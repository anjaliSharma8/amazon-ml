```python
import re
import unicodedata

def normalize_text(text):
    if pd.isna(text):
        return ""

    text = str(text).lower()

    # Unicode normalization
    text = unicodedata.normalize("NFKC", text)

    # Keep letters, marks, numbers and spaces
    result = []

    for ch in text:
        category = unicodedata.category(ch)

        if category[0] in {"L", "M", "N"} or ch.isspace():
            result.append(ch)
        else:
            result.append(" ")

    text = "".join(result)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
# Read only 100,000 rows from S2 and S3
s2_sample = pd.read_csv(
    "train_source2.tsv",
    sep="\t",
    nrows=100000
)

s3_sample = pd.read_csv(
    "train_source3.tsv",
    sep="\t",
    nrows=100000
)

# Normalize name and address
s2_sample["name_norm"] = s2_sample["business_name"].apply(normalize_text)
s2_sample["address_norm"] = s2_sample["business_address"].apply(normalize_text)

s3_sample["name_norm"] = s3_sample["business_name"].apply(normalize_text)
s3_sample["address_norm"] = s3_sample["business_address"].apply(normalize_text)

print("S2 sample:", s2_sample.shape)
print("S3 sample:", s3_sample.shape)

print("\nS2 unique normalized names:")
print(s2_sample["name_norm"].nunique())

print("\nS3 unique normalized names:")
print(s3_sample["name_norm"].nunique())

print("\nSample S2:")
display(
    s2_sample[
        ["entity_id", "business_name", "name_norm",
         "business_address", "address_norm", "country"]
    ].head(10)
)
```

```python
s1_id = first_gt["source1_entity_id"]

s1_record = train_s1[train_s1["entity_id"] == s1_id]

display(s1_record)
```

```python
# Get the correct matching IDs
matched_ids = first_gt["matched_entity_ids"].split(",")

s2_ids = {x for x in matched_ids if x.startswith("S2-")}
s3_ids = {x for x in matched_ids if x.startswith("S3-")}

print("S2 IDs:", s2_ids)
print("S3 IDs:", s3_ids)

found_s2 = []
found_s3 = []

# Search S2 in chunks
for chunk in pd.read_csv(
    "train_source2.tsv",
    sep="\t",
    chunksize=100000,
    dtype=str
):
    rows = chunk[chunk["entity_id"].isin(s2_ids)]
    if len(rows) > 0:
        found_s2.append(rows)

    if sum(len(x) for x in found_s2) == len(s2_ids):
        break

# Search S3 in chunks
for chunk in pd.read_csv(
    "train_source3.tsv",
    sep="\t",
    chunksize=100000,
    dtype=str
):
    rows = chunk[chunk["entity_id"].isin(s3_ids)]
    if len(rows) > 0:
        found_s3.append(rows)

    if sum(len(x) for x in found_s3) == len(s3_ids):
        break

matched_s2 = pd.concat(found_s2, ignore_index=True)
matched_s3 = pd.concat(found_s3, ignore_index=True)

print("MATCHED S2 RECORDS")
display(matched_s2)

print("MATCHED S3 RECORDS")
display(matched_s3)
```

```python
# Positive match IDs from ground truth
positive_ids = set(matched_s2["entity_id"]) | set(matched_s3["entity_id"])

# S1 information
s1_row = s1_record.iloc[0]

# -------------------------
# Positive pairs -> label 1
# -------------------------

positive_s2 = matched_s2.copy()
positive_s3 = matched_s3.copy()

positive_s2["s1_id"] = s1_row["entity_id"]
positive_s2["label"] = 1

positive_s3["s1_id"] = s1_row["entity_id"]
positive_s3["label"] = 1

# -------------------------
# Negative pairs -> label 0
# Take records from same country
# but which are NOT true matches
# -------------------------

negative_s2 = s2_sample[
    (s2_sample["country"] == s1_row["country"]) &
    (~s2_sample["entity_id"].isin(positive_ids))
].head(5).copy()

negative_s3 = s3_sample[
    (s3_sample["country"] == s1_row["country"]) &
    (~s3_sample["entity_id"].isin(positive_ids))
].head(5).copy()

negative_s2["s1_id"] = s1_row["entity_id"]
negative_s2["label"] = 0

negative_s3["s1_id"] = s1_row["entity_id"]
negative_s3["label"] = 0

# Combine everything
training_pairs = pd.concat(
    [positive_s2, positive_s3, negative_s2, negative_s3],
    ignore_index=True
)

print("Training pairs:")
display(
    training_pairs[
        ["s1_id", "entity_id", "business_name",
         "business_address", "country", "label"]
    ]
)
```

```python
import re
import unicodedata
from difflib import SequenceMatcher

def normalize_text(text):
    if pd.isna(text):
        return ""

    text = str(text).lower()
    text = unicodedata.normalize("NFKC", text)

    result = []

    for ch in text:
        category = unicodedata.category(ch)

        if category[0] in {"L", "M", "N"} or ch.isspace():
            result.append(ch)
        else:
            result.append(" ")

    text = "".join(result)
    text = re.sub(r"\s+", " ", text).strip()

    return text


# Recreate normalized columns
train_s1["name_norm"] = train_s1["business_name"].apply(normalize_text)
train_s1["address_norm"] = train_s1["business_address"].apply(normalize_text)

# Recreate our S1 record
s1_id = "S1-965667"
s1_record = train_s1[train_s1["entity_id"] == s1_id]

print("S1 record restored ✅")
display(s1_record)
```

```python
def similarity(a, b):
    return SequenceMatcher(None, str(a), str(b)).ratio()


s1_name = s1_record.iloc[0]["name_norm"]
s1_address = s1_record.iloc[0]["address_norm"]
s1_country = s1_record.iloc[0]["country"]

training_pairs["name_similarity"] = training_pairs["business_name"].apply(
    lambda x: similarity(s1_name, normalize_text(x))
)

training_pairs["address_similarity"] = training_pairs["business_address"].apply(
    lambda x: similarity(s1_address, normalize_text(x))
)

training_pairs["country_same"] = (
    training_pairs["country"] == s1_country
).astype(int)

display(
    training_pairs[
        [
            "s1_id",
            "entity_id",
            "label",
            "name_similarity",
            "address_similarity",
            "country_same"
        ]
    ]
)
```

```python
import re
from collections import defaultdict

def name_block_key(name):
    name = normalize_text(name)
    # spaces hatao
    name = re.sub(r"\s+", "", name)
    # first 12 characters
    return name[:12]

def address_number(address):
    if pd.isna(address):
        return ""
    
    nums = re.findall(r"\d+", str(address))
    
    if nums:
        return nums[0]
    
    return ""


# Create blocking keys
s2_sample["name_block"] = s2_sample["business_name"].apply(name_block_key)
s3_sample["name_block"] = s3_sample["business_name"].apply(name_block_key)

s2_sample["address_number"] = s2_sample["business_address"].apply(address_number)
s3_sample["address_number"] = s3_sample["business_address"].apply(address_number)

print("Blocking keys created ✅")
```

```python
# Name index
name_index = defaultdict(list)

for entity_id, country, key in zip(
    s2_sample["entity_id"],
    s2_sample["country"],
    s2_sample["name_block"]
):
    if key:
        name_index[(country, key)].append(entity_id)

for entity_id, country, key in zip(
    s3_sample["entity_id"],
    s3_sample["country"],
    s3_sample["name_block"]
):
    if key:
        name_index[(country, key)].append(entity_id)


# Address-number index
address_index = defaultdict(list)

for entity_id, country, key in zip(
    s2_sample["entity_id"],
    s2_sample["country"],
    s2_sample["address_number"]
):
    if key:
        address_index[(country, key)].append(entity_id)

for entity_id, country, key in zip(
    s3_sample["entity_id"],
    s3_sample["country"],
    s3_sample["address_number"]
):
    if key:
        address_index[(country, key)].append(entity_id)

print("Indexes created ✅")
```

```python
# Our example S1
row = s1_record.iloc[0]

country = row["country"]
name_key = name_block_key(row["business_name"])
number_key = address_number(row["business_address"])

candidates = set()

# Block 1: name
candidates.update(
    name_index.get((country, name_key), [])
)

# Block 2: address number
candidates.update(
    address_index.get((country, number_key), [])
)

print("S1:", row["entity_id"])
print("Name block:", name_key)
print("Address number:", number_key)

print("\nNumber of candidates:", len(candidates))
print("\nCandidates:")
print(candidates)
```

```python
true_matches = set(matched_ids)

found = true_matches.intersection(candidates)
missed = true_matches - candidates

print("True matches:", true_matches)
print("\nFound by blocking:", found)
print("\nMissed by blocking:", missed)

print("\nBlocking recall:",
      len(found) / len(true_matches))
```

```python
def get_name_tokens(name):
    name = normalize_text(name)
    tokens = name.split()

    # Very short/common tokens ignore
    tokens = [t for t in tokens if len(t) >= 3]

    return tokens


# Create token-prefix index for S2 and S3
token_index = defaultdict(list)

for entity_id, country, name in zip(
    s2_sample["entity_id"],
    s2_sample["country"],
    s2_sample["business_name"]
):
    tokens = get_name_tokens(name)

    for token in set(tokens):
        key = (country, token[:5])
        token_index[key].append(entity_id)


for entity_id, country, name in zip(
    s3_sample["entity_id"],
    s3_sample["country"],
    s3_sample["business_name"]
):
    tokens = get_name_tokens(name)

    for token in set(tokens):
        key = (country, token[:5])
        token_index[key].append(entity_id)


# Get candidates for our S1
row = s1_record.iloc[0]

s1_tokens = get_name_tokens(row["business_name"])

token_candidates = set()

for token in s1_tokens:
    key = (row["country"], token[:5])
    token_candidates.update(token_index.get(key, []))


print("S1:", row["entity_id"])
print("S1 tokens:", s1_tokens)
print("Number of token-block candidates:", len(token_candidates))

print("\nTrue matches found:")
print(true_matches.intersection(token_candidates))

print("\nMissed true matches:")
print(true_matches - token_candidates)

print("\nToken blocking recall:")
print(
    len(true_matches.intersection(token_candidates))
    / len(true_matches)
)
```

```python
target_ids = true_matches

found_records = []

for file in ["train_source2.tsv", "train_source3.tsv"]:

    print("Searching:", file)

    for chunk in pd.read_csv(
        file,
        sep="\t",
        chunksize=100000,
        dtype=str
    ):
        found = chunk[chunk["entity_id"].isin(target_ids)]

        if not found.empty:
            found_records.append(found)

        # If all 5 are found, we can stop
        found_ids = set(
            pd.concat(found_records)["entity_id"]
        ) if found_records else set()

        if target_ids.issubset(found_ids):
            break

all_true_records = pd.concat(found_records, ignore_index=True)

display(
    all_true_records[
        ["entity_id", "business_name", "business_address", "country"]
    ]
)
```

```python
from rapidfuzz import fuzz
import re

s1 = s1_record.iloc[0]

def get_numbers(text):
    if pd.isna(text):
        return set()
    return set(re.findall(r"\d+", str(text)))


s1_name = normalize_text(s1["business_name"])
s1_address = normalize_text(s1["business_address"])
s1_numbers = get_numbers(s1["business_address"])

results = []

for _, row in all_true_records.iterrows():

    candidate_name = normalize_text(row["business_name"])
    candidate_address = normalize_text(row["business_address"])

    name_score = fuzz.token_set_ratio(
        s1_name,
        candidate_name
    ) / 100

    address_score = fuzz.token_set_ratio(
        s1_address,
        candidate_address
    ) / 100

    candidate_numbers = get_numbers(row["business_address"])

    if len(s1_numbers | candidate_numbers) == 0:
        number_overlap = 0
    else:
        number_overlap = len(s1_numbers & candidate_numbers) / len(
            s1_numbers | candidate_numbers
        )

    country_same = int(
        s1["country"] == row["country"]
    )

    results.append({
        "entity_id": row["entity_id"],
        "business_name": row["business_name"],
        "name_score": round(name_score, 3),
        "address_score": round(address_score, 3),
        "number_overlap": round(number_overlap, 3),
        "country_same": country_same
    })

signal_table = pd.DataFrame(results)

display(signal_table)
```

```python
rule_candidates = signal_table[
    (signal_table["name_score"] >= 0.80) |
    (signal_table["address_score"] >= 0.70) |
    (signal_table["number_overlap"] > 0)
]

print("Candidates captured by rule:")
display(rule_candidates)

print("True matches captured:",
      len(rule_candidates))

print("Total true matches:",
      len(signal_table))

print("Recall:",
      len(rule_candidates) / len(signal_table))
```

```python
# Apply our improved blocking rule to the sample

s2_name_score = s2_sample["business_name"].apply(
    lambda x: fuzz.token_set_ratio(
        s1_name,
        normalize_text(x)
    ) / 100
)

s2_address_score = s2_sample["business_address"].apply(
    lambda x: fuzz.token_set_ratio(
        s1_address,
        normalize_text(x)
    ) / 100
)

s3_name_score = s3_sample["business_name"].apply(
    lambda x: fuzz.token_set_ratio(
        s1_name,
        normalize_text(x)
    ) / 100
)

s3_address_score = s3_sample["business_address"].apply(
    lambda x: fuzz.token_set_ratio(
        s1_address,
        normalize_text(x)
    ) / 100
)


# IMPORTANT: each comparison has its own brackets
s2_candidate_mask = (
    (s2_sample["country"] == s1["country"]) &
    (
        (s2_name_score >= 0.80) |
        (s2_address_score >= 0.70)
    )
)

s3_candidate_mask = (
    (s3_sample["country"] == s1["country"]) &
    (
        (s3_name_score >= 0.80) |
        (s3_address_score >= 0.70)
    )
)


sample_candidates = set(
    s2_sample.loc[s2_candidate_mask, "entity_id"]
).union(
    set(s3_sample.loc[s3_candidate_mask, "entity_id"])
)

print("Number of candidates:", len(sample_candidates))

print("\nTrue matches found:")
print(true_matches.intersection(sample_candidates))

print("\nNumber of true matches found:",
      len(true_matches.intersection(sample_candidates)))
```

```python
def token_prefixes(name):
    tokens = normalize_text(name).split()
    return [t[:5] for t in tokens if len(t) >= 3]


s1_tokens = token_prefixes(s1["business_name"])

print("S1:", s1["business_name"])
print("Token prefixes:", s1_tokens)

print("\nTrue-match token prefixes:")

for _, row in all_true_records.iterrows():
    print(row["entity_id"], "->", token_prefixes(row["business_name"]))
```

```python
from collections import Counter

token_counts = Counter()

for name in s2_sample["business_name"]:
    for token in set(token_prefixes(name)):
        token_counts[token] += 1

for name in s3_sample["business_name"]:
    for token in set(token_prefixes(name)):
        token_counts[token] += 1

for token in s1_tokens:
    print(token, "->", token_counts[token], "records")
```

```python
# Keep only informative name tokens
MAX_TOKEN_FREQ = 1000

# Build a filtered token index
filtered_token_index = defaultdict(list)

for entity_id, country, name in zip(
    s2_sample["entity_id"],
    s2_sample["country"],
    s2_sample["business_name"]
):
    for token in set(token_prefixes(name)):
        if token_counts[token] <= MAX_TOKEN_FREQ:
            filtered_token_index[(country, token)].append(entity_id)

for entity_id, country, name in zip(
    s3_sample["entity_id"],
    s3_sample["country"],
    s3_sample["business_name"]
):
    for token in set(token_prefixes(name)):
        if token_counts[token] <= MAX_TOKEN_FREQ:
            filtered_token_index[(country, token)].append(entity_id)


# Generate candidates for our S1
rare_token_candidates = set()

for token in s1_tokens:
    if token_counts[token] <= MAX_TOKEN_FREQ:
        rare_token_candidates.update(
            filtered_token_index.get(
                (s1["country"], token),
                []
            )
        )

print("S1:", s1["entity_id"])
print("Useful tokens:", [
    t for t in s1_tokens
    if token_counts[t] <= MAX_TOKEN_FREQ
])

print("\nCandidates from rare-token blocking:",
      len(rare_token_candidates))

print("\nTrue matches found:")
print(true_matches.intersection(rare_token_candidates))

print("\nTrue matches found:",
      len(true_matches.intersection(rare_token_candidates)),
      "out of", len(true_matches))
```

```python
def four_char_tokens(name):
    tokens = normalize_text(name).split()
    return {t[:4] for t in tokens if len(t) >= 3}


s1_tokens4 = four_char_tokens(s1["business_name"])
s1_numbers = get_numbers(s1["business_address"])

print("S1 4-char tokens:", s1_tokens4)
print("S1 address numbers:", s1_numbers)

print("\nChecking 5 true matches:\n")

for _, row in all_true_records.iterrows():

    candidate_tokens = four_char_tokens(row["business_name"])
    candidate_numbers = get_numbers(row["business_address"])

    token_hit = bool(s1_tokens4 & candidate_tokens)
    number_hit = bool(s1_numbers & candidate_numbers)

    captured = token_hit or number_hit

    print(
        row["entity_id"],
        "| name hit:", token_hit,
        "| number hit:", number_hit,
        "| captured:", captured
    )
```

```python
import re
import pandas as pd

# Our S1
row = s1_record.iloc[0]

country = row["country"]

# From our earlier experiment:
name_prefixes = {"maur", "will", "colo"}
address_number = "85"

# Regex for any useful 4-character name prefix
name_pattern = r"(?<!\w)(?:maur|will|colo)"

# Regex for address number 85
number_pattern = r"(?<!\d)85(?!\d)"

all_candidates_full = set()

for filename in ["train_source2.tsv", "train_source3.tsv"]:

    print("\nScanning:", filename)

    rows_found = 0

    for chunk in pd.read_csv(
        filename,
        sep="\t",
        chunksize=100000,
        dtype=str,
        usecols=[
            "entity_id",
            "business_name",
            "business_address",
            "country"
        ]
    ):

        # Only same-country records
        chunk = chunk[
            chunk["country"].eq(country)
        ].copy()

        if chunk.empty:
            continue

        # Lightweight normalization for this US example
        name = (
            chunk["business_name"]
            .fillna("")
            .str.lower()
            .str.replace(r"[^\w\s]", " ", regex=True)
            .str.replace(r"\s+", " ", regex=True)
            .str.strip()
        )

        address = (
            chunk["business_address"]
            .fillna("")
            .str.lower()
            .str.replace(r"[^\w\s]", " ", regex=True)
            .str.replace(r"\s+", " ", regex=True)
            .str.strip()
        )

        # Blocking condition
        name_hit = name.str.contains(
            name_pattern,
            regex=True,
            na=False
        )

        number_hit = address.str.contains(
            number_pattern,
            regex=True,
            na=False
        )

        candidates = chunk[name_hit | number_hit]

        all_candidates_full.update(
            candidates["entity_id"].tolist()
        )

        rows_found += len(candidates)

        if rows_found % 10000 == 0 and rows_found > 0:
            print("Candidates found so far:", rows_found)

print("\n==============================")
print("FULL DATA BLOCKING RESULT")
print("==============================")

print("Total candidates:", len(all_candidates_full))

print("\nTrue matches found:")
print(true_matches.intersection(all_candidates_full))

print("\nTrue matches missed:")
print(true_matches - all_candidates_full)

print(
    "\nTrue matches captured:",
    len(true_matches.intersection(all_candidates_full)),
    "out of",
    len(true_matches)
)
```

```python
# SMARTER BLOCKING TEST
# We are still testing only S1-965667

s1 = s1_record.iloc[0]

all_smart_candidates = set()

for filename in ["train_source2.tsv", "train_source3.tsv"]:

    print("Scanning:", filename)

    for chunk in pd.read_csv(
        filename,
        sep="\t",
        chunksize=100000,
        dtype=str,
        usecols=[
            "entity_id",
            "business_name",
            "business_address",
            "country"
        ]
    ):

        # Same country only
        chunk = chunk[chunk["country"] == s1["country"]].copy()

        if chunk.empty:
            continue

        # Normalize
        names = (
            chunk["business_name"]
            .fillna("")
            .str.lower()
            .str.replace(r"[^\w\s]", " ", regex=True)
            .str.replace(r"\s+", " ", regex=True)
            .str.strip()
        )

        addresses = (
            chunk["business_address"]
            .fillna("")
            .str.lower()
            .str.replace(r"[^\w\s]", " ", regex=True)
            .str.replace(r"\s+", " ", regex=True)
            .str.strip()
        )

        # --------------------------------
        # NAME BLOCK
        # --------------------------------

        maur_hit = names.str.contains(
            r"(?<!\w)maur",
            regex=True,
            na=False
        )

        will_hit = names.str.contains(
            r"(?<!\w)will",
            regex=True,
            na=False
        )

        colo_hit = names.str.contains(
            r"(?<!\w)colo",
            regex=True,
            na=False
        )

        # Count how many useful name tokens matched
        name_hit_count = (
            maur_hit.astype(int)
            + will_hit.astype(int)
            + colo_hit.astype(int)
        )

        two_name_hits = name_hit_count >= 2

        # --------------------------------
        # ADDRESS BLOCK
        # --------------------------------

        number_hit = addresses.str.contains(
            r"(?<!\d)85(?!\d)",
            regex=True,
            na=False
        )

        address_token_hit = addresses.str.contains(
            r"(?<!\w)(?:wayn|tico)",
            regex=True,
            na=False
        )

        # --------------------------------
        # FINAL BLOCKING RULE
        # --------------------------------

        mask = (
            two_name_hits
            | number_hit
            | address_token_hit
        )

        candidates = chunk.loc[mask, "entity_id"]

        all_smart_candidates.update(candidates.tolist())


print("\n==============================")
print("SMART BLOCKING RESULT")
print("==============================")

print("Total candidates:", len(all_smart_candidates))

found = true_matches.intersection(all_smart_candidates)
missed = true_matches - all_smart_candidates

print("\nTrue matches found:")
print(found)

print("\nTrue matches missed:")
print(missed)

print(
    "\nTrue matches captured:",
    len(found),
    "out of",
    len(true_matches)
)

print(
    "\nRecall:",
    len(found) / len(true_matches)
)
```

```python
from itertools import combinations
import re

# Common legal suffixes:
# These are not removed from the original data.
# They are only avoided as primary blocking keys.
LEGAL_SUFFIXES = {
    "inc", "incorporated",
    "ltd", "limited",
    "llc",
    "corp", "corporation",
    "co", "company",
    "pvt", "private",
    "public", "plc"
}


def get_name_prefixes(name):
    name = normalize_text(name)
    tokens = name.split()

    prefixes = set()

    for token in tokens:
        if len(token) >= 4 and token not in LEGAL_SUFFIXES:
            prefixes.add(token[:4])

    return sorted(prefixes)


def get_address_prefixes(address):
    address = normalize_text(address)
    tokens = address.split()

    prefixes = set()

    for token in tokens:
        if len(token) >= 4:
            prefixes.add(token[:4])

    return sorted(prefixes)


def get_numbers(address):
    if pd.isna(address):
        return []

    return re.findall(r"\d+", str(address))


# Create automatic blocking keys for every S1
sample_100_s1["name_prefixes"] = sample_100_s1["business_name"].apply(
    get_name_prefixes
)

sample_100_s1["name_pairs"] = sample_100_s1["name_prefixes"].apply(
    lambda x: list(combinations(x, 2))
)

sample_100_s1["address_prefixes"] = sample_100_s1["business_address"].apply(
    get_address_prefixes
)

sample_100_s1["address_numbers"] = sample_100_s1["business_address"].apply(
    get_numbers
)


display(
    sample_100_s1[
        [
            "entity_id",
            "business_name",
            "name_prefixes",
            "name_pairs",
            "address_numbers",
            "address_prefixes"
        ]
    ].head(10)
)
```

```python
LEGAL_SUFFIXES = {
    "inc", "incorporated",
    "ltd", "limited",
    "llc",
    "pllc",
    "corp", "corporation",
    "co", "company",
    "pvt", "private",
    "public", "plc"
}

# Recreate the name prefixes after the correction
sample_100_s1["name_prefixes"] = sample_100_s1["business_name"].apply(
    get_name_prefixes
)

sample_100_s1["name_pairs"] = sample_100_s1["name_prefixes"].apply(
    lambda x: list(combinations(x, 2))
)

display(
    sample_100_s1[
        ["entity_id", "business_name", "name_prefixes", "name_pairs"]
    ].head(10)
)
```

```python
from collections import defaultdict
from itertools import combinations

# -----------------------------------------
# 1. Build lookup maps from our 100 S1s
# -----------------------------------------

name_pair_to_s1 = defaultdict(set)
address_pair_to_s1 = defaultdict(set)

# Also keep exact normalized names as an additional block
exact_name_to_s1 = defaultdict(set)

for _, row in sample_100_s1.iterrows():

    s1_id = row["entity_id"]
    country = row["country"]

    # ----- Name pair keys -----
    prefixes = set(row["name_prefixes"])

    for pair in combinations(sorted(prefixes), 2):
        name_pair_to_s1[(country, pair)].add(s1_id)

    # ----- Exact normalized name -----
    exact_name = normalize_text(row["business_name"])

    if exact_name:
        exact_name_to_s1[(country, exact_name)].add(s1_id)

    # ----- Address pair keys -----
    numbers = set(row["address_numbers"])
    addr_prefixes = set(row["address_prefixes"])

    for number in numbers:
        for prefix in addr_prefixes:
            address_pair_to_s1[(country, number, prefix)].add(s1_id)


print("Blocking lookup created ✅")
print("Name-pair keys:", len(name_pair_to_s1))
print("Exact-name keys:", len(exact_name_to_s1))
print("Address keys:", len(address_pair_to_s1))
```

```python
# Create small lookup tables from our 100 S1 records

name_lookup_rows = []
exact_lookup_rows = []
address_lookup_rows = []

for _, row in sample_100_s1.iterrows():

    s1_id = row["entity_id"]
    country = row["country"]

    # -------------------------
    # Exact normalized name
    # -------------------------
    exact_name = normalize_text(row["business_name"])

    if exact_name:
        exact_lookup_rows.append({
            "s1_id": s1_id,
            "country": country,
            "exact_name": exact_name
        })

    # -------------------------
    # Name prefix keys
    # -------------------------
    for prefix in set(row["name_prefixes"]):
        name_lookup_rows.append({
            "s1_id": s1_id,
            "country": country,
            "prefix": prefix
        })

    # -------------------------
    # Address number + prefix
    # -------------------------
    numbers = set(row["address_numbers"])
    prefixes = set(row["address_prefixes"])

    for number in numbers:
        for prefix in prefixes:
            address_lookup_rows.append({
                "s1_id": s1_id,
                "country": country,
                "number": str(number),
                "prefix": prefix
            })


name_lookup_df = pd.DataFrame(name_lookup_rows)
exact_lookup_df = pd.DataFrame(exact_lookup_rows)
address_lookup_df = pd.DataFrame(address_lookup_rows)

con.register("name_lookup", name_lookup_df)
con.register("exact_lookup", exact_lookup_df)
con.register("address_lookup", address_lookup_df)

print("Lookup tables ready ✅")
print("Name keys:", len(name_lookup_df))
print("Exact-name keys:", len(exact_lookup_df))
print("Address keys:", len(address_lookup_df))
```

```python
con.execute("""
PRAGMA threads=4;

CREATE OR REPLACE TEMP TABLE blocking_candidates AS

WITH source_data AS (

    SELECT
        entity_id,
        country,
        lower(
            regexp_replace(
                coalesce(business_name, ''),
                '[^\\p{L}\\p{N}]+',
                ' ',
                'g'
            )
        ) AS name_clean,
        lower(
            regexp_replace(
                coalesce(business_address, ''),
                '[^\\p{L}\\p{N}]+',
                ' ',
                'g'
            )
        ) AS address_clean

    FROM read_csv(
        'train_source2.tsv',
        delim='\\t',
        header=true,
        auto_detect=true
    )

    UNION ALL

    SELECT
        entity_id,
        country,
        lower(
            regexp_replace(
                coalesce(business_name, ''),
                '[^\\p{L}\\p{N}]+',
                ' ',
                'g'
            )
        ) AS name_clean,
        lower(
            regexp_replace(
                coalesce(business_address, ''),
                '[^\\p{L}\\p{N}]+',
                ' ',
                'g'
            )
        ) AS address_clean

    FROM read_csv(
        'train_source3.tsv',
        delim='\\t',
        header=true,
        auto_detect=true
    )
),

-- -------------------------
-- Exact normalized name
-- -------------------------

exact_matches AS (

    SELECT DISTINCT
        e.s1_id,
        s.entity_id AS candidate_id

    FROM source_data s

    JOIN exact_lookup e
      ON s.country = e.country
     AND s.name_clean = e.exact_name

    WHERE s.name_clean <> ''
),

-- -------------------------
-- Name token prefixes
-- -------------------------

source_name_tokens AS (

    SELECT
        s.entity_id,
        s.country,
        substr(trim(token), 1, 4) AS prefix

    FROM source_data s,
    UNNEST(string_split(s.name_clean, ' ')) AS u(token)

    WHERE length(trim(token)) >= 3
      AND lower(trim(token)) NOT IN (
          'inc', 'incorporated',
          'ltd', 'limited',
          'llc', 'pllc',
          'corp', 'corporation',
          'co', 'company',
          'pvt', 'private',
          'public', 'plc'
      )
),

name_matches AS (

    SELECT
        l.s1_id,
        t.entity_id AS candidate_id

    FROM source_name_tokens t

    JOIN name_lookup l
      ON t.country = l.country
     AND t.prefix = l.prefix

    GROUP BY
        l.s1_id,
        t.entity_id

    HAVING COUNT(DISTINCT t.prefix) >= 2
),

-- -------------------------
-- Address numbers
-- -------------------------

source_address_numbers AS (

    SELECT
        s.entity_id,
        s.country,
        number

    FROM source_data s,
    UNNEST(
        regexp_extract_all(s.address_clean, '\\d+')
    ) AS u(number)

    WHERE number <> ''
),

-- -------------------------
-- Address token prefixes
-- -------------------------

source_address_tokens AS (

    SELECT
        s.entity_id,
        s.country,
        substr(trim(token), 1, 4) AS prefix

    FROM source_data s,
    UNNEST(string_split(s.address_clean, ' ')) AS u(token)

    WHERE length(trim(token)) >= 4
),

address_matches AS (

    SELECT DISTINCT
        l.s1_id,
        n.entity_id AS candidate_id

    FROM source_address_numbers n

    JOIN source_address_tokens a
      ON n.entity_id = a.entity_id
     AND n.country = a.country

    JOIN address_lookup l
      ON n.country = l.country
     AND n.number = l.number
     AND a.prefix = l.prefix
)

-- -------------------------
-- UNION all blocking passes
-- -------------------------

SELECT DISTINCT s1_id, candidate_id
FROM exact_matches

UNION

SELECT DISTINCT s1_id, candidate_id
FROM name_matches

UNION

SELECT DISTINCT s1_id, candidate_id
FROM address_matches
""").fetchdf()

print("✅ Blocking candidates created")
```

```python
# -----------------------------------------
# Evaluate DuckDB blocking candidates
# -----------------------------------------

candidate_df = con.execute("""
    SELECT s1_id, candidate_id
    FROM blocking_candidates
""").fetchdf()

print("Total candidate pairs:", len(candidate_df))

# Convert candidates into:
# S1 -> set of candidate IDs
candidates_by_s1 = defaultdict(set)

for s1_id, candidate_id in zip(
    candidate_df["s1_id"],
    candidate_df["candidate_id"]
):
    candidates_by_s1[s1_id].add(candidate_id)


# -----------------------------------------
# Compare with Ground Truth
# -----------------------------------------

evaluation = []

for _, row in gt_100.iterrows():

    s1_id = row["source1_entity_id"]

    if row["matched_entity_ids"] == "":
        true_matches = set()
    else:
        true_matches = set(
            row["matched_entity_ids"].split(",")
        )

    candidates = candidates_by_s1.get(
        s1_id,
        set()
    )

    found = true_matches.intersection(candidates)

    missed = true_matches - candidates

    recall = (
        len(found) / len(true_matches)
        if len(true_matches) > 0
        else None
    )

    evaluation.append({
        "s1_id": s1_id,
        "true_matches": len(true_matches),
        "candidates": len(candidates),
        "found": len(found),
        "missed": len(missed),
        "recall": recall
    })


blocking_eval = pd.DataFrame(evaluation)

# -----------------------------------------
# Summary
# -----------------------------------------

non_singletons = blocking_eval[
    blocking_eval["true_matches"] > 0
]

singletons = blocking_eval[
    blocking_eval["true_matches"] == 0
]

print("\n========== BLOCKING RESULTS ==========")

print("S1 tested:", len(blocking_eval))

print(
    "Overall blocking recall:",
    round(
        non_singletons["found"].sum()
        / non_singletons["true_matches"].sum(),
        4
    )
)

print(
    "Average candidates per S1:",
    round(
        blocking_eval["candidates"].mean(),
        2
    )
)

print(
    "Median candidates per S1:",
    round(
        blocking_eval["candidates"].median(),
        2
    )
)

print(
    "Maximum candidates:",
    blocking_eval["candidates"].max()
)

print(
    "\nSingleton S1s:",
    len(singletons)
)

print(
    "Singletons with candidates:",
    (singletons["candidates"] > 0).sum()
)

print("\nWorst 10 S1s:")
display(
    blocking_eval
    .sort_values(
        ["recall", "candidates"],
        na_position="first"
    )
    .head(10)
)
```

```python
# Find non-singleton S1s where our blocker missed true matches

missed_rows = []

for _, row in gt_100.iterrows():

    s1_id = row["source1_entity_id"]

    if row["matched_entity_ids"] == "":
        continue

    true_ids = set(row["matched_entity_ids"].split(","))

    candidate_ids = candidates_by_s1.get(s1_id, set())

    missed = true_ids - candidate_ids

    if missed:
        missed_rows.append({
            "s1_id": s1_id,
            "true_matches": len(true_ids),
            "found": len(true_ids & candidate_ids),
            "missed": len(missed),
            "missed_ids": ",".join(sorted(missed))
        })

missed_df = pd.DataFrame(missed_rows)

print("S1s with missed true matches:", len(missed_df))

display(
    missed_df.sort_values(
        "missed",
        ascending=False
    ).head(10)
)
```

```python
# ------------------------------------------------
# Get the actual records for missed true matches
# ------------------------------------------------

# Collect all missed S2/S3 IDs
missed_ids = set()

for ids in missed_df["missed_ids"]:
    if ids:
        missed_ids.update(ids.split(","))

# Get the S1 records
missed_s1_ids = set(missed_df["s1_id"])

missed_s1_records = train_s1[
    train_s1["entity_id"].isin(missed_s1_ids)
].copy()

# Register missed target IDs with DuckDB
missed_id_df = pd.DataFrame({
    "entity_id": list(missed_ids)
})

con.register("missed_id_lookup", missed_id_df)

# Get missed S2 records
missed_s2 = con.execute("""
    SELECT
        s.entity_id,
        s.business_name,
        s.business_address,
        s.country
    FROM read_csv(
        'train_source2.tsv',
        delim='\t',
        header=true
    ) s
    INNER JOIN missed_id_lookup m
        ON s.entity_id = m.entity_id
""").fetchdf()

# Get missed S3 records
missed_s3 = con.execute("""
    SELECT
        s.entity_id,
        s.business_name,
        s.business_address,
        s.country
    FROM read_csv(
        'train_source3.tsv',
        delim='\t',
        header=true
    ) s
    INNER JOIN missed_id_lookup m
        ON s.entity_id = m.entity_id
""").fetchdf()

print("Missed S2 records:", len(missed_s2))
print("Missed S3 records:", len(missed_s3))

print("\nMISSED S2 RECORDS")
display(missed_s2)

print("\nMISSED S3 RECORDS")
display(missed_s3)
```

```python
# ==========================================
# IMPROVED BLOCKING V2
# ==========================================

# New candidates will be added to existing
# fast_candidates / blocking_candidates

improved_candidates = defaultdict(set)

# ------------------------------------------
# Build lookup tables for 100 S1
# ------------------------------------------

compact_name_lookup = defaultdict(set)
short_token_lookup = defaultdict(set)
address_token_pair_lookup = defaultdict(set)
address_number_lookup = defaultdict(set)

for _, row in sample_100_s1.iterrows():

    s1_id = row["entity_id"]
    country = row["country"]

    # ======================================
    # 1. COMPACT NAME PREFIX
    # ======================================
    name = normalize_text(row["business_name"])

    # remove spaces -> useful for:
    # ABC Corporation -> abccorporation
    # abc.com -> abccom
    compact = re.sub(r"[^a-z0-9]", "", name.lower())

    if len(compact) >= 6:
        key = compact[:8]
        compact_name_lookup[(country, key)].add(s1_id)

    # ======================================
    # 2. SHORT / FIRST TOKEN BLOCK
    # ======================================
    tokens = name.split()

    for token in tokens:

        # normal 3+ character informative tokens
        if len(token) >= 3 and token not in LEGAL_SUFFIXES:

            short_key = token[:3]

            short_token_lookup[
                (country, short_key)
            ].add(s1_id)

        # 2-character tokens
        # only keep if not a legal/common word
        elif len(token) == 2 and token.isalpha():

            short_token_lookup[
                (country, token)
            ].add(s1_id)

    # ======================================
    # 3. ADDRESS TOKEN PAIRS
    # ======================================
    addr = normalize_text(row["business_address"])

    addr_tokens = [
        t[:4]
        for t in addr.split()
        if len(t) >= 4
    ]

    addr_tokens = sorted(set(addr_tokens))

    for pair in combinations(addr_tokens, 2):

        address_token_pair_lookup[
            (country, pair)
        ].add(s1_id)

    # ======================================
    # 4. NORMALIZED ADDRESS NUMBERS
    # ======================================
    numbers = re.findall(r"\d+", addr)

    for number in numbers:

        # remove leading zeros
        normalized_number = number.lstrip("0")

        if normalized_number == "":
            normalized_number = "0"

        address_number_lookup[
            (country, normalized_number)
        ].add(s1_id)


print("New blocking lookups ready ✅")

print("Compact-name keys:",
      len(compact_name_lookup))

print("Short-token keys:",
      len(short_token_lookup))

print("Address-token-pair keys:",
      len(address_token_pair_lookup))

print("Address-number keys:",
      len(address_number_lookup))
```

```python
# ==========================================
# CHECK FREQUENCY OF OUR CURRENT BLOCKING KEYS
# ==========================================

# -----------------------------
# 1. Short-name keys
# -----------------------------

short_key_rows = []

for (country, key), s1_ids in short_token_lookup.items():
    short_key_rows.append({
        "country": country,
        "key": key
    })

short_key_df = pd.DataFrame(short_key_rows).drop_duplicates()


# -----------------------------
# 2. Address token-pair keys
# -----------------------------

address_key_rows = []

for (country, pair), s1_ids in address_token_pair_lookup.items():

    address_key_rows.append({
        "country": country,
        "prefix1": pair[0],
        "prefix2": pair[1]
    })

address_key_df = pd.DataFrame(address_key_rows).drop_duplicates()


con.register("short_key_lookup", short_key_df)
con.register("address_key_lookup", address_key_df)

print("Short keys:", len(short_key_df))
print("Address-pair keys:", len(address_key_df))


# ==========================================
# SHORT NAME TOKEN FREQUENCY
# ==========================================

short_freq = con.execute("""
WITH source_data AS (

    SELECT
        entity_id,
        country,
        lower(
            regexp_replace(
                coalesce(business_name, ''),
                '[^\\p{L}\\p{N}]+',
                ' ',
                'g'
            )
        ) AS name_clean

    FROM read_csv(
        'train_source2.tsv',
        delim='\\t',
        header=true
    )

    UNION ALL

    SELECT
        entity_id,
        country,
        lower(
            regexp_replace(
                coalesce(business_name, ''),
                '[^\\p{L}\\p{N}]+',
                ' ',
                'g'
            )
        ) AS name_clean

    FROM read_csv(
        'train_source3.tsv',
        delim='\\t',
        header=true
    )
),

tokens AS (

    SELECT DISTINCT
        entity_id,
        country,

        CASE
            WHEN length(trim(token)) = 2
            THEN lower(trim(token))
            ELSE substr(lower(trim(token)), 1, 3)
        END AS key

    FROM source_data,
    UNNEST(string_split(name_clean, ' '))
    AS t(token)

    WHERE length(trim(token)) >= 2
)

SELECT
    l.country,
    l.key,
    COUNT(DISTINCT t.entity_id) AS record_count

FROM short_key_lookup l

LEFT JOIN tokens t
    ON t.country = l.country
   AND t.key = l.key

GROUP BY
    l.country,
    l.key

ORDER BY record_count DESC
""").fetchdf()


print("\n===== SHORT TOKEN FREQUENCY =====")
display(short_freq.head(20))
```

```python
# ==========================================
# 4-CHARACTER NAME KEY FREQUENCY
# ==========================================

name4_rows = []

for _, row in sample_100_s1.iterrows():

    country = row["country"]

    for prefix in set(row["name_prefixes"]):

        # Only 4+ character prefixes
        if len(prefix) >= 4:
            name4_rows.append({
                "country": country,
                "key": prefix[:4]
            })

name4_lookup_df = pd.DataFrame(
    name4_rows
).drop_duplicates()

con.register("name4_lookup", name4_lookup_df)

print(
    "Unique 4-char S1 keys:",
    len(name4_lookup_df)
)


name4_freq = con.execute("""
WITH source_data AS (

    SELECT
        entity_id,
        country,
        lower(
            regexp_replace(
                coalesce(business_name, ''),
                '[^\\p{L}\\p{N}]+',
                ' ',
                'g'
            )
        ) AS name_clean

    FROM read_csv(
        'train_source2.tsv',
        delim='\t',
        header=true
    )

    UNION ALL

    SELECT
        entity_id,
        country,
        lower(
            regexp_replace(
                coalesce(business_name, ''),
                '[^\\p{L}\\p{N}]+',
                ' ',
                'g'
            )
        ) AS name_clean

    FROM read_csv(
        'train_source3.tsv',
        delim='\t',
        header=true
    )
),

source_keys AS (

    SELECT DISTINCT
        entity_id,
        country,
        substr(lower(trim(token)), 1, 4) AS key

    FROM source_data,
    UNNEST(string_split(name_clean, ' '))
    AS t(token)

    WHERE length(trim(token)) >= 4
)

SELECT
    l.country,
    l.key,
    COUNT(DISTINCT s.entity_id) AS record_count

FROM name4_lookup l

LEFT JOIN source_keys s
    ON s.country = l.country
   AND s.key = l.key

GROUP BY
    l.country,
    l.key

ORDER BY record_count
""").fetchdf()


print("\n===== 4-CHARACTER KEY FREQUENCY =====")
display(name4_freq.head(30))
```

```python
# ==========================================
# MAP USEFUL 4-CHAR KEYS TO ACTUAL S1 IDs
# ==========================================

useful_4_set = set(
    zip(
        useful_4char["country"],
        useful_4char["key"]
    )
)

rows = []

for _, row in sample_100_s1.iterrows():

    s1_id = row["entity_id"]
    country = row["country"]

    for prefix in set(row["name_prefixes"]):

        key = prefix[:4]

        if (country, key) in useful_4_set:

            rows.append({
                "s1_id": s1_id,
                "country": country,
                "key": key
            })

useful_4_s1_df = pd.DataFrame(rows).drop_duplicates()

con.register(
    "useful_4_s1",
    useful_4_s1_df
)

print("Useful 4-char S1-key mappings:",
      len(useful_4_s1_df))

display(useful_4_s1_df.head(20))
```

```python
# ==========================================
# GENERATE CANDIDATES USING USEFUL 4-CHAR KEYS
# ==========================================

con.execute("""
PRAGMA threads=4;

CREATE OR REPLACE TEMP TABLE useful_4_candidates AS

WITH source_data AS (

    SELECT
        entity_id,
        country,
        lower(
            regexp_replace(
                coalesce(business_name, ''),
                '[^\\p{L}\\p{N}]+',
                ' ',
                'g'
            )
        ) AS name_clean

    FROM read_csv(
        'train_source2.tsv',
        delim='\\t',
        header=true
    )

    UNION ALL

    SELECT
        entity_id,
        country,
        lower(
            regexp_replace(
                coalesce(business_name, ''),
                '[^\\p{L}\\p{N}]+',
                ' ',
                'g'
            )
        ) AS name_clean

    FROM read_csv(
        'train_source3.tsv',
        delim='\\t',
        header=true
    )
),

source_keys AS (

    SELECT DISTINCT
        s.entity_id,
        s.country,
        substr(
            lower(trim(u.token)),
            1,
            4
        ) AS key

    FROM source_data s,
    UNNEST(
        string_split(s.name_clean, ' ')
    ) AS u(token)

    WHERE length(trim(u.token)) >= 4

      AND lower(trim(u.token)) NOT IN (
          'inc',
          'ltd',
          'llc',
          'pllc',
          'corp',
          'company',
          'private',
          'public'
      )
)

SELECT DISTINCT
    l.s1_id,
    s.entity_id AS candidate_id

FROM source_keys s

JOIN useful_4_s1 l
  ON s.country = l.country
 AND s.key = l.key
""")

print("✅ Useful 4-char candidates generated")
```

```python
# ==========================================
# COMBINE OLD + NEW CANDIDATES
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE improved_candidates AS

SELECT DISTINCT
    s1_id,
    candidate_id
FROM blocking_candidates

UNION

SELECT DISTINCT
    s1_id,
    candidate_id
FROM useful_4_candidates
""")

result = con.execute("""
SELECT
    COUNT(*) AS total_candidate_pairs,
    COUNT(DISTINCT s1_id) AS s1_count
FROM improved_candidates
""").fetchdf()

display(result)
```

```python
from rapidfuzz import fuzz
import re

# Combine the missed records we already found
all_missed_records = pd.concat(
    [missed_s2, missed_s3],
    ignore_index=True
)

# Make lookup: candidate_id -> record
candidate_lookup = {
    row["entity_id"]: row
    for _, row in all_missed_records.iterrows()
}

# Make lookup: S1 id -> record
s1_lookup = {
    row["entity_id"]: row
    for _, row in train_s1.iterrows()
}

analysis_rows = []

for _, pair in missed_true_pairs_df.iterrows():

    s1_id = pair["s1_id"]
    candidate_id = pair["missed_id"]

    s1_row = s1_lookup[s1_id]
    c_row = candidate_lookup[candidate_id]

    s1_name = normalize_text(s1_row["business_name"])
    c_name = normalize_text(c_row["business_name"])

    s1_addr = normalize_text(s1_row["business_address"])
    c_addr = normalize_text(c_row["business_address"])

    name_score = fuzz.token_set_ratio(
        s1_name,
        c_name
    ) / 100

    address_score = fuzz.token_set_ratio(
        s1_addr,
        c_addr
    ) / 100

    s1_numbers = set(
        re.findall(r"\d+", s1_addr)
    )

    c_numbers = set(
        re.findall(r"\d+", c_addr)
    )

    number_overlap = len(
        s1_numbers & c_numbers
    )

    analysis_rows.append({
        "s1_id": s1_id,
        "missed_id": candidate_id,
        "s1_name": s1_row["business_name"],
        "candidate_name": c_row["business_name"],
        "s1_address": s1_row["business_address"],
        "candidate_address": c_row["business_address"],
        "name_score": round(name_score, 3),
        "address_score": round(address_score, 3),
        "common_numbers": number_overlap
    })

missed_analysis = pd.DataFrame(analysis_rows)

display(missed_analysis)
```

```python
# ==========================================
# ANALYZE THE 16 MISSED TRUE PAIRS
# Simple and safe version
# ==========================================

def core_name(text):
    text = normalize_text(text)

    tokens = text.split()
    cleaned = []

    for token in tokens:
        if token in LEGAL_SUFFIXES:
            continue

        if token in {"com", "www", "net", "org"}:
            continue

        cleaned.append(token)

    return "".join(cleaned)


def address_tokens(text):
    if pd.isna(text):
        return set()

    text = normalize_text(text)

    stopwords = {
        "street", "st", "road", "rd",
        "avenue", "ave", "lane", "ln",
        "drive", "dr", "court", "ct",
        "boulevard", "blvd",
        "house", "no", "number",
        "floor", "unit", "block"
    }

    result = set()

    for token in text.split():
        if len(token) >= 4 and token not in stopwords:
            result.add(token[:4])

    return result


def normalized_numbers(text):
    if pd.isna(text):
        return set()

    nums = re.findall(r"\d+", str(text))

    result = set()

    for num in nums:
        num = num.lstrip("0")

        if num == "":
            num = "0"

        result.add(num)

    return result


# Useful 3-character keys
useful_short_set = set()

for _, row in useful_short.iterrows():
    useful_short_set.add(
        (row["country"], row["key"])
    )


results = []

for _, row in missed_analysis.iterrows():

    s1_id = row["s1_id"]
    missed_id = row["missed_id"]

    country = s1_lookup[s1_id]["country"]

    s1_name = row["s1_name"]
    c_name = row["candidate_name"]

    s1_address = row["s1_address"]
    c_address = row["candidate_address"]

    # 1. Core name similarity
    s1_core = core_name(s1_name)
    c_core = core_name(c_name)

    core_match = False

    if len(s1_core) >= 8 and len(c_core) >= 8:
        if s1_core == c_core:
            core_match = True
        elif s1_core in c_core:
            core_match = True
        elif c_core in s1_core:
            core_match = True

    # 2. Rare/short token
    s1_tokens = normalize_text(s1_name).split()
    c_tokens = normalize_text(c_name).split()

    s1_short = set()
    c_short = set()

    for token in s1_tokens:
        key = token[:3]

        if (country, key) in useful_short_set:
            s1_short.add(key)

    for token in c_tokens:
        key = token[:3]

        if (country, key) in useful_short_set:
            c_short.add(key)

    short_match = bool(
        s1_short.intersection(c_short)
    )

    # 3. Address token overlap
    s1_addr = address_tokens(s1_address)
    c_addr = address_tokens(c_address)

    common_addr = s1_addr.intersection(c_addr)

    address_match = len(common_addr) >= 2

    # 4. Normalized address numbers
    s1_nums = normalized_numbers(s1_address)
    c_nums = normalized_numbers(c_address)

    number_match = bool(
        s1_nums.intersection(c_nums)
    )

    # 5. Number + address token
    strong_address = (
        number_match and
        len(common_addr) >= 1
    )

    # Final experimental capture
    captured = (
        core_match
        or short_match
        or address_match
        or number_match
        or strong_address
    )

    results.append({
        "s1_id": s1_id,
        "missed_id": missed_id,
        "core_name": core_match,
        "short_token": short_match,
        "address_tokens": address_match,
        "number_match": number_match,
        "strong_address": strong_address,
        "captured": captured
    })


better_test = pd.DataFrame(results)

display(better_test)

print(
    "\nCaptured:",
    int(better_test["captured"].sum()),
    "out of",
    len(better_test)
)
```

```python
print("===== RECOVERY SIGNAL COVERAGE =====")

print(
    "Core name:",
    int(better_test["core_name"].sum()),
    "of",
    len(better_test)
)

print(
    "Rare/short token:",
    int(better_test["short_token"].sum()),
    "of",
    len(better_test)
)

print(
    "Address tokens:",
    int(better_test["address_tokens"].sum()),
    "of",
    len(better_test)
)

print(
    "Normalized number:",
    int(better_test["number_match"].sum()),
    "of",
    len(better_test)
)

print(
    "Strong address:",
    int(better_test["strong_address"].sum()),
    "of",
    len(better_test)
)
```

```python
# ==========================================
# TEST GENERALIZED RECOVERY RULES
# ==========================================

def compact_key(text):
    """
    Remove legal/domain suffixes and spaces.
    Keep first 8 characters.
    """
    return core_name(text)[:8]


def rare_tokens(text, country):
    tokens = normalize_text(text).split()

    result = set()

    for token in tokens:
        key = token[:3]

        if (country, key) in useful_short_set:
            result.add(key)

    return result


def addr_prefix_tokens(text):
    if pd.isna(text):
        return set()

    tokens = normalize_text(text).split()

    stopwords = {
        "street", "st", "road", "rd",
        "avenue", "ave", "lane", "ln",
        "drive", "dr", "court", "ct",
        "boulevard", "blvd",
        "house", "no", "number",
        "floor", "unit", "block",
        "building"
    }

    return {
        token[:4]
        for token in tokens
        if len(token) >= 4
        and token not in stopwords
    }


recovery_test = []

for _, row in missed_analysis.iterrows():

    s1_id = row["s1_id"]
    country = s1_lookup[s1_id]["country"]

    s1_name = row["s1_name"]
    c_name = row["candidate_name"]

    s1_address = row["s1_address"]
    c_address = row["candidate_address"]

    # -----------------------------
    # 1. Compact name
    # -----------------------------
    s1_compact = compact_key(s1_name)
    c_compact = compact_key(c_name)

    compact_match = (
        len(s1_compact) >= 6
        and len(c_compact) >= 6
        and s1_compact == c_compact
    )

    # -----------------------------
    # 2. Rare short token
    # -----------------------------
    s1_rare = rare_tokens(s1_name, country)
    c_rare = rare_tokens(c_name, country)

    rare_match = bool(
        s1_rare.intersection(c_rare)
    )

    # -----------------------------
    # 3. Two informative address tokens
    # -----------------------------
    s1_addr = addr_prefix_tokens(s1_address)
    c_addr = addr_prefix_tokens(c_address)

    common_addr = s1_addr.intersection(c_addr)

    address_pair_match = (
        len(common_addr) >= 2
    )

    # -----------------------------
    # Final recovery rule
    # -----------------------------
    captured = (
        compact_match
        or rare_match
        or address_pair_match
    )

    recovery_test.append({
        "s1_id": s1_id,
        "missed_id": row["missed_id"],
        "compact_match": compact_match,
        "rare_token_match": rare_match,
        "address_pair_match": address_pair_match,
        "captured": captured
    })


recovery_test = pd.DataFrame(recovery_test)

display(recovery_test)

print(
    "\nRecovery captured:",
    int(recovery_test["captured"].sum()),
    "out of",
    len(recovery_test)
)
```

```python
# ==========================================
# BUILD RECOVERY LOOKUPS FOR 100 S1
# ==========================================

recovery_name_lookup = defaultdict(set)
recovery_short_lookup = defaultdict(set)
recovery_addr_pair_lookup = defaultdict(set)

for _, row in sample_100_s1.iterrows():

    s1_id = row["entity_id"]
    country = row["country"]

    # --------------------------------------
    # 1. COMPACT NAME LOOKUP
    # --------------------------------------

    compact = compact_key(row["business_name"])

    if len(compact) >= 8:
        recovery_name_lookup[
            (country, compact[:8])
        ].add(s1_id)

    # --------------------------------------
    # 2. RARE SHORT TOKEN LOOKUP
    # --------------------------------------

    for token in normalize_text(
        row["business_name"]
    ).split():

        key = token[:3]

        if (country, key) in useful_short_set:
            recovery_short_lookup[
                (country, key)
            ].add(s1_id)

    # --------------------------------------
    # 3. ADDRESS TOKEN-PAIR LOOKUP
    # --------------------------------------

    addr_tokens = sorted(
        addr_prefix_tokens(
            row["business_address"]
        )
    )

    for pair in combinations(addr_tokens, 2):

        recovery_addr_pair_lookup[
            (country, pair)
        ].add(s1_id)


print("Recovery lookup created ✅")

print(
    "Compact-name keys:",
    len(recovery_name_lookup)
)

print(
    "Rare-token keys:",
    len(recovery_short_lookup)
)

print(
    "Address-pair keys:",
    len(recovery_addr_pair_lookup)
)
```

```python
# ==========================================
# REGISTER RECOVERY LOOKUPS IN DUCKDB
# ==========================================

# -------- Compact-name lookup --------
compact_rows = []

for (country, key), s1_ids in recovery_name_lookup.items():
    for s1_id in s1_ids:
        compact_rows.append({
            "s1_id": s1_id,
            "country": country,
            "compact_key": key
        })

compact_df = pd.DataFrame(compact_rows).drop_duplicates()


# -------- Rare-token lookup --------
short_rows = []

for (country, key), s1_ids in recovery_short_lookup.items():
    for s1_id in s1_ids:
        short_rows.append({
            "s1_id": s1_id,
            "country": country,
            "short_key": key
        })

recovery_short_df = pd.DataFrame(short_rows).drop_duplicates()


# -------- Address-pair lookup --------
addr_rows = []

for (country, pair), s1_ids in recovery_addr_pair_lookup.items():
    for s1_id in s1_ids:
        addr_rows.append({
            "s1_id": s1_id,
            "country": country,
            "addr_key1": pair[0],
            "addr_key2": pair[1]
        })

recovery_addr_df = pd.DataFrame(addr_rows).drop_duplicates()


# Register in DuckDB
con.register("recovery_compact", compact_df)
con.register("recovery_short", recovery_short_df)
con.register("recovery_addr", recovery_addr_df)

print("Recovery lookups registered ✅")

print("Compact mappings:", len(compact_df))
print("Rare-token mappings:", len(recovery_short_df))
print("Address-pair mappings:", len(recovery_addr_df))
```

```python
# ==========================================
# ADD RECOVERY BLOCKING TO EXISTING CANDIDATES
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE recovery_candidates AS

WITH source_data AS (

    SELECT
        entity_id,
        country,

        -- Clean business name
        lower(
            regexp_replace(
                coalesce(business_name, ''),
                '[^\\p{L}\\p{N}]+',
                ' ',
                'g'
            )
        ) AS name_clean,

        -- Clean address
        lower(
            regexp_replace(
                coalesce(business_address, ''),
                '[^\\p{L}\\p{N}]+',
                ' ',
                'g'
            )
        ) AS address_clean

    FROM read_csv(
        'train_source2.tsv',
        delim='\t',
        header=true
    )

    UNION ALL

    SELECT
        entity_id,
        country,

        lower(
            regexp_replace(
                coalesce(business_name, ''),
                '[^\\p{L}\\p{N}]+',
                ' ',
                'g'
            )
        ) AS name_clean,

        lower(
            regexp_replace(
                coalesce(business_address, ''),
                '[^\\p{L}\\p{N}]+',
                ' ',
                'g'
            )
        ) AS address_clean

    FROM read_csv(
        'train_source3.tsv',
        delim='\t',
        header=true
    )
),

source_name_keys AS (

    SELECT DISTINCT
        entity_id,
        country,

        -- first 8 chars of compact name
        substr(
            regexp_replace(name_clean, ' ', '', 'g'),
            1,
            8
        ) AS compact_key,

        -- all 3-char token keys
        substr(
            lower(trim(token)),
            1,
            3
        ) AS short_key

    FROM source_data,
    UNNEST(
        string_split(name_clean, ' ')
    ) AS t(token)

    WHERE length(trim(token)) >= 2
),

source_address_tokens AS (

    SELECT DISTINCT
        entity_id,
        country,
        substr(
            lower(trim(token)),
            1,
            4
        ) AS addr_key

    FROM source_data,
    UNNEST(
        string_split(address_clean, ' ')
    ) AS t(token)

    WHERE length(trim(token)) >= 4
),

source_address_numbers AS (

    SELECT DISTINCT
        entity_id,
        country,
        number

    FROM source_data,
    UNNEST(
        regexp_extract_all(
            address_clean,
            '\\d+'
        )
    ) AS t(number)

    WHERE number <> ''
),

-- ========================================
-- Compact name recovery
-- ========================================

compact_matches AS (

    SELECT DISTINCT
        l.s1_id,
        s.entity_id AS candidate_id

    FROM source_name_keys s

    JOIN recovery_compact l
      ON s.country = l.country
     AND s.compact_key = l.compact_key
),

-- ========================================
-- Rare token recovery
-- ========================================

short_matches AS (

    SELECT DISTINCT
        l.s1_id,
        s.entity_id AS candidate_id

    FROM source_name_keys s

    JOIN recovery_short l
      ON s.country = l.country
     AND s.short_key = l.short_key
),

-- ========================================
-- Address-pair recovery
-- ========================================

address_matches AS (

    SELECT
        l.s1_id,
        t.entity_id AS candidate_id

    FROM source_address_tokens t

    JOIN recovery_addr l
      ON t.country = l.country
     AND t.addr_key IN (
         l.addr_key1,
         l.addr_key2
     )

    GROUP BY
        l.s1_id,
        t.entity_id

    HAVING COUNT(
        DISTINCT t.addr_key
    ) >= 2
)

-- ========================================
-- Combine all recovery routes
-- ========================================

SELECT DISTINCT
    s1_id,
    candidate_id
FROM compact_matches

UNION

SELECT DISTINCT
    s1_id,
    candidate_id
FROM short_matches

UNION

SELECT DISTINCT
    s1_id,
    candidate_id
FROM address_matches
""")

print("✅ Recovery candidates created")
```

```python
# ==========================================
# STEP 1: CREATE ML TRAINING PAIRS
# ==========================================

# Ground Truth -> positive pairs
gt_pairs = []

for _, row in gt_100.iterrows():

    s1_id = row["source1_entity_id"]
    matched = row["matched_entity_ids"]

    if pd.notna(matched) and matched != "":
        for candidate_id in matched.split(","):
            gt_pairs.append({
                "s1_id": s1_id,
                "candidate_id": candidate_id,
                "label": 1
            })

gt_pairs_100 = pd.DataFrame(gt_pairs)

con.register("gt_pairs_100", gt_pairs_100)

print("Ground-truth positive pairs:", len(gt_pairs_100))


# ------------------------------------------
# Label every blocking candidate
# ------------------------------------------

con.execute("""
CREATE OR REPLACE TEMP TABLE labeled_candidates AS

SELECT
    b.s1_id,
    b.candidate_id,

    CASE
        WHEN g.candidate_id IS NOT NULL
        THEN 1
        ELSE 0
    END AS label

FROM blocking_candidates b

LEFT JOIN gt_pairs_100 g
    ON b.s1_id = g.s1_id
   AND b.candidate_id = g.candidate_id
""")


# ------------------------------------------
# Keep all positives
# + 25 negative candidates per S1
# ------------------------------------------

con.execute("""
CREATE OR REPLACE TEMP TABLE ml_pairs AS

WITH negatives AS (

    SELECT
        s1_id,
        candidate_id,
        label,
        ROW_NUMBER() OVER (
            PARTITION BY s1_id
            ORDER BY hash(candidate_id)
        ) AS rn

    FROM labeled_candidates

    WHERE label = 0
)

SELECT
    s1_id,
    candidate_id,
    label

FROM labeled_candidates

WHERE label = 1

UNION ALL

SELECT
    s1_id,
    candidate_id,
    label

FROM negatives

WHERE rn <= 25
""")


# ------------------------------------------
# Check result
# ------------------------------------------

result = con.execute("""
SELECT
    label,
    COUNT(*) AS pair_count
FROM ml_pairs
GROUP BY label
ORDER BY label
""").fetchdf()

print("\nML pair distribution:")
display(result)

total = con.execute("""
SELECT COUNT(*) FROM ml_pairs
""").fetchone()[0]

print("Total ML pairs:", total)
```

```python
# ==========================================
# FETCH S1 + CANDIDATE S2/S3 RECORDS
# ==========================================

ml_pairs = con.execute("""
    SELECT
        s.s1_id,
        s.candidate_id,
        s.label,

        s1.business_name AS s1_name,
        s1.business_address AS s1_address,
        s1.country AS s1_country,

        c.business_name AS candidate_name,
        c.business_address AS candidate_address,
        c.country AS candidate_country

    FROM ml_pairs s

    INNER JOIN (
        SELECT
            entity_id,
            business_name,
            business_address,
            country
        FROM read_csv(
            'train_source1.tsv',
            delim='\t',
            header=true
        )
    ) s1
        ON s.s1_id = s1.entity_id

    INNER JOIN (
        SELECT
            entity_id,
            business_name,
            business_address,
            country

        FROM read_csv(
            'train_source2.tsv',
            delim='\t',
            header=true
        )

        UNION ALL

        SELECT
            entity_id,
            business_name,
            business_address,
            country

        FROM read_csv(
            'train_source3.tsv',
            delim='\t',
            header=true
        )
    ) c
        ON s.candidate_id = c.entity_id
""").fetchdf()

print("ML records fetched ✅")
print("Rows:", len(ml_pairs))

display(ml_pairs.head(10))
```

```python
# ==========================================
# STEP 2: FEATURE ENGINEERING
# ==========================================

from rapidfuzz import fuzz
import re
import numpy as np


def safe_text(x):
    if pd.isna(x):
        return ""
    return str(x)


def get_numbers(text):
    text = safe_text(text)

    nums = re.findall(r"\d+", text)

    # Normalize 00253 -> 253
    result = set()

    for num in nums:
        num = num.lstrip("0")

        if num == "":
            num = "0"

        result.add(num)

    return result


def token_overlap(a, b):
    a_tokens = set(normalize_text(a).split())
    b_tokens = set(normalize_text(b).split())

    if len(a_tokens) == 0 and len(b_tokens) == 0:
        return 1.0

    if len(a_tokens) == 0 or len(b_tokens) == 0:
        return 0.0

    return len(a_tokens & b_tokens) / len(a_tokens | b_tokens)


def number_overlap(a, b):
    a_nums = get_numbers(a)
    b_nums = get_numbers(b)

    if len(a_nums) == 0 or len(b_nums) == 0:
        return 0.0

    return len(a_nums & b_nums) / len(a_nums | b_nums)


# ------------------------------------------
# Name features
# ------------------------------------------

ml_pairs["name_ratio"] = ml_pairs.apply(
    lambda r: fuzz.ratio(
        safe_text(r["s1_name"]),
        safe_text(r["candidate_name"])
    ) / 100,
    axis=1
)

ml_pairs["name_token_set"] = ml_pairs.apply(
    lambda r: fuzz.token_set_ratio(
        safe_text(r["s1_name"]),
        safe_text(r["candidate_name"])
    ) / 100,
    axis=1
)

ml_pairs["name_partial"] = ml_pairs.apply(
    lambda r: fuzz.partial_ratio(
        safe_text(r["s1_name"]),
        safe_text(r["candidate_name"])
    ) / 100,
    axis=1
)

ml_pairs["name_token_overlap"] = ml_pairs.apply(
    lambda r: token_overlap(
        r["s1_name"],
        r["candidate_name"]
    ),
    axis=1
)


# ------------------------------------------
# Address features
# ------------------------------------------

ml_pairs["address_ratio"] = ml_pairs.apply(
    lambda r: fuzz.ratio(
        safe_text(r["s1_address"]),
        safe_text(r["candidate_address"])
    ) / 100,
    axis=1
)

ml_pairs["address_token_set"] = ml_pairs.apply(
    lambda r: fuzz.token_set_ratio(
        safe_text(r["s1_address"]),
        safe_text(r["candidate_address"])
    ) / 100,
    axis=1
)

ml_pairs["address_partial"] = ml_pairs.apply(
    lambda r: fuzz.partial_ratio(
        safe_text(r["s1_address"]),
        safe_text(r["candidate_address"])
    ) / 100,
    axis=1
)

ml_pairs["address_token_overlap"] = ml_pairs.apply(
    lambda r: token_overlap(
        r["s1_address"],
        r["candidate_address"]
    ),
    axis=1
)

ml_pairs["number_overlap"] = ml_pairs.apply(
    lambda r: number_overlap(
        r["s1_address"],
        r["candidate_address"]
    ),
    axis=1
)


# ------------------------------------------
# Country
# ------------------------------------------

ml_pairs["country_same"] = (
    ml_pairs["s1_country"] ==
    ml_pairs["candidate_country"]
).astype(int)


# ------------------------------------------
# Missing-value features
# ------------------------------------------

ml_pairs["candidate_name_missing"] = (
    ml_pairs["candidate_name"]
    .isna()
    .astype(int)
)

ml_pairs["candidate_address_missing"] = (
    ml_pairs["candidate_address"]
    .isna()
    .astype(int)
)


print("Feature engineering complete ✅")

feature_columns = [
    "name_ratio",
    "name_token_set",
    "name_partial",
    "name_token_overlap",
    "address_ratio",
    "address_token_set",
    "address_partial",
    "address_token_overlap",
    "number_overlap",
    "country_same",
    "candidate_name_missing",
    "candidate_address_missing"
]

display(
    ml_pairs[
        ["s1_id", "candidate_id", "label"] +
        feature_columns
    ].head(10)
)
```

```python
import numpy as np
import pandas as pd

def f05_score(true_set, predicted_set):

    if len(true_set) == 0:
        return 1.0 if len(predicted_set) == 0 else 0.0

    if len(predicted_set) == 0:
        return 0.0

    tp = len(true_set & predicted_set)
    fp = len(predicted_set - true_set)
    fn = len(true_set - predicted_set)

    precision = tp / (tp + fp) if (tp + fp) else 0
    recall = tp / (tp + fn) if (tp + fn) else 0

    if precision == 0 and recall == 0:
        return 0.0

    return (
        1.25 * precision * recall
    ) / (
        0.25 * precision + recall
    )


# Ground truth lookup
gt_lookup = {}

for _, row in gt_100.iterrows():

    matched = row["matched_entity_ids"]

    if pd.isna(matched) or matched == "":
        gt_lookup[row["source1_entity_id"]] = set()
    else:
        gt_lookup[row["source1_entity_id"]] = set(
            matched.split(",")
        )


threshold_results = []

for threshold in np.arange(0.50, 0.96, 0.01):

    scores = []

    for s1_id in val_s1_ids:

        rows = val_data[
            val_data["s1_id"] == s1_id
        ]

        predicted = set(
            rows.loc[
                rows["model_prob"] >= threshold,
                "candidate_id"
            ]
        )

        true_set = gt_lookup.get(
            s1_id,
            set()
        )

        score = f05_score(
            true_set,
            predicted
        )

        scores.append(score)

    threshold_results.append({
        "threshold": round(float(threshold), 2),
        "macro_F0.5": np.mean(scores)
    })


threshold_df = pd.DataFrame(threshold_results)

print("===== THRESHOLD RESULTS =====")

display(
    threshold_df.sort_values(
        "macro_F0.5",
        ascending=False
    ).head(10)
)
```

```python
# ==========================================
# STEP 5: INSPECT VALIDATION PREDICTIONS
# ==========================================

BEST_THRESHOLD = 0.94

val_data = val_data.copy()

val_data["pred_match"] = (
    val_data["model_prob"] >= BEST_THRESHOLD
).astype(int)

print("Threshold:", BEST_THRESHOLD)

print("\nPredicted matches:",
      val_data["pred_match"].sum())

print("Actual positive labels:",
      val_data["label"].sum())

print("\nPredicted label distribution:")
print(val_data["pred_match"].value_counts())

print("\nHighest probability pairs:")

display(
    val_data[
        [
            "s1_id",
            "candidate_id",
            "label",
            "model_prob",
            "name_token_set",
            "address_token_set",
            "number_overlap",
            "country_same"
        ]
    ]
    .sort_values(
        "model_prob",
        ascending=False
    )
    .head(20)
)
```

```python
# ==========================================
# CHECK VALIDATION PERFORMANCE
# ==========================================

from sklearn.metrics import confusion_matrix, precision_score, recall_score

y_true = val_data["label"]
y_pred = val_data["pred_match"]

cm = confusion_matrix(
    y_true,
    y_pred
)

print("Confusion Matrix:")
print(cm)

precision = precision_score(
    y_true,
    y_pred,
    zero_division=0
)

recall = recall_score(
    y_true,
    y_pred,
    zero_division=0
)

print("\nPrecision:", round(precision, 4))
print("Recall:", round(recall, 4))
print("Predicted matches:", int(y_pred.sum()))
print("Actual matches:", int(y_true.sum()))
```

```python
# ==========================================
# STEP 6: FINAL MODEL ON ALL 100 S1s
# ==========================================

from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.utils.class_weight import compute_sample_weight

# Use all 2,799 labeled pairs
X_all = ml_pairs[feature_columns]
y_all = ml_pairs["label"]

# Handle class imbalance
all_weights = compute_sample_weight(
    class_weight="balanced",
    y=y_all
)

# Train final model
final_model = HistGradientBoostingClassifier(
    max_iter=250,
    learning_rate=0.05,
    max_leaf_nodes=15,
    l2_regularization=1.0,
    random_state=42
)

final_model.fit(
    X_all,
    y_all,
    sample_weight=all_weights
)

print("✅ Final model trained")
print("Training pairs:", len(ml_pairs))
print("Features:", len(feature_columns))
print("Final threshold:", 0.94)
```

```python
# ==========================================
# STEP 8: TEST 4-CHAR NAME KEY FREQUENCY
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE test_name4_freq AS

WITH source_data AS (

    SELECT
        entity_id,
        country,
        lower(
            regexp_replace(
                coalesce(business_name, ''),
                '[^\\p{L}\\p{N}]+',
                ' ',
                'g'
            )
        ) AS name_clean

    FROM read_csv(
        'test_source2.tsv',
        delim='\\t',
        header=true
    )

    UNION ALL

    SELECT
        entity_id,
        country,
        lower(
            regexp_replace(
                coalesce(business_name, ''),
                '[^\\p{L}\\p{N}]+',
                ' ',
                'g'
            )
        ) AS name_clean

    FROM read_csv(
        'test_source3.tsv',
        delim='\\t',
        header=true
    )
),

tokens AS (

    SELECT DISTINCT
        entity_id,
        country,
        substr(
            lower(trim(token)),
            1,
            4
        ) AS key

    FROM source_data,
    UNNEST(
        string_split(name_clean, ' ')
    ) AS t(token)

    WHERE length(trim(token)) >= 4

      AND lower(trim(token)) NOT IN (
          'inc',
          'ltd',
          'llc',
          'pllc',
          'corp',
          'company',
          'private',
          'public',
          'com',
          'www',
          'net',
          'org'
      )
)

SELECT
    country,
    key,
    COUNT(DISTINCT entity_id) AS record_count

FROM tokens

GROUP BY
    country,
    key
""")

print("✅ Test 4-character frequency table created")

freq_summary = con.execute("""
SELECT
    country,
    COUNT(*) AS unique_keys,
    MIN(record_count) AS minimum_frequency,
    MAX(record_count) AS maximum_frequency
FROM test_name4_freq
GROUP BY country
ORDER BY country
""").fetchdf()

display(freq_summary)
```

```python
# ==========================================
# HOW MANY TEST KEYS ARE RARE?
# ==========================================

distribution = con.execute("""
SELECT
    SUM(CASE WHEN record_count <= 10 THEN 1 ELSE 0 END) AS le_10,
    SUM(CASE WHEN record_count <= 50 THEN 1 ELSE 0 END) AS le_50,
    SUM(CASE WHEN record_count <= 100 THEN 1 ELSE 0 END) AS le_100,
    SUM(CASE WHEN record_count <= 500 THEN 1 ELSE 0 END) AS le_500,
    SUM(CASE WHEN record_count <= 1000 THEN 1 ELSE 0 END) AS le_1000,
    COUNT(*) AS total_keys
FROM test_name4_freq
""").fetchdf()

display(distribution)
```

```python
# ==========================================
# FAST TEST SOURCE TABLE
# ==========================================

con.execute("PRAGMA threads=8")

con.execute("""
CREATE OR REPLACE TEMP TABLE test_source_fast AS

SELECT
    entity_id AS candidate_id,
    country,

    -- normalized full name
    lower(
        regexp_replace(
            coalesce(business_name, ''),
            '[^\\p{L}\\p{N}]+',
            ' ',
            'g'
        )
    ) AS name_clean,

    -- compact first 8 characters
    substr(
        regexp_replace(
            lower(coalesce(business_name, '')),
            '[^\\p{L}\\p{N}]',
            '',
            'g'
        ),
        1,
        8
    ) AS compact8,

    -- first 4 name-token prefixes
    substr(
        regexp_extract(
            lower(
                regexp_replace(
                    coalesce(business_name, ''),
                    '[^\\p{L}\\p{N}]+',
                    ' ',
                    'g'
                )
            ),
            '^([^ ]+)',
            1
        ),
        1, 4
    ) AS token1,

    substr(
        regexp_extract(
            lower(
                regexp_replace(
                    coalesce(business_name, ''),
                    '[^\\p{L}\\p{N}]+',
                    ' ',
                    'g'
                )
            ),
            '^[^ ]+ ([^ ]+)',
            1
        ),
        1, 4
    ) AS token2,

    substr(
        regexp_extract(
            lower(
                regexp_replace(
                    coalesce(business_name, ''),
                    '[^\\p{L}\\p{N}]+',
                    ' ',
                    'g'
                )
            ),
            '^[^ ]+ [^ ]+ ([^ ]+)',
            1
        ),
        1, 4
    ) AS token3

FROM read_csv(
    'test_source2.tsv',
    delim='\t',
    header=true
)

UNION ALL

SELECT
    entity_id AS candidate_id,
    country,

    lower(
        regexp_replace(
            coalesce(business_name, ''),
            '[^\\p{L}\\p{N}]+',
            ' ',
            'g'
        )
    ) AS name_clean,

    substr(
        regexp_replace(
            lower(coalesce(business_name, '')),
            '[^\\p{L}\\p{N}]',
            '',
            'g'
        ),
        1,
        8
    ) AS compact8,

    substr(
        regexp_extract(
            lower(
                regexp_replace(
                    coalesce(business_name, ''),
                    '[^\\p{L}\\p{N}]+',
                    ' ',
                    'g'
                )
            ),
            '^([^ ]+)',
            1
        ),
        1, 4
    ) AS token1,

    substr(
        regexp_extract(
            lower(
                regexp_replace(
                    coalesce(business_name, ''),
                    '[^\\p{L}\\p{N}]+',
                    ' ',
                    'g'
                )
            ),
            '^[^ ]+ ([^ ]+)',
            1
        ),
        1, 4
    ) AS token2,

    substr(
        regexp_extract(
            lower(
                regexp_replace(
                    coalesce(business_name, ''),
                    '[^\\p{L}\\p{N}]+',
                    ' ',
                    'g'
                )
            ),
            '^[^ ]+ [^ ]+ ([^ ]+)',
            1
        ),
        1, 4
    ) AS token3

FROM read_csv(
    'test_source3.tsv',
    delim='\t',
    header=true
)
""")

print("✅ Fast test source table created")
```

```python
# ==========================================
# ADD TOKEN 2 AND TOKEN 3
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE test_tokens AS

SELECT
    entity_id AS candidate_id,
    country,

    substr(split_part(name_clean, ' ', 1), 1, 4) AS token1,
    substr(split_part(name_clean, ' ', 2), 1, 4) AS token2,
    substr(split_part(name_clean, ' ', 3), 1, 4) AS token3

FROM (
    SELECT
        entity_id,
        country,
        lower(
            regexp_replace(
                coalesce(business_name, ''),
                '[^\\p{L}\\p{N}]+',
                ' ',
                'g'
            )
        ) AS name_clean
    FROM read_csv(
        'test_source2.tsv',
        delim='\t',
        header=true
    )

    UNION ALL

    SELECT
        entity_id,
        country,
        lower(
            regexp_replace(
                coalesce(business_name, ''),
                '[^\\p{L}\\p{N}]+',
                ' ',
                'g'
            )
        ) AS name_clean
    FROM read_csv(
        'test_source3.tsv',
        delim='\t',
        header=true
    )
)
""")

print("✅ token1 + token2 + token3 table created")
```

```python
# ==========================================
# PREPARE 10K S1 WITH TOKEN 1/2/3
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE test_s1_tokens_10k AS

SELECT
    entity_id AS source1_entity_id,
    country,

    substr(split_part(name_clean, ' ', 1), 1, 4) AS token1,
    substr(split_part(name_clean, ' ', 2), 1, 4) AS token2,
    substr(split_part(name_clean, ' ', 3), 1, 4) AS token3

FROM (
    SELECT
        entity_id,
        country,
        lower(
            regexp_replace(
                coalesce(business_name, ''),
                '[^\\p{L}\\p{N}]+',
                ' ',
                'g'
            )
        ) AS name_clean
    FROM read_csv(
        'test_source1.tsv',
        delim='\t',
        header=true
    )
)
LIMIT 10000
""")

print("✅ 10K S1 token1/token2/token3 ready")
```

```python
# ==========================================
# FREQUENCY OF TOKEN 1 / 2 / 3
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE token_freq AS

SELECT country, token, COUNT(*) AS freq
FROM (
    SELECT country, token1 AS token
    FROM test_tokens
    WHERE token1 <> ''

    UNION ALL

    SELECT country, token2 AS token
    FROM test_tokens
    WHERE token2 <> ''

    UNION ALL

    SELECT country, token3 AS token
    FROM test_tokens
    WHERE token3 <> ''
)
GROUP BY country, token
""")

print("✅ Token frequency table ready")

print(
    con.execute("""
        SELECT
            MIN(freq) AS min_freq,
            AVG(freq) AS avg_freq,
            MAX(freq) AS max_freq
        FROM token_freq
    """).fetchdf()
)
```

```python
# ==========================================
# RARE TOKEN LOOKUP
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE rare_tokens AS
SELECT
    country,
    token
FROM token_freq
WHERE token <> ''
  AND freq <= 50
""")

print("✅ Rare token lookup created")

print(
    con.execute("""
        SELECT COUNT(*) AS rare_tokens
        FROM rare_tokens
    """).fetchdf()
)
```

```python
# ==========================================
# TEST 3-TOKEN RARE BLOCKING ON 10K S1
# ==========================================

result = con.execute("""
WITH candidates AS (

    -- Match using token1
    SELECT
        s.source1_entity_id,
        t.candidate_id
    FROM test_s1_tokens_10k s
    JOIN test_tokens t
        ON s.country = t.country
       AND s.token1 <> ''
       AND s.token1 = t.token1
    JOIN rare_tokens r
        ON t.country = r.country
       AND t.token1 = r.token

    UNION

    -- Match using token2
    SELECT
        s.source1_entity_id,
        t.candidate_id
    FROM test_s1_tokens_10k s
    JOIN test_tokens t
        ON s.country = t.country
       AND s.token2 <> ''
       AND s.token2 = t.token2
    JOIN rare_tokens r
        ON t.country = r.country
       AND t.token2 = r.token

    UNION

    -- Match using token3
    SELECT
        s.source1_entity_id,
        t.candidate_id
    FROM test_s1_tokens_10k s
    JOIN test_tokens t
        ON s.country = t.country
       AND s.token3 <> ''
       AND s.token3 = t.token3
    JOIN rare_tokens r
        ON t.country = r.country
       AND t.token3 = r.token
)

SELECT COUNT(*) AS candidate_pairs
FROM candidates
""").fetchdf()

print(result)
```

```python
# ==========================================
# TRAIN S2 + S3 TOKENS
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE train_tokens AS

SELECT
    entity_id AS candidate_id,
    country,

    substr(split_part(name_clean, ' ', 1), 1, 4) AS token1,
    substr(split_part(name_clean, ' ', 2), 1, 4) AS token2,
    substr(split_part(name_clean, ' ', 3), 1, 4) AS token3

FROM (
    SELECT
        entity_id,
        country,
        lower(
            regexp_replace(
                coalesce(business_name, ''),
                '[^\\p{L}\\p{N}]+',
                ' ',
                'g'
            )
        ) AS name_clean
    FROM read_csv(
        'train_source2.tsv',
        delim='\t',
        header=true
    )

    UNION ALL

    SELECT
        entity_id,
        country,
        lower(
            regexp_replace(
                coalesce(business_name, ''),
                '[^\\p{L}\\p{N}]+',
                ' ',
                'g'
            )
        ) AS name_clean
    FROM read_csv(
        'train_source3.tsv',
        delim='\t',
        header=true
    )
)
""")

print("✅ Training S2 + S3 token table created")
```

```python
# ==========================================
# TRAIN TOKEN FREQUENCIES
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE train_token_freq AS

SELECT
    country,
    token,
    COUNT(*) AS freq
FROM (
    SELECT country, token1 AS token
    FROM train_tokens
    WHERE token1 <> ''

    UNION ALL

    SELECT country, token2 AS token
    FROM train_tokens
    WHERE token2 <> ''

    UNION ALL

    SELECT country, token3 AS token
    FROM train_tokens
    WHERE token3 <> ''
)
GROUP BY country, token
""")

print("✅ Training token frequency table created")
```

```python
# ==========================================
# RARE TRAINING TOKENS
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE train_rare_tokens AS
SELECT
    country,
    token
FROM train_token_freq
WHERE token <> ''
  AND freq <= 50
""")

print(
    con.execute("""
        SELECT COUNT(*) AS rare_tokens
        FROM train_rare_tokens
    """).fetchdf()
)
```

```python
# ==========================================
# GENERATE RARE-TOKEN CANDIDATES
# FOR 100 TRAINING S1
# ==========================================

# Prepare the 3 tokens for our 100 S1
con.execute("""
CREATE OR REPLACE TEMP TABLE val_s1_tokens AS
SELECT
    entity_id AS source1_entity_id,
    country,

    substr(split_part(name_clean, ' ', 1), 1, 4) AS token1,
    substr(split_part(name_clean, ' ', 2), 1, 4) AS token2,
    substr(split_part(name_clean, ' ', 3), 1, 4) AS token3

FROM (
    SELECT
        entity_id,
        country,
        lower(
            regexp_replace(
                coalesce(business_name, ''),
                '[^\\p{L}\\p{N}]+',
                ' ',
                'g'
            )
        ) AS name_clean
    FROM read_csv(
        'train_source1.tsv',
        delim='\t',
        header=true
    )
)
WHERE entity_id IN (
    SELECT source1_entity_id
    FROM val_gt_100
)
""")

# Generate candidates using each token separately
con.execute("""
CREATE OR REPLACE TEMP TABLE val_candidates AS

SELECT DISTINCT
    s.source1_entity_id,
    t.candidate_id

FROM val_s1_tokens s
JOIN train_tokens t
    ON s.country = t.country
   AND s.token1 <> ''
   AND s.token1 = t.token1
JOIN train_token_freq f
    ON t.country = f.country
   AND t.token1 = f.token
WHERE f.freq <= 50

UNION

SELECT DISTINCT
    s.source1_entity_id,
    t.candidate_id

FROM val_s1_tokens s
JOIN train_tokens t
    ON s.country = t.country
   AND s.token2 <> ''
   AND s.token2 = t.token2
JOIN train_token_freq f
    ON t.country = f.country
   AND t.token2 = f.token
WHERE f.freq <= 50

UNION

SELECT DISTINCT
    s.source1_entity_id,
    t.candidate_id

FROM val_s1_tokens s
JOIN train_tokens t
    ON s.country = t.country
   AND s.token3 <> ''
   AND s.token3 = t.token3
JOIN train_token_freq f
    ON t.country = f.country
   AND t.token3 = f.token
WHERE f.freq <= 50
""")

print("✅ Candidates generated")

print(
    con.execute("""
        SELECT
            COUNT(*) AS candidate_pairs,
            COUNT(DISTINCT source1_entity_id) AS s1_count
        FROM val_candidates
    """).fetchdf()
)
```

```python
# ==========================================
# CALCULATE BLOCKING RECALL
# ==========================================

result = con.execute("""
WITH gt_pairs AS (
    SELECT
        source1_entity_id,
        trim(x) AS candidate_id
    FROM val_gt_100,
    UNNEST(
        string_split(
            COALESCE(matched_entity_ids, ''),
            ','
        )
    ) AS t(x)
    WHERE trim(x) <> ''
),

captured_pairs AS (
    SELECT
        g.source1_entity_id,
        g.candidate_id
    FROM gt_pairs g
    INNER JOIN val_candidates c
        ON g.source1_entity_id = c.source1_entity_id
       AND g.candidate_id = c.candidate_id
)

SELECT
    (SELECT COUNT(*) FROM gt_pairs) AS total_true_matches,
    (SELECT COUNT(*) FROM captured_pairs) AS captured_true_matches,
    ROUND(
        100.0 *
        (SELECT COUNT(*) FROM captured_pairs) /
        NULLIF((SELECT COUNT(*) FROM gt_pairs), 0),
        2
    ) AS blocking_recall_percent
""").fetchdf()

print(result)
```

```python
# ==========================================
# TWO-TOKEN NAME BLOCKING
# ==========================================

# Create an order-independent pair from token1 + token2
con.execute("""
CREATE OR REPLACE TEMP TABLE train_pair12 AS

SELECT
    candidate_id,
    country,

    CASE
        WHEN token1 <= token2
        THEN token1 || '|' || token2
        ELSE token2 || '|' || token1
    END AS pair_key

FROM train_tokens

WHERE token1 <> ''
  AND token2 <> ''
""")

con.execute("""
CREATE OR REPLACE TEMP TABLE val_pair12 AS

SELECT
    source1_entity_id,
    country,

    CASE
        WHEN token1 <= token2
        THEN token1 || '|' || token2
        ELSE token2 || '|' || token1
    END AS pair_key

FROM val_s1_tokens

WHERE token1 <> ''
  AND token2 <> ''
""")

# Generate candidates
con.execute("""
CREATE OR REPLACE TEMP TABLE pair_candidates AS

SELECT DISTINCT
    s.source1_entity_id,
    t.candidate_id

FROM val_pair12 s

JOIN train_pair12 t
    ON s.country = t.country
   AND s.pair_key = t.pair_key
""")

print("✅ Two-token candidates generated")

print(
    con.execute("""
        SELECT
            COUNT(*) AS candidate_pairs,
            COUNT(DISTINCT source1_entity_id) AS s1_count
        FROM pair_candidates
    """).fetchdf()
)
```

```python
# ==========================================
# RECALL OF TWO-TOKEN BLOCKING
# ==========================================

result = con.execute("""
WITH gt_pairs AS (
    SELECT
        source1_entity_id,
        trim(x) AS candidate_id
    FROM val_gt_100,
    UNNEST(
        string_split(
            COALESCE(matched_entity_ids, ''),
            ','
        )
    ) AS t(x)
    WHERE trim(x) <> ''
),

captured_pairs AS (
    SELECT
        g.source1_entity_id,
        g.candidate_id
    FROM gt_pairs g
    INNER JOIN pair_candidates c
        ON g.source1_entity_id = c.source1_entity_id
       AND g.candidate_id = c.candidate_id
)

SELECT
    (SELECT COUNT(*) FROM gt_pairs) AS total_true_matches,
    (SELECT COUNT(*) FROM captured_pairs) AS captured_true_matches,
    ROUND(
        100.0 * (SELECT COUNT(*) FROM captured_pairs)
        / NULLIF((SELECT COUNT(*) FROM gt_pairs), 0),
        2
    ) AS blocking_recall_percent
""").fetchdf()

print(result)
```

```python
# ==========================================
# FIND MISSED TRUE MATCHES
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE missed_pairs AS

WITH gt_pairs AS (
    SELECT
        source1_entity_id,
        trim(x) AS candidate_id
    FROM val_gt_100,
    UNNEST(
        string_split(
            COALESCE(matched_entity_ids, ''),
            ','
        )
    ) AS t(x)
    WHERE trim(x) <> ''
)

SELECT
    g.source1_entity_id,
    g.candidate_id
FROM gt_pairs g

LEFT JOIN pair_candidates c
    ON g.source1_entity_id = c.source1_entity_id
   AND g.candidate_id = c.candidate_id

WHERE c.candidate_id IS NULL
""")

print(
    con.execute("""
        SELECT COUNT(*) AS missed_true_matches
        FROM missed_pairs
    """).fetchdf()
)

print("\nSample missed pairs:")
print(
    con.execute("""
        SELECT *
        FROM missed_pairs
        LIMIT 20
    """).fetchdf()
)
```

```python
# ==========================================
# INSPECT MISSED MATCHES
# ==========================================

result = con.execute("""
WITH missed AS (
    SELECT *
    FROM missed_pairs
    LIMIT 20
),

s1 AS (
    SELECT
        entity_id,
        business_name,
        business_address,
        country
    FROM read_csv(
        'train_source1.tsv',
        delim='\t',
        header=true
    )
),

s2 AS (
    SELECT
        entity_id,
        business_name,
        business_address,
        country
    FROM read_csv(
        'train_source2.tsv',
        delim='\t',
        header=true
    )
),

s3 AS (
    SELECT
        entity_id,
        business_name,
        business_address,
        country
    FROM read_csv(
        'train_source3.tsv',
        delim='\t',
        header=true
    )
),

candidates AS (
    SELECT * FROM s2
    UNION ALL
    SELECT * FROM s3
)

SELECT
    m.source1_entity_id,
    m.candidate_id,

    s1.business_name AS s1_name,
    c.business_name AS candidate_name,

    s1.business_address AS s1_address,
    c.business_address AS candidate_address,

    s1.country AS s1_country,
    c.country AS candidate_country

FROM missed m
JOIN s1
    ON m.source1_entity_id = s1.entity_id
JOIN candidates c
    ON m.candidate_id = c.entity_id
""").fetchdf()

print(result.to_string(index=False))
```

```python
# ==========================================
# ADDRESS BLOCKING KEYS
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE train_address_keys AS

SELECT
    entity_id AS candidate_id,
    country,

    -- first numeric part of address
    regexp_extract(
        coalesce(business_address, ''),
        '[0-9]+',
        0
    ) AS address_number,

    -- first address token
    substr(
        split_part(
            lower(
                regexp_replace(
                    coalesce(business_address, ''),
                    '[^\\p{L}\\p{N}]+',
                    ' ',
                    'g'
                )
            ),
            ' ',
            1
        ),
        1, 5
    ) AS addr_token1,

    -- second address token
    substr(
        split_part(
            lower(
                regexp_replace(
                    coalesce(business_address, ''),
                    '[^\\p{L}\\p{N}]+',
                    ' ',
                    'g'
                )
            ),
            ' ',
            2
        ),
        1, 5
    ) AS addr_token2

FROM read_csv(
    'train_source2.tsv',
    delim='\t',
    header=true
)

UNION ALL

SELECT
    entity_id AS candidate_id,
    country,

    regexp_extract(
        coalesce(business_address, ''),
        '[0-9]+',
        0
    ) AS address_number,

    substr(
        split_part(
            lower(
                regexp_replace(
                    coalesce(business_address, ''),
                    '[^\\p{L}\\p{N}]+',
                    ' ',
                    'g'
                )
            ),
            ' ',
            1
        ),
        1, 5
    ) AS addr_token1,

    substr(
        split_part(
            lower(
                regexp_replace(
                    coalesce(business_address, ''),
                    '[^\\p{L}\\p{N}]+',
                    ' ',
                    'g'
                )
            ),
            ' ',
            2
        ),
        1, 5
    ) AS addr_token2

FROM read_csv(
    'train_source3.tsv',
    delim='\t',
    header=true
)
""")

print("✅ Training address keys created")
```

```python
# ==========================================
# ADDRESS KEYS FOR 100 VALIDATION S1
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE val_address_keys AS

SELECT
    entity_id AS source1_entity_id,
    country,

    regexp_extract(
        coalesce(business_address, ''),
        '[0-9]+',
        0
    ) AS address_number,

    substr(
        split_part(
            lower(
                regexp_replace(
                    coalesce(business_address, ''),
                    '[^\\p{L}\\p{N}]+',
                    ' ',
                    'g'
                )
            ),
            ' ',
            1
        ),
        1, 5
    ) AS addr_token1,

    substr(
        split_part(
            lower(
                regexp_replace(
                    coalesce(business_address, ''),
                    '[^\\p{L}\\p{N}]+',
                    ' ',
                    'g'
                )
            ),
            ' ',
            2
        ),
        1, 5
    ) AS addr_token2

FROM read_csv(
    'train_source1.tsv',
    delim='\t',
    header=true
)

WHERE entity_id IN (
    SELECT source1_entity_id
    FROM val_gt_100
)
""")

print(
    con.execute("""
        SELECT COUNT(*) AS validation_s1
        FROM val_address_keys
    """).fetchdf()
)
```

```python
# ==========================================
# ADDRESS KEY FREQUENCIES
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE train_address_freq AS

SELECT
    country,
    key_type,
    key_value,
    COUNT(*) AS freq
FROM (
    SELECT
        country,
        'number' AS key_type,
        address_number AS key_value
    FROM train_address_keys
    WHERE address_number <> ''

    UNION ALL

    SELECT
        country,
        'token1' AS key_type,
        addr_token1 AS key_value
    FROM train_address_keys
    WHERE addr_token1 <> ''

    UNION ALL

    SELECT
        country,
        'token2' AS key_type,
        addr_token2 AS key_value
    FROM train_address_keys
    WHERE addr_token2 <> ''
)
GROUP BY country, key_type, key_value
""")

print("✅ Address frequency table created")

print(
    con.execute("""
        SELECT
            key_type,
            COUNT(*) AS unique_keys,
            AVG(freq) AS avg_freq,
            MAX(freq) AS max_freq
        FROM train_address_freq
        GROUP BY key_type
        ORDER BY key_type
    """).fetchdf()
)
```

```python
# ==========================================
# CHECK ADDRESS RECOVERY ON 133 MISSES
# ==========================================

result = con.execute("""
SELECT
    COUNT(*) AS missed_pairs,

    SUM(
        CASE
            WHEN
                (
                    a.address_number <> ''
                    AND b.address_number <> ''
                    AND try_cast(a.address_number AS BIGINT)
                        = try_cast(b.address_number AS BIGINT)
                )
                OR
                (
                    a.addr_token1 <> ''
                    AND a.addr_token1 = b.addr_token1
                )
                OR
                (
                    a.addr_token2 <> ''
                    AND a.addr_token2 = b.addr_token2
                )
            THEN 1
            ELSE 0
        END
    ) AS recovered_by_address,

    SUM(
        CASE
            WHEN
                a.address_number <> ''
                AND b.address_number <> ''
                AND try_cast(a.address_number AS BIGINT)
                    = try_cast(b.address_number AS BIGINT)
            THEN 1 ELSE 0
        END
    ) AS number_matches,

    SUM(
        CASE
            WHEN a.addr_token1 <> ''
                 AND a.addr_token1 = b.addr_token1
            THEN 1 ELSE 0
        END
    ) AS token1_matches,

    SUM(
        CASE
            WHEN a.addr_token2 <> ''
                 AND a.addr_token2 = b.addr_token2
            THEN 1 ELSE 0
        END
    ) AS token2_matches

FROM missed_pairs m

JOIN val_address_keys a
    ON m.source1_entity_id = a.source1_entity_id

JOIN train_address_keys b
    ON m.candidate_id = b.candidate_id
""").fetchdf()

print(result)
```

```python
# ==========================================
# FREQUENCY OF ADDRESS KEYS IN MISSED MATCHES
# ==========================================

result = con.execute("""
SELECT
    COUNT(*) AS missed_pairs,

    AVG(fnum.freq) AS avg_number_freq,
    MAX(fnum.freq) AS max_number_freq,

    AVG(ft1.freq) AS avg_token1_freq,
    MAX(ft1.freq) AS max_token1_freq,

    AVG(ft2.freq) AS avg_token2_freq,
    MAX(ft2.freq) AS max_token2_freq

FROM missed_pairs m

JOIN val_address_keys a
    ON m.source1_entity_id = a.source1_entity_id

JOIN train_address_keys b
    ON m.candidate_id = b.candidate_id

LEFT JOIN train_address_freq fnum
    ON b.country = fnum.country
   AND fnum.key_type = 'number'
   AND b.address_number = fnum.key_value

LEFT JOIN train_address_freq ft1
    ON b.country = ft1.country
   AND ft1.key_type = 'token1'
   AND b.addr_token1 = ft1.key_value

LEFT JOIN train_address_freq ft2
    ON b.country = ft2.country
   AND ft2.key_type = 'token2'
   AND b.addr_token2 = ft2.key_value
""").fetchdf()

print(result)
```

```python
# ==========================================
# COMPOUND ADDRESS KEY
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE train_addr_compound AS

SELECT
    candidate_id,
    country,

    CASE
        WHEN address_number <> '' AND addr_token1 <> ''
        THEN ltrim(address_number, '0') || '|' || addr_token1
        ELSE ''
    END AS compound_key

FROM train_address_keys
WHERE address_number <> ''
  AND addr_token1 <> ''
""")

con.execute("""
CREATE OR REPLACE TEMP TABLE val_addr_compound AS

SELECT
    source1_entity_id,
    country,

    CASE
        WHEN address_number <> '' AND addr_token1 <> ''
        THEN ltrim(address_number, '0') || '|' || addr_token1
        ELSE ''
    END AS compound_key

FROM val_address_keys
WHERE address_number <> ''
  AND addr_token1 <> ''
""")

# Frequency of compound keys
con.execute("""
CREATE OR REPLACE TEMP TABLE compound_freq AS

SELECT
    country,
    compound_key,
    COUNT(*) AS freq
FROM train_addr_compound
WHERE compound_key <> ''
GROUP BY country, compound_key
""")

print(
    con.execute("""
        SELECT
            COUNT(*) AS unique_compound_keys,
            AVG(freq) AS avg_freq,
            MAX(freq) AS max_freq
        FROM compound_freq
    """).fetchdf()
)
```

```python
# ==========================================
# COMPOUND ADDRESS RECOVERY
# ==========================================

result = con.execute("""
SELECT
    COUNT(*) AS missed_pairs,

    SUM(
        CASE
            WHEN a.compound_key <> ''
                 AND b.compound_key <> ''
                 AND a.compound_key = b.compound_key
                 AND f.freq <= 50
            THEN 1
            ELSE 0
        END
    ) AS recovered_with_rare_compound

FROM missed_pairs m

JOIN val_addr_compound a
    ON m.source1_entity_id = a.source1_entity_id

JOIN train_addr_compound b
    ON m.candidate_id = b.candidate_id

LEFT JOIN compound_freq f
    ON b.country = f.country
   AND b.compound_key = f.compound_key
""").fetchdf()

print(result)
```

```python
# ==========================================
# BETTER ADDRESS KEY
# number + first alphabetic street token
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE train_address_better AS

SELECT
    candidate_id,
    country,

    ltrim(address_number, '0') AS number_clean,

    lower(
        regexp_extract(
            lower(
                regexp_replace(
                    coalesce(business_address, ''),
                    '[^\\p{L}\\p{N}]+',
                    ' ',
                    'g'
                )
            ),
            '[^0-9 ]+',
            0
        )
    ) AS street_token

FROM (
    SELECT
        entity_id AS candidate_id,
        country,
        business_address,
        regexp_extract(
            coalesce(business_address, ''),
            '[0-9]+',
            0
        ) AS address_number
    FROM read_csv(
        'train_source2.tsv',
        delim='\t',
        header=true
    )

    UNION ALL

    SELECT
        entity_id AS candidate_id,
        country,
        business_address,
        regexp_extract(
            coalesce(business_address, ''),
            '[0-9]+',
            0
        ) AS address_number
    FROM read_csv(
        'train_source3.tsv',
        delim='\t',
        header=true
    )
)
""")

print("✅ Better address keys created")
```

```python
# ==========================================
# BETTER ADDRESS KEY FOR VALIDATION S1
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE val_address_better AS

SELECT
    entity_id AS source1_entity_id,
    country,

    ltrim(
        regexp_extract(
            coalesce(business_address, ''),
            '[0-9]+',
            0
        ),
        '0'
    ) AS number_clean,

    lower(
        regexp_extract(
            lower(
                regexp_replace(
                    coalesce(business_address, ''),
                    '[^\\p{L}\\p{N}]+',
                    ' ',
                    'g'
                )
            ),
            '[^0-9 ]+',
            0
        )
    ) AS street_token

FROM read_csv(
    'train_source1.tsv',
    delim='\t',
    header=true
)

WHERE entity_id IN (
    SELECT source1_entity_id
    FROM val_gt_100
)
""")

print(
    con.execute("""
        SELECT COUNT(*) AS validation_s1
        FROM val_address_better
    """).fetchdf()
)
```

```python
# ==========================================
# TEST BETTER ADDRESS KEY
# ==========================================

result = con.execute("""
SELECT
    COUNT(*) AS missed_pairs,

    SUM(
        CASE
            WHEN a.number_clean <> ''
             AND a.street_token <> ''
             AND b.number_clean <> ''
             AND b.street_token <> ''
             AND a.number_clean = b.number_clean
             AND a.street_token = b.street_token
            THEN 1
            ELSE 0
        END
    ) AS recovered_by_better_address

FROM missed_pairs m

JOIN val_address_better a
    ON m.source1_entity_id = a.source1_entity_id

JOIN train_address_better b
    ON m.candidate_id = b.candidate_id
   AND a.country = b.country
""").fetchdf()

print(result)
```

```python
# ==========================================
# COMBINE NAME + ADDRESS BLOCKING
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE address_candidates AS

SELECT DISTINCT
    a.source1_entity_id,
    b.candidate_id

FROM val_address_better a

JOIN train_address_better b
    ON a.country = b.country
   AND a.number_clean <> ''
   AND a.street_token <> ''
   AND b.number_clean <> ''
   AND b.street_token <> ''
   AND a.number_clean = b.number_clean
   AND a.street_token = b.street_token
""")

print("✅ Address candidates generated")

print(
    con.execute("""
        SELECT
            COUNT(*) AS address_candidate_pairs,
            COUNT(DISTINCT source1_entity_id) AS s1_count
        FROM address_candidates
    """).fetchdf()
)
```

```python
# ==========================================
# COMBINE NAME + ADDRESS CANDIDATES
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE combined_candidates AS

SELECT
    source1_entity_id,
    candidate_id
FROM pair_candidates

UNION

SELECT
    source1_entity_id,
    candidate_id
FROM address_candidates
""")

result = con.execute("""
WITH gt_pairs AS (
    SELECT
        source1_entity_id,
        trim(x) AS candidate_id
    FROM val_gt_100,
    UNNEST(
        string_split(
            COALESCE(matched_entity_ids, ''),
            ','
        )
    ) AS t(x)
    WHERE trim(x) <> ''
),

captured AS (
    SELECT
        g.source1_entity_id,
        g.candidate_id
    FROM gt_pairs g
    INNER JOIN combined_candidates c
        ON g.source1_entity_id = c.source1_entity_id
       AND g.candidate_id = c.candidate_id
)

SELECT
    (SELECT COUNT(*) FROM gt_pairs) AS total_true_matches,

    (SELECT COUNT(*) FROM captured) AS captured_true_matches,

    ROUND(
        100.0 *
        (SELECT COUNT(*) FROM captured)
        / NULLIF((SELECT COUNT(*) FROM gt_pairs), 0),
        2
    ) AS blocking_recall_percent,

    (SELECT COUNT(*) FROM combined_candidates)
        AS total_candidate_pairs,

    (SELECT COUNT(DISTINCT source1_entity_id)
     FROM combined_candidates)
        AS s1_with_candidates
""").fetchdf()

print(result)
```

```python
# ==========================================
# FIND REMAINING 61 MISSES
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE remaining_misses AS

WITH gt_pairs AS (
    SELECT
        source1_entity_id,
        trim(x) AS candidate_id
    FROM val_gt_100,
    UNNEST(
        string_split(
            COALESCE(matched_entity_ids, ''),
            ','
        )
    ) AS t(x)
    WHERE trim(x) <> ''
)

SELECT
    g.source1_entity_id,
    g.candidate_id
FROM gt_pairs g

LEFT JOIN combined_candidates c
    ON g.source1_entity_id = c.source1_entity_id
   AND g.candidate_id = c.candidate_id

WHERE c.candidate_id IS NULL
""")

# ==========================================
# TEST ADDRESS NUMBER ONLY
# ==========================================

result = con.execute("""
SELECT
    COUNT(*) AS remaining_misses,

    SUM(
        CASE
            WHEN a.address_number <> ''
             AND b.address_number <> ''
             AND try_cast(a.address_number AS BIGINT)
                 = try_cast(b.address_number AS BIGINT)
            THEN 1
            ELSE 0
        END
    ) AS recoverable_by_number

FROM remaining_misses m

JOIN val_address_keys a
    ON m.source1_entity_id = a.source1_entity_id

JOIN train_address_keys b
    ON m.candidate_id = b.candidate_id
""").fetchdf()

print(result)
```

```python
# ==========================================
# CHECK ADDRESS NUMBER FREQUENCY
# ==========================================

result = con.execute("""
SELECT
    COUNT(*) AS remaining_misses,

    SUM(CASE WHEN f.freq <= 10 THEN 1 ELSE 0 END)
        AS freq_le_10,

    SUM(CASE WHEN f.freq <= 50 THEN 1 ELSE 0 END)
        AS freq_le_50,

    SUM(CASE WHEN f.freq <= 100 THEN 1 ELSE 0 END)
        AS freq_le_100,

    SUM(CASE WHEN f.freq <= 500 THEN 1 ELSE 0 END)
        AS freq_le_500,

    SUM(CASE WHEN f.freq <= 1000 THEN 1 ELSE 0 END)
        AS freq_le_1000,

    SUM(CASE WHEN f.freq <= 5000 THEN 1 ELSE 0 END)
        AS freq_le_5000

FROM remaining_misses m

JOIN val_address_keys a
    ON m.source1_entity_id = a.source1_entity_id

JOIN train_address_keys b
    ON m.candidate_id = b.candidate_id

LEFT JOIN train_address_freq f
    ON b.country = f.country
   AND f.key_type = 'number'
   AND b.address_number = f.key_value

WHERE a.address_number <> ''
  AND b.address_number <> ''
  AND try_cast(a.address_number AS BIGINT)
      = try_cast(b.address_number AS BIGINT)
""").fetchdf()

print(result)
```

```python
# ==========================================
# SELECTIVE ADDRESS NUMBER BLOCKING
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE train_number_rare AS

SELECT
    b.candidate_id,
    b.country,
    ltrim(b.address_number, '0') AS number_clean

FROM train_address_keys b

JOIN (
    SELECT
        country,
        ltrim(address_number, '0') AS number_clean,
        COUNT(*) AS freq

    FROM train_address_keys

    WHERE address_number <> ''

    GROUP BY
        country,
        ltrim(address_number, '0')

    HAVING COUNT(*) <= 500
) f

ON b.country = f.country
AND ltrim(b.address_number, '0') = f.number_clean

WHERE b.address_number <> ''
""")

con.execute("""
CREATE OR REPLACE TEMP TABLE number_candidates AS

SELECT DISTINCT
    a.source1_entity_id,
    b.candidate_id

FROM val_address_keys a

JOIN train_number_rare b
    ON a.country = b.country
   AND ltrim(a.address_number, '0') = b.number_clean

WHERE a.address_number <> ''
""")

print("✅ Selective number candidates generated")

print(
    con.execute("""
        SELECT
            COUNT(*) AS candidate_pairs,
            COUNT(DISTINCT source1_entity_id) AS s1_count
        FROM number_candidates
    """).fetchdf()
)
```

```python
# ==========================================
# COMBINE NAME + ADDRESS + NUMBER BLOCKING
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE combined_candidates_v2 AS

SELECT source1_entity_id, candidate_id
FROM pair_candidates

UNION

SELECT source1_entity_id, candidate_id
FROM address_candidates

UNION

SELECT source1_entity_id, candidate_id
FROM number_candidates
""")

result = con.execute("""
WITH gt_pairs AS (
    SELECT
        source1_entity_id,
        trim(x) AS candidate_id
    FROM val_gt_100,
    UNNEST(
        string_split(
            COALESCE(matched_entity_ids, ''),
            ','
        )
    ) AS t(x)
    WHERE trim(x) <> ''
),

captured AS (
    SELECT
        g.source1_entity_id,
        g.candidate_id
    FROM gt_pairs g
    INNER JOIN combined_candidates_v2 c
        ON g.source1_entity_id = c.source1_entity_id
       AND g.candidate_id = c.candidate_id
)

SELECT
    (SELECT COUNT(*) FROM gt_pairs) AS total_true_matches,
    (SELECT COUNT(*) FROM captured) AS captured_true_matches,

    ROUND(
        100.0 *
        (SELECT COUNT(*) FROM captured)
        / NULLIF((SELECT COUNT(*) FROM gt_pairs), 0),
        2
    ) AS blocking_recall_percent,

    (SELECT COUNT(*) FROM combined_candidates_v2)
        AS total_candidate_pairs,

    (SELECT COUNT(DISTINCT source1_entity_id)
     FROM combined_candidates_v2)
        AS s1_with_candidates
""").fetchdf()

print(result)
```

```python
# ==========================================
# INSPECT REMAINING MISSED MATCHES
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE remaining_misses_v2 AS

WITH gt_pairs AS (
    SELECT
        source1_entity_id,
        trim(x) AS candidate_id
    FROM val_gt_100,
    UNNEST(
        string_split(
            COALESCE(matched_entity_ids, ''),
            ','
        )
    ) AS t(x)
    WHERE trim(x) <> ''
)

SELECT
    g.source1_entity_id,
    g.candidate_id
FROM gt_pairs g

LEFT JOIN combined_candidates_v2 c
    ON g.source1_entity_id = c.source1_entity_id
   AND g.candidate_id = c.candidate_id

WHERE c.candidate_id IS NULL
""")

print(
    con.execute("""
        SELECT COUNT(*) AS remaining_misses
        FROM remaining_misses_v2
    """).fetchdf()
)

print("\nSample remaining misses:")

result = con.execute("""
SELECT
    m.source1_entity_id,
    m.candidate_id,

    s1.business_name AS s1_name,
    c.business_name AS candidate_name,

    s1.business_address AS s1_address,
    c.business_address AS candidate_address,

    s1.country AS country
FROM remaining_misses_v2 m

JOIN read_csv(
    'train_source1.tsv',
    delim='\t',
    header=true
) s1
    ON m.source1_entity_id = s1.entity_id

JOIN (
    SELECT entity_id, business_name, business_address, country
    FROM read_csv(
        'train_source2.tsv',
        delim='\t',
        header=true
    )

    UNION ALL

    SELECT entity_id, business_name, business_address, country
    FROM read_csv(
        'train_source3.tsv',
        delim='\t',
        header=true
    )
) c
    ON m.candidate_id = c.entity_id

LIMIT 20
""").fetchdf()

print(result.to_string(index=False))
```

```python
# ==========================================
# SINGLE NAME TOKEN BLOCKING (FREQ <= 500)
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE single_token_500_candidates AS

SELECT DISTINCT
    s.source1_entity_id,
    t.candidate_id

FROM val_s1_tokens s

JOIN train_tokens t
    ON s.country = t.country
   AND s.token1 <> ''
   AND s.token1 = t.token1

JOIN train_token_freq f
    ON t.country = f.country
   AND t.token1 = f.token
WHERE f.freq <= 500

UNION

SELECT DISTINCT
    s.source1_entity_id,
    t.candidate_id

FROM val_s1_tokens s

JOIN train_tokens t
    ON s.country = t.country
   AND s.token2 <> ''
   AND s.token2 = t.token2

JOIN train_token_freq f
    ON t.country = f.country
   AND t.token2 = f.token
WHERE f.freq <= 500

UNION

SELECT DISTINCT
    s.source1_entity_id,
    t.candidate_id

FROM val_s1_tokens s

JOIN train_tokens t
    ON s.country = t.country
   AND s.token3 <> ''
   AND s.token3 = t.token3

JOIN train_token_freq f
    ON t.country = f.country
   AND t.token3 = f.token
WHERE f.freq <= 500
""")

print("✅ Single-token candidates generated")

print(
    con.execute("""
        SELECT
            COUNT(*) AS candidate_pairs,
            COUNT(DISTINCT source1_entity_id) AS s1_count
        FROM single_token_500_candidates
    """).fetchdf()
)
```

```python
# ==========================================
# COMBINE + CHECK RECALL
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE combined_candidates_v3 AS

SELECT source1_entity_id, candidate_id
FROM combined_candidates_v2

UNION

SELECT source1_entity_id, candidate_id
FROM single_token_500_candidates
""")

result = con.execute("""
WITH gt_pairs AS (
    SELECT
        source1_entity_id,
        trim(x) AS candidate_id
    FROM val_gt_100,
    UNNEST(
        string_split(
            COALESCE(matched_entity_ids, ''),
            ','
        )
    ) AS t(x)
    WHERE trim(x) <> ''
),

captured AS (
    SELECT
        g.source1_entity_id,
        g.candidate_id
    FROM gt_pairs g
    JOIN combined_candidates_v3 c
        ON g.source1_entity_id = c.source1_entity_id
       AND g.candidate_id = c.candidate_id
)

SELECT
    COUNT(*) AS total_true_matches,
    COUNT(captured.candidate_id) AS captured_true_matches,

    ROUND(
        100.0 * COUNT(captured.candidate_id)
        / COUNT(*),
        2
    ) AS blocking_recall_percent,

    (
        SELECT COUNT(*)
        FROM combined_candidates_v3
    ) AS total_candidate_pairs

FROM gt_pairs
LEFT JOIN captured
    ON gt_pairs.source1_entity_id = captured.source1_entity_id
   AND gt_pairs.candidate_id = captured.candidate_id
""").fetchdf()

print(result)
```

```python
# ==========================================
# ADDRESS NUMBER BLOCKING - FREQ <= 5000
# ==========================================

result = con.execute("""
WITH rare_numbers AS (
    SELECT
        country,
        ltrim(address_number, '0') AS number_clean,
        COUNT(*) AS freq
    FROM train_address_keys
    WHERE address_number <> ''
    GROUP BY
        country,
        ltrim(address_number, '0')
    HAVING COUNT(*) <= 5000
)

SELECT COUNT(DISTINCT
    a.source1_entity_id || '|' || b.candidate_id
) AS candidate_pairs

FROM val_address_keys a

JOIN rare_numbers r
    ON a.country = r.country
   AND ltrim(a.address_number, '0') = r.number_clean

JOIN train_address_keys b
    ON b.country = r.country
   AND ltrim(b.address_number, '0') = r.number_clean

WHERE a.address_number <> ''
""").fetchdf()

print(result)
```

```python
# ==========================================
# CHECK NUMBER FREQ <= 5000 RECOVERY
# ==========================================

result = con.execute("""
WITH rare_numbers AS (
    SELECT
        country,
        ltrim(address_number, '0') AS number_clean,
        COUNT(*) AS freq
    FROM train_address_keys
    WHERE address_number <> ''
    GROUP BY
        country,
        ltrim(address_number, '0')
    HAVING COUNT(*) <= 5000
)

SELECT
    COUNT(*) AS remaining_misses,

    COUNT(
        CASE
            WHEN ltrim(a.address_number, '0') = r.number_clean
             AND ltrim(b.address_number, '0') = r.number_clean
            THEN 1
        END
    ) AS recovered_by_number_5000

FROM remaining_misses_v2 m

JOIN val_address_keys a
    ON m.source1_entity_id = a.source1_entity_id

JOIN train_address_keys b
    ON m.candidate_id = b.candidate_id

LEFT JOIN rare_numbers r
    ON b.country = r.country
   AND ltrim(b.address_number, '0') = r.number_clean
""").fetchdf()

print(result)
```

```python
# ==========================================
# CHARACTER-LEVEL SIMILARITY ON 55 MISSES
# ==========================================

import pandas as pd
from rapidfuzz.fuzz import ratio, WRatio

# Get the 55 missed pairs with their names
missed_names = con.execute("""
WITH source_records AS (
    SELECT
        entity_id,
        business_name
    FROM read_csv(
        'train_source2.tsv',
        delim='\t',
        header=true
    )

    UNION ALL

    SELECT
        entity_id,
        business_name
    FROM read_csv(
        'train_source3.tsv',
        delim='\t',
        header=true
    )
)

SELECT
    m.source1_entity_id,
    m.candidate_id,
    s.business_name AS s1_name,
    c.business_name AS candidate_name
FROM remaining_misses_v2 m
JOIN val_s1_100 s
    ON m.source1_entity_id = s.entity_id
JOIN source_records c
    ON m.candidate_id = c.entity_id
""").fetchdf()

def clean_name(x):
    if pd.isna(x):
        return ""
    return str(x).lower().strip()

missed_names["name_ratio"] = missed_names.apply(
    lambda r: ratio(
        clean_name(r["s1_name"]),
        clean_name(r["candidate_name"])
    ),
    axis=1
)

missed_names["wratio"] = missed_names.apply(
    lambda r: WRatio(
        clean_name(r["s1_name"]),
        clean_name(r["candidate_name"])
    ),
    axis=1
)

print("✅ Character similarity calculated")
print(
    missed_names[
        ["source1_entity_id", "candidate_id",
         "s1_name", "candidate_name",
         "name_ratio", "wratio"]
    ].head(20).to_string(index=False)
)

print("\nSummary:")
print(missed_names[["name_ratio", "wratio"]].describe())
```

```python
# ==========================================
# CURRENT REMAINING MISSES
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE remaining_misses_v3 AS

WITH gt_pairs AS (
    SELECT
        source1_entity_id,
        trim(x) AS candidate_id
    FROM val_gt_100,
    UNNEST(
        string_split(
            COALESCE(matched_entity_ids, ''),
            ','
        )
    ) AS t(x)
    WHERE trim(x) <> ''
)

SELECT
    g.source1_entity_id,
    g.candidate_id
FROM gt_pairs g

LEFT JOIN combined_candidates_v3 c
    ON g.source1_entity_id = c.source1_entity_id
   AND g.candidate_id = c.candidate_id

WHERE c.candidate_id IS NULL
""")

# ==========================================
# TEST SINGLE TOKEN WITH FREQ <= 5000
# ==========================================

result = con.execute("""
SELECT
    COUNT(*) AS remaining_misses,

    SUM(
        CASE
            WHEN
                (
                    s.token1 <> ''
                    AND s.token1 = t.token1
                    AND f1.freq <= 5000
                )
                OR
                (
                    s.token1 <> ''
                    AND s.token1 = t.token2
                    AND f2.freq <= 5000
                )
                OR
                (
                    s.token1 <> ''
                    AND s.token1 = t.token3
                    AND f3.freq <= 5000
                )
                OR
                (
                    s.token2 <> ''
                    AND s.token2 = t.token1
                    AND f1.freq <= 5000
                )
                OR
                (
                    s.token2 <> ''
                    AND s.token2 = t.token2
                    AND f2.freq <= 5000
                )
                OR
                (
                    s.token2 <> ''
                    AND s.token2 = t.token3
                    AND f3.freq <= 5000
                )
                OR
                (
                    s.token3 <> ''
                    AND s.token3 = t.token1
                    AND f1.freq <= 5000
                )
                OR
                (
                    s.token3 <> ''
                    AND s.token3 = t.token2
                    AND f2.freq <= 5000
                )
                OR
                (
                    s.token3 <> ''
                    AND s.token3 = t.token3
                    AND f3.freq <= 5000
                )
            THEN 1
            ELSE 0
        END
    ) AS recoverable

FROM remaining_misses_v3 m

JOIN val_s1_tokens s
    ON m.source1_entity_id = s.source1_entity_id

JOIN train_tokens t
    ON m.candidate_id = t.candidate_id

LEFT JOIN train_token_freq f1
    ON t.country = f1.country
   AND t.token1 = f1.token

LEFT JOIN train_token_freq f2
    ON t.country = f2.country
   AND t.token2 = f2.token

LEFT JOIN train_token_freq f3
    ON t.country = f3.country
   AND t.token3 = f3.token
""").fetchdf()

print(result)
```

```python
# ==========================================
# TEST ADDRESS TOKEN 2 ON REMAINING MISSES
# ==========================================

result = con.execute("""
SELECT
    COUNT(*) AS remaining_misses,

    SUM(
        CASE
            WHEN a.addr_token2 <> ''
             AND b.addr_token2 <> ''
             AND a.addr_token2 = b.addr_token2
            THEN 1
            ELSE 0
        END
    ) AS recoverable_by_addr_token2

FROM remaining_misses_v3 m

JOIN val_address_keys a
    ON m.source1_entity_id = a.source1_entity_id

JOIN train_address_keys b
    ON m.candidate_id = b.candidate_id
   AND a.country = b.country
""").fetchdf()

print(result)
```

```python
import re
import pandas as pd

remaining_ids = con.execute("""
    SELECT source1_entity_id, candidate_id
    FROM remaining_misses_v3
""").fetchdf()

test = missed_names.merge(
    remaining_ids,
    on=["source1_entity_id", "candidate_id"],
    how="inner"
)

def tokens(text):
    if pd.isna(text):
        return set()

    text = str(text).lower()
    text = re.sub(
        r'[^a-z0-9\u0900-\u097F\u0D00-\u0D7F]+',
        ' ',
        text
    )

    return {
        x for x in text.split()
        if len(x) >= 3
    }

test["s1_tokens"] = test["s1_name"].apply(tokens)
test["candidate_tokens"] = test["candidate_name"].apply(tokens)

test["common_token"] = test.apply(
    lambda r: r["s1_tokens"] & r["candidate_tokens"],
    axis=1
)

print("Remaining misses:", len(test))
print(
    "With at least one common name token:",
    (test["common_token"].str.len() > 0).sum()
)

print("\nExamples:")
print(
    test[
        ["s1_name", "candidate_name", "common_token"]
    ].head(20).to_string(index=False)
)
```

```python
# ==========================================
# ORDER-INDEPENDENT NAME PAIRS
# ON 50 REMAINING MISSES
# ==========================================

from itertools import combinations

def word_pairs(text):
    words = tokens(text)

    return {
        tuple(sorted(pair))
        for pair in combinations(words, 2)
    }

test["s1_pairs"] = test["s1_name"].apply(word_pairs)
test["candidate_pairs"] = test["candidate_name"].apply(word_pairs)

test["common_pairs"] = test.apply(
    lambda r: r["s1_pairs"] & r["candidate_pairs"],
    axis=1
)

print("Remaining misses:", len(test))

print(
    "Recovered by order-independent 2-word pair:",
    (test["common_pairs"].str.len() > 0).sum()
)

print("\nExamples:")
print(
    test[
        ["s1_name", "candidate_name", "common_pairs"]
    ].head(20).to_string(index=False)
)
```

```python
# ==========================================
# ORDER-INDEPENDENT 2-WORD PAIR CANDIDATES
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE pair_candidates_v2 AS

WITH s AS (
    SELECT
        source1_entity_id,
        country,
        LEAST(token1, token2) AS p1a,
        GREATEST(token1, token2) AS p1b,
        LEAST(token1, token3) AS p2a,
        GREATEST(token1, token3) AS p2b,
        LEAST(token2, token3) AS p3a,
        GREATEST(token2, token3) AS p3b
    FROM val_s1_tokens
)

SELECT DISTINCT
    s.source1_entity_id,
    t.candidate_id
FROM s
JOIN train_tokens t
    ON s.country = t.country
   AND (
        (
            s.p1a <> ''
            AND s.p1b <> ''
            AND s.p1a = LEAST(t.token1, t.token2)
            AND s.p1b = GREATEST(t.token1, t.token2)
        )
        OR
        (
            s.p1a <> ''
            AND s.p1b <> ''
            AND s.p1a = LEAST(t.token1, t.token3)
            AND s.p1b = GREATEST(t.token1, t.token3)
        )
        OR
        (
            s.p1a <> ''
            AND s.p1b <> ''
            AND s.p1a = LEAST(t.token2, t.token3)
            AND s.p1b = GREATEST(t.token2, t.token3)
        )
        OR
        (
            s.p2a <> ''
            AND s.p2b <> ''
            AND s.p2a = LEAST(t.token1, t.token2)
            AND s.p2b = GREATEST(t.token1, t.token2)
        )
        OR
        (
            s.p2a <> ''
            AND s.p2b <> ''
            AND s.p2a = LEAST(t.token1, t.token3)
            AND s.p2b = GREATEST(t.token1, t.token3)
        )
        OR
        (
            s.p2a <> ''
            AND s.p2b <> ''
            AND s.p2a = LEAST(t.token2, t.token3)
            AND s.p2b = GREATEST(t.token2, t.token3)
        )
        OR
        (
            s.p3a <> ''
            AND s.p3b <> ''
            AND s.p3a = LEAST(t.token1, t.token2)
            AND s.p3b = GREATEST(t.token1, t.token2)
        )
        OR
        (
            s.p3a <> ''
            AND s.p3b <> ''
            AND s.p3a = LEAST(t.token1, t.token3)
            AND s.p3b = GREATEST(t.token1, t.token3)
        )
        OR
        (
            s.p3a <> ''
            AND s.p3b <> ''
            AND s.p3a = LEAST(t.token2, t.token3)
            AND s.p3b = GREATEST(t.token2, t.token3)
        )
   )
""")

print("✅ Order-independent pair candidates created")

print(
    con.execute("""
        SELECT
            COUNT(*) AS candidate_pairs,
            COUNT(DISTINCT source1_entity_id) AS s1_count
        FROM pair_candidates_v2
    """).fetchdf()
)
```

```python
# ==========================================
# CHECK NAME PREFIX / SUFFIX ON 50 MISSES
# ==========================================

import re

def normalize_name_simple(text):
    if pd.isna(text):
        return ""
    text = str(text).lower()
    text = re.sub(r'[^a-z0-9]+', '', text)
    return text

test["s1_compact"] = test["s1_name"].apply(normalize_name_simple)
test["candidate_compact"] = test["candidate_name"].apply(normalize_name_simple)

test["same_prefix6"] = test.apply(
    lambda r:
        len(r["s1_compact"]) >= 6
        and len(r["candidate_compact"]) >= 6
        and r["s1_compact"][:6] == r["candidate_compact"][:6],
    axis=1
)

test["same_suffix6"] = test.apply(
    lambda r:
        len(r["s1_compact"]) >= 6
        and len(r["candidate_compact"]) >= 6
        and r["s1_compact"][-6:] == r["candidate_compact"][-6:],
    axis=1
)

print("Remaining misses:", len(test))
print("Same first 6 characters:", test["same_prefix6"].sum())
print("Same last 6 characters:", test["same_suffix6"].sum())
```

```python
# ==========================================
# TEST TRANSLITERATION ON 50 MISSES
# ==========================================

from unidecode import unidecode

def transliterated(text):
    if pd.isna(text):
        return ""
    
    return re.sub(
        r'[^a-z0-9]+',
        '',
        unidecode(str(text).lower())
    )

test["s1_translit"] = test["s1_name"].apply(transliterated)
test["candidate_translit"] = test["candidate_name"].apply(transliterated)

test["translit_ratio"] = test.apply(
    lambda r: ratio(
        r["s1_translit"],
        r["candidate_translit"]
    ),
    axis=1
)

print("Remaining misses:", len(test))

print(
    "Transliterated similarity >= 80:",
    (test["translit_ratio"] >= 80).sum()
)

print(
    "Transliterated similarity >= 70:",
    (test["translit_ratio"] >= 70).sum()
)

print("\nExamples:")
print(
    test[
        ["s1_name", "candidate_name", "translit_ratio"]
    ].head(20).to_string(index=False)
)
```

```python
# ==========================================
# RARE 2-WORD PAIR BLOCKING TEST
# ==========================================

# Frequency of each order-independent pair
con.execute("""
CREATE OR REPLACE TEMP TABLE pair12_freq AS
SELECT
    pair_key,
    COUNT(*) AS freq
FROM train_pair12
WHERE pair_key <> ''
GROUP BY pair_key
""")

# Test several frequency limits
for limit in [5, 10, 20, 50]:

    result = con.execute(f"""
    WITH candidates AS (

        SELECT DISTINCT
            s.source1_entity_id,
            t.candidate_id

        FROM val_pair12 s

        JOIN pair12_freq f
            ON s.pair_key = f.pair_key
           AND f.freq <= {limit}

        JOIN train_pair12 t
            ON t.pair_key = s.pair_key
    ),

    gt_pairs AS (
        SELECT
            source1_entity_id,
            trim(x) AS candidate_id
        FROM val_gt_100,
        UNNEST(
            string_split(
                COALESCE(matched_entity_ids, ''),
                ','
            )
        ) AS t(x)
        WHERE trim(x) <> ''
    ),

    captured AS (
        SELECT
            g.source1_entity_id,
            g.candidate_id
        FROM gt_pairs g
        JOIN candidates c
            ON g.source1_entity_id = c.source1_entity_id
           AND g.candidate_id = c.candidate_id
    )

    SELECT
        {limit} AS max_pair_frequency,
        (SELECT COUNT(*) FROM candidates) AS candidate_pairs,
        (SELECT COUNT(*) FROM captured) AS captured_true_matches,
        ROUND(
            100.0 *
            (SELECT COUNT(*) FROM captured)
            / 352,
            2
        ) AS recall_percent
    """).fetchdf()

    print(result)
```

```python
# ==========================================
# NAME TOKEN + ADDRESS NUMBER
# RECOVERY TEST ON 50 MISSES
# ==========================================

result = con.execute("""
SELECT
    COUNT(*) AS remaining_misses,

    SUM(
        CASE
            WHEN
                a.address_number <> ''
                AND b.address_number <> ''
                AND try_cast(a.address_number AS BIGINT)
                    = try_cast(b.address_number AS BIGINT)

                AND (
                    s.token1 = t.token1
                    OR s.token1 = t.token2
                    OR s.token1 = t.token3
                    OR s.token2 = t.token1
                    OR s.token2 = t.token2
                    OR s.token2 = t.token3
                    OR s.token3 = t.token1
                    OR s.token3 = t.token2
                    OR s.token3 = t.token3
                )

            THEN 1
            ELSE 0
        END
    ) AS recoverable

FROM remaining_misses_v3 m

JOIN val_address_keys a
    ON m.source1_entity_id = a.source1_entity_id

JOIN train_address_keys b
    ON m.candidate_id = b.candidate_id
   AND a.country = b.country

JOIN val_s1_tokens s
    ON m.source1_entity_id = s.source1_entity_id

JOIN train_tokens t
    ON m.candidate_id = t.candidate_id
""").fetchdf()

print(result)
```

```python
# ==========================================
# POSTAL / PIN CODE RECOVERY
# ==========================================

result = con.execute("""
WITH s1_addr AS (
    SELECT
        entity_id AS source1_entity_id,
        business_address
    FROM read_csv(
        'train_source1.tsv',
        delim='\t',
        header=true
    )
),

candidate_addr AS (
    SELECT
        entity_id AS candidate_id,
        business_address
    FROM read_csv(
        'train_source2.tsv',
        delim='\t',
        header=true
    )

    UNION ALL

    SELECT
        entity_id AS candidate_id,
        business_address
    FROM read_csv(
        'train_source3.tsv',
        delim='\t',
        header=true
    )
)

SELECT
    COUNT(*) AS remaining_misses,

    SUM(
        CASE
            WHEN
                regexp_extract(
                    coalesce(a.business_address, ''),
                    '[0-9]{5,6}',
                    0
                ) <> ''

                AND regexp_extract(
                    coalesce(b.business_address, ''),
                    '[0-9]{5,6}',
                    0
                ) <> ''

                AND regexp_extract(
                    coalesce(a.business_address, ''),
                    '[0-9]{5,6}',
                    0
                )
                =
                regexp_extract(
                    coalesce(b.business_address, ''),
                    '[0-9]{5,6}',
                    0
                )

            THEN 1
            ELSE 0
        END
    ) AS recovered_by_postal

FROM remaining_misses_v3 m

JOIN s1_addr a
    ON m.source1_entity_id = a.source1_entity_id

JOIN candidate_addr b
    ON m.candidate_id = b.candidate_id
""").fetchdf()

print(result)
```

```python
# ==========================================
# MEANINGFUL ADDRESS TOKEN TEST
# ==========================================

import re

def address_words(text):
    if pd.isna(text):
        return []

    text = str(text).lower()

    # Keep words, remove punctuation
    words = re.findall(r'[^\W\d_]+', text, flags=re.UNICODE)

    # Ignore very short words
    return [w for w in words if len(w) >= 3]


# Load only the address information needed for the 50 missed pairs
addr_data = con.execute("""
WITH source_records AS (
    SELECT
        entity_id,
        business_address
    FROM read_csv(
        'train_source2.tsv',
        delim='\t',
        header=true
    )

    UNION ALL

    SELECT
        entity_id,
        business_address
    FROM read_csv(
        'train_source3.tsv',
        delim='\t',
        header=true
    )
)

SELECT
    m.source1_entity_id,
    m.candidate_id,

    s.business_address AS s1_address,
    c.business_address AS candidate_address

FROM remaining_misses_v3 m

JOIN read_csv(
    'train_source1.tsv',
    delim='\t',
    header=true
) s
    ON m.source1_entity_id = s.entity_id

JOIN source_records c
    ON m.candidate_id = c.entity_id
""").fetchdf()


addr_data["s1_words"] = addr_data["s1_address"].apply(address_words)
addr_data["candidate_words"] = addr_data["candidate_address"].apply(address_words)

addr_data["common_words"] = addr_data.apply(
    lambda r: set(r["s1_words"]) & set(r["candidate_words"]),
    axis=1
)

print("Remaining misses:", len(addr_data))

print(
    "At least 2 common meaningful address words:",
    (addr_data["common_words"].str.len() >= 2).sum()
)

print(
    "At least 1 common meaningful address word:",
    (addr_data["common_words"].str.len() >= 1).sum()
)

print("\nExamples:")
print(
    addr_data[
        ["s1_address", "candidate_address", "common_words"]
    ].head(20).to_string(index=False)
)
```

```python
# ==========================================
# MEANINGFUL ADDRESS WORDS
# ==========================================

import re
import pandas as pd

def first_two_addr_words(text):
    if pd.isna(text):
        return ("", "")

    words = re.findall(
        r'[^\W\d_]+',
        str(text).lower(),
        flags=re.UNICODE
    )

    words = [w for w in words if len(w) >= 3]

    if len(words) >= 2:
        return (words[0], words[1])
    elif len(words) == 1:
        return (words[0], "")
    else:
        return ("", "")

# addr_data already contains the 100 validation S1 addresses
val_addr_words = addr_data[
    ["source1_entity_id", "s1_address"]
].drop_duplicates().copy()

val_addr_words[["addr_w1", "addr_w2"]] = val_addr_words[
    "s1_address"
].apply(
    lambda x: pd.Series(first_two_addr_words(x))
)

print("✅ Meaningful address keys prepared")

print(
    val_addr_words[
        ["source1_entity_id", "addr_w1", "addr_w2"]
    ].head(10).to_string(index=False)
)
```

```python
# ==========================================
# FIXED: VALIDATION ADDRESS PAIR
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE val_addr_pair AS

SELECT
    w.source1_entity_id,
    a.country,

    CASE
        WHEN w.addr_w1 <= w.addr_w2
        THEN w.addr_w1 || '|' || w.addr_w2
        ELSE w.addr_w2 || '|' || w.addr_w1
    END AS pair_key

FROM val_addr_words w

JOIN val_address_keys a
    ON w.source1_entity_id = a.source1_entity_id

WHERE w.addr_w1 <> ''
  AND w.addr_w2 <> ''
""")

print("✅ Validation address pairs created")

print(
    con.execute("""
        SELECT COUNT(*) AS rows
        FROM val_addr_pair
    """).fetchdf()
)
```

```python
# ==========================================
# 3-CHARACTER TOKEN PREFIX TEST
# ON 50 REMAINING MISSES
# ==========================================

def three_char_prefixes(text):
    if pd.isna(text):
        return set()

    words = re.findall(
        r'[^\W\d_]+',
        str(text).lower(),
        flags=re.UNICODE
    )

    return {
        w[:3]
        for w in words
        if len(w) >= 3
    }

test["s1_3prefix"] = test["s1_name"].apply(three_char_prefixes)
test["candidate_3prefix"] = test["candidate_name"].apply(three_char_prefixes)

test["common_3prefix"] = test.apply(
    lambda r: r["s1_3prefix"] & r["candidate_3prefix"],
    axis=1
)

print("Remaining misses:", len(test))

print(
    "At least 1 common 3-char prefix:",
    (test["common_3prefix"].str.len() >= 1).sum()
)

print(
    "At least 2 common 3-char prefixes:",
    (test["common_3prefix"].str.len() >= 2).sum()
)

print("\nExamples:")
print(
    test[
        ["s1_name", "candidate_name", "common_3prefix"]
    ].head(20).to_string(index=False)
)
```

```python
# ==========================================
# RECREATE TRAIN S2 + S3 TOKEN TABLE
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE train_tokens AS

SELECT
    entity_id AS candidate_id,
    country,

    substr(split_part(name_clean, ' ', 1), 1, 4) AS token1,
    substr(split_part(name_clean, ' ', 2), 1, 4) AS token2,
    substr(split_part(name_clean, ' ', 3), 1, 4) AS token3

FROM (
    SELECT
        entity_id,
        country,
        lower(
            regexp_replace(
                coalesce(business_name, ''),
                '[^\\p{L}\\p{N}]+',
                ' ',
                'g'
            )
        ) AS name_clean

    FROM read_csv(
        'train_source2.tsv',
        delim='\t',
        header=true
    )

    UNION ALL

    SELECT
        entity_id,
        country,
        lower(
            regexp_replace(
                coalesce(business_name, ''),
                '[^\\p{L}\\p{N}]+',
                ' ',
                'g'
            )
        ) AS name_clean

    FROM read_csv(
        'train_source3.tsv',
        delim='\t',
        header=true
    )
)
""")

print("✅ train_tokens recreated")

print(
    con.execute("""
        SELECT COUNT(*) AS total_candidates
        FROM train_tokens
    """).fetchdf()
)
```

```python
# ==========================================
# RECREATE VALIDATION S1 TOKENS
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE val_s1_tokens AS

SELECT
    entity_id AS source1_entity_id,
    country,

    substr(split_part(name_clean, ' ', 1), 1, 4) AS token1,
    substr(split_part(name_clean, ' ', 2), 1, 4) AS token2,
    substr(split_part(name_clean, ' ', 3), 1, 4) AS token3

FROM (
    SELECT
        entity_id,
        country,

        lower(
            regexp_replace(
                coalesce(business_name, ''),
                '[^\\p{L}\\p{N}]+',
                ' ',
                'g'
            )
        ) AS name_clean

    FROM read_csv(
        'train_source1.tsv',
        delim='\t',
        header=true
    )
)

WHERE entity_id IN (
    SELECT source1_entity_id
    FROM (
        SELECT
            g.source1_entity_id
        FROM read_csv(
            'train_ground_truth.tsv',
            delim='\t',
            header=true
        ) g
        USING SAMPLE reservoir(100 ROWS) REPEATABLE(42)
    )
)
""")

print("✅ Validation S1 tokens recreated")
```

```python
# ==========================================
# 3-CHARACTER PREFIX FREQUENCIES
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE train_prefix_freq AS

SELECT
    country,
    prefix,
    COUNT(*) AS freq
FROM (
    SELECT country, substr(token1, 1, 3) AS prefix
    FROM train_tokens
    WHERE token1 <> ''

    UNION ALL

    SELECT country, substr(token2, 1, 3) AS prefix
    FROM train_tokens
    WHERE token2 <> ''

    UNION ALL

    SELECT country, substr(token3, 1, 3) AS prefix
    FROM train_tokens
    WHERE token3 <> ''
)
GROUP BY country, prefix
""")

print("✅ 3-character prefix frequencies created")

print(
    con.execute("""
        SELECT
            COUNT(*) AS unique_prefixes,
            MIN(freq) AS min_freq,
            AVG(freq) AS avg_freq,
            MAX(freq) AS max_freq
        FROM train_prefix_freq
    """).fetchdf()
)
```

```python
# ==========================================
# 3-CHAR PREFIX BLOCKING - FREQ <= 50
# ==========================================

# Recreate ground truth for our current 100 validation S1
con.execute("""
CREATE OR REPLACE TEMP TABLE val_gt_100 AS
SELECT
    g.source1_entity_id,
    g.matched_entity_ids
FROM read_csv(
    'train_ground_truth.tsv',
    delim='\t',
    header=true
) g
JOIN (
    SELECT DISTINCT source1_entity_id
    FROM val_s1_tokens
) v
ON g.source1_entity_id = v.source1_entity_id
""")

# Generate candidates using each token position separately
con.execute("""
CREATE OR REPLACE TEMP TABLE prefix3_candidates AS

SELECT DISTINCT
    s.source1_entity_id,
    t.candidate_id
FROM val_s1_tokens s
JOIN train_tokens t
    ON s.country = t.country
   AND substr(s.token1, 1, 3) <> ''
   AND substr(s.token1, 1, 3) = substr(t.token1, 1, 3)
JOIN train_prefix_freq f
    ON t.country = f.country
   AND f.prefix = substr(t.token1, 1, 3)
   AND f.freq <= 50

UNION

SELECT DISTINCT
    s.source1_entity_id,
    t.candidate_id
FROM val_s1_tokens s
JOIN train_tokens t
    ON s.country = t.country
   AND substr(s.token1, 1, 3) <> ''
   AND substr(s.token1, 1, 3) = substr(t.token2, 1, 3)
JOIN train_prefix_freq f
    ON t.country = f.country
   AND f.prefix = substr(t.token2, 1, 3)
   AND f.freq <= 50

UNION

SELECT DISTINCT
    s.source1_entity_id,
    t.candidate_id
FROM val_s1_tokens s
JOIN train_tokens t
    ON s.country = t.country
   AND substr(s.token1, 1, 3) <> ''
   AND substr(s.token1, 1, 3) = substr(t.token3, 1, 3)
JOIN train_prefix_freq f
    ON t.country = f.country
   AND f.prefix = substr(t.token3, 1, 3)
   AND f.freq <= 50

UNION

SELECT DISTINCT
    s.source1_entity_id,
    t.candidate_id
FROM val_s1_tokens s
JOIN train_tokens t
    ON s.country = t.country
   AND substr(s.token2, 1, 3) <> ''
   AND substr(s.token2, 1, 3) = substr(t.token1, 1, 3)
JOIN train_prefix_freq f
    ON t.country = f.country
   AND f.prefix = substr(t.token1, 1, 3)
   AND f.freq <= 50

UNION

SELECT DISTINCT
    s.source1_entity_id,
    t.candidate_id
FROM val_s1_tokens s
JOIN train_tokens t
    ON s.country = t.country
   AND substr(s.token2, 1, 3) <> ''
   AND substr(s.token2, 1, 3) = substr(t.token2, 1, 3)
JOIN train_prefix_freq f
    ON t.country = f.country
   AND f.prefix = substr(t.token2, 1, 3)
   AND f.freq <= 50

UNION

SELECT DISTINCT
    s.source1_entity_id,
    t.candidate_id
FROM val_s1_tokens s
JOIN train_tokens t
    ON s.country = t.country
   AND substr(s.token2, 1, 3) <> ''
   AND substr(s.token2, 1, 3) = substr(t.token3, 1, 3)
JOIN train_prefix_freq f
    ON t.country = f.country
   AND f.prefix = substr(t.token3, 1, 3)
   AND f.freq <= 50

UNION

SELECT DISTINCT
    s.source1_entity_id,
    t.candidate_id
FROM val_s1_tokens s
JOIN train_tokens t
    ON s.country = t.country
   AND substr(s.token3, 1, 3) <> ''
   AND substr(s.token3, 1, 3) = substr(t.token1, 1, 3)
JOIN train_prefix_freq f
    ON t.country = f.country
   AND f.prefix = substr(t.token1, 1, 3)
   AND f.freq <= 50

UNION

SELECT DISTINCT
    s.source1_entity_id,
    t.candidate_id
FROM val_s1_tokens s
JOIN train_tokens t
    ON s.country = t.country
   AND substr(s.token3, 1, 3) <> ''
   AND substr(s.token3, 1, 3) = substr(t.token2, 1, 3)
JOIN train_prefix_freq f
    ON t.country = f.country
   AND f.prefix = substr(t.token2, 1, 3)
   AND f.freq <= 50

UNION

SELECT DISTINCT
    s.source1_entity_id,
    t.candidate_id
FROM val_s1_tokens s
JOIN train_tokens t
    ON s.country = t.country
   AND substr(s.token3, 1, 3) <> ''
   AND substr(s.token3, 1, 3) = substr(t.token3, 1, 3)
JOIN train_prefix_freq f
    ON t.country = f.country
   AND f.prefix = substr(t.token3, 1, 3)
   AND f.freq <= 50
""")

print("✅ 3-character prefix candidates generated")

print(
    con.execute("""
        SELECT
            COUNT(*) AS candidate_pairs,
            COUNT(DISTINCT source1_entity_id) AS s1_count
        FROM prefix3_candidates
    """).fetchdf()
)
```

```python
# ==========================================
# TEST 2-TOKEN PAIR WITH FREQUENCY LIMITS
# ==========================================

for limit in [50, 100, 250, 500, 1000]:

    result = con.execute(f"""
    WITH candidates AS (

        SELECT DISTINCT
            s.source1_entity_id,
            t.candidate_id

        FROM val_pair12 s

        JOIN pair12_freq f
            ON s.country = f.country
           AND s.pair_key = f.pair_key
           AND f.freq <= {limit}

        JOIN train_pair12 t
            ON s.country = t.country
           AND s.pair_key = t.pair_key
    ),

    gt_pairs AS (
        SELECT
            source1_entity_id,
            trim(x) AS candidate_id

        FROM val_gt_100,

        UNNEST(
            string_split(
                COALESCE(matched_entity_ids, ''),
                ','
            )
        ) AS u(x)

        WHERE trim(x) <> ''
    ),

    captured AS (
        SELECT
            g.source1_entity_id,
            g.candidate_id

        FROM gt_pairs g

        JOIN candidates c
            ON g.source1_entity_id = c.source1_entity_id
           AND g.candidate_id = c.candidate_id
    )

    SELECT
        {limit} AS max_frequency,
        (SELECT COUNT(*) FROM candidates) AS candidate_pairs,
        (SELECT COUNT(*) FROM captured) AS captured_true_matches,
        ROUND(
            100.0 *
            (SELECT COUNT(*) FROM captured)
            / 352,
            2
        ) AS recall_percent
    """).fetchdf()

    print(result.to_string(index=False))
```

```python
# ==========================================
# PROPER ADDRESS WORD-PAIR BLOCKER
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE train_addr_pair AS

SELECT
    entity_id AS candidate_id,
    country,

    list_extract(
        regexp_extract_all(
            lower(coalesce(business_address, '')),
            '[^\\W\\d_]+'
        ),
        1
    ) AS addr_w1,

    list_extract(
        regexp_extract_all(
            lower(coalesce(business_address, '')),
            '[^\\W\\d_]+'
        ),
        2
    ) AS addr_w2

FROM (
    SELECT entity_id, country, business_address
    FROM read_csv(
        'train_source2.tsv',
        delim='\t',
        header=true
    )

    UNION ALL

    SELECT entity_id, country, business_address
    FROM read_csv(
        'train_source3.tsv',
        delim='\t',
        header=true
    )
)
""")

con.execute("""
CREATE OR REPLACE TEMP TABLE val_addr_pair AS

SELECT
    entity_id AS source1_entity_id,
    country,

    list_extract(
        regexp_extract_all(
            lower(coalesce(business_address, '')),
            '[^\\W\\d_]+'
        ),
        1
    ) AS addr_w1,

    list_extract(
        regexp_extract_all(
            lower(coalesce(business_address, '')),
            '[^\\W\\d_]+'
        ),
        2
    ) AS addr_w2

FROM read_csv(
    'train_source1.tsv',
    delim='\t',
    header=true
)

WHERE entity_id IN (
    SELECT source1_entity_id
    FROM val_gt_100
)
""")

con.execute("""
CREATE OR REPLACE TEMP TABLE address_pair_candidates AS

SELECT DISTINCT
    s.source1_entity_id,
    t.candidate_id

FROM val_addr_pair s

JOIN train_addr_pair t
    ON s.country = t.country
   AND s.addr_w1 <> ''
   AND s.addr_w2 <> ''
   AND t.addr_w1 <> ''
   AND t.addr_w2 <> ''

   AND (
        (
            s.addr_w1 = t.addr_w1
            AND s.addr_w2 = t.addr_w2
        )
        OR
        (
            s.addr_w1 = t.addr_w2
            AND s.addr_w2 = t.addr_w1
        )
   )
""")

print("✅ Address pair candidates created")

print(
    con.execute("""
        SELECT
            COUNT(*) AS candidate_pairs,
            COUNT(DISTINCT source1_entity_id) AS s1_count
        FROM address_pair_candidates
    """).fetchdf()
)
```

```python
# ==========================================
# NAME + ADDRESS PAIR BENEFIT
# ==========================================

result = con.execute("""
WITH gt_pairs AS (
    SELECT
        source1_entity_id,
        trim(x) AS candidate_id
    FROM val_gt_100,
    UNNEST(
        string_split(
            COALESCE(matched_entity_ids, ''),
            ','
        )
    ) AS u(x)
    WHERE trim(x) <> ''
),

name_captured AS (
    SELECT
        g.source1_entity_id,
        g.candidate_id
    FROM gt_pairs g
    JOIN pair_candidates p
        ON g.source1_entity_id = p.source1_entity_id
       AND g.candidate_id = p.candidate_id
),

address_captured AS (
    SELECT
        g.source1_entity_id,
        g.candidate_id
    FROM gt_pairs g
    JOIN address_pair_candidates a
        ON g.source1_entity_id = a.source1_entity_id
       AND g.candidate_id = a.candidate_id
),

combined_captured AS (
    SELECT source1_entity_id, candidate_id
    FROM name_captured

    UNION

    SELECT source1_entity_id, candidate_id
    FROM address_captured
),

new_from_address AS (
    SELECT
        a.source1_entity_id,
        a.candidate_id
    FROM address_captured a
    LEFT JOIN name_captured n
        ON a.source1_entity_id = n.source1_entity_id
       AND a.candidate_id = n.candidate_id
    WHERE n.candidate_id IS NULL
)

SELECT
    (SELECT COUNT(*) FROM gt_pairs) AS total_true_matches,

    (SELECT COUNT(*) FROM name_captured)
        AS name_true_matches,

    ROUND(
        100.0 * (SELECT COUNT(*) FROM name_captured)
        / (SELECT COUNT(*) FROM gt_pairs),
        2
    ) AS name_recall_percent,

    (SELECT COUNT(*) FROM new_from_address)
        AS new_true_matches_from_address,

    (SELECT COUNT(*) FROM combined_captured)
        AS combined_true_matches,

    ROUND(
        100.0 * (SELECT COUNT(*) FROM combined_captured)
        / (SELECT COUNT(*) FROM gt_pairs),
        2
    ) AS combined_recall_percent,

    (SELECT COUNT(*) FROM address_pair_candidates)
        AS address_candidate_pairs

""").fetchdf()

print(result)
```

```python
# ==========================================
# TARGETED ADDRESS KEY
# NUMBER + STREET WORD
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE train_addr_targeted AS

SELECT
    entity_id AS candidate_id,
    country,

    regexp_extract(
        coalesce(business_address, ''),
        '[0-9]+',
        0
    ) AS address_number,

    lower(
        regexp_extract(
            regexp_replace(
                coalesce(business_address, ''),
                '^[^\\p{L}]*[0-9]+[^\\p{L}]+',
                '',
                'g'
            ),
            '([\\p{L}]{3,})',
            1
        )
    ) AS street_word

FROM (
    SELECT entity_id, country, business_address
    FROM read_csv(
        'train_source2.tsv',
        delim='\t',
        header=true
    )

    UNION ALL

    SELECT entity_id, country, business_address
    FROM read_csv(
        'train_source3.tsv',
        delim='\t',
        header=true
    )
)
""")

print("✅ Targeted address table created")

print(
    con.execute("""
        SELECT COUNT(*) AS rows
        FROM train_addr_targeted
    """).fetchdf()
)
```

```python
# ==========================================
# TARGETED ADDRESS KEY FOR VALIDATION S1
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE val_addr_targeted AS

SELECT
    entity_id AS source1_entity_id,
    country,

    regexp_extract(
        coalesce(business_address, ''),
        '[0-9]+',
        0
    ) AS address_number,

    lower(
        regexp_extract(
            regexp_replace(
                coalesce(business_address, ''),
                '^[^\\p{L}]*[0-9]+[^\\p{L}]+',
                '',
                'g'
            ),
            '([\\p{L}]{3,})',
            1
        )
    ) AS street_word

FROM read_csv(
    'train_source1.tsv',
    delim='\t',
    header=true
)

WHERE entity_id IN (
    SELECT source1_entity_id
    FROM val_gt_100
)
""")

print(
    con.execute("""
        SELECT
            COUNT(*) AS validation_s1,
            SUM(CASE WHEN address_number <> '' THEN 1 ELSE 0 END)
                AS with_number,
            SUM(CASE WHEN street_word <> '' THEN 1 ELSE 0 END)
                AS with_street_word
        FROM val_addr_targeted
    """).fetchdf()
)
```

```python
# ==========================================
# TARGETED ADDRESS CANDIDATE COUNT
# country + number + street word
# ==========================================

result = con.execute("""
SELECT
    COUNT(*) AS candidate_pairs,
    COUNT(DISTINCT v.source1_entity_id) AS s1_count
FROM val_addr_targeted v

JOIN train_addr_targeted t
    ON v.country = t.country
   AND v.address_number <> ''
   AND t.address_number <> ''
   AND ltrim(v.address_number, '0')
       = ltrim(t.address_number, '0')
   AND v.street_word <> ''
   AND t.street_word <> ''
   AND v.street_word = t.street_word
""").fetchdf()

print(result)
```

```python
# ==========================================
# TARGETED ADDRESS BLOCKING RECALL
# ==========================================

result = con.execute("""
WITH gt_pairs AS (
    SELECT
        source1_entity_id,
        trim(x) AS candidate_id
    FROM val_gt_100,
    UNNEST(
        string_split(
            COALESCE(matched_entity_ids, ''),
            ','
        )
    ) AS u(x)
    WHERE trim(x) <> ''
),

captured AS (
    SELECT
        g.source1_entity_id,
        g.candidate_id
    FROM gt_pairs g

    JOIN val_addr_targeted s
        ON g.source1_entity_id = s.source1_entity_id

    JOIN train_addr_targeted t
        ON g.candidate_id = t.candidate_id
       AND s.country = t.country
       AND s.address_number <> ''
       AND t.address_number <> ''
       AND ltrim(s.address_number, '0')
           = ltrim(t.address_number, '0')
       AND s.street_word <> ''
       AND t.street_word <> ''
       AND s.street_word = t.street_word
)

SELECT
    (SELECT COUNT(*) FROM gt_pairs) AS total_true_matches,

    (SELECT COUNT(*)
     FROM captured) AS captured_true_matches,

    ROUND(
        100.0 *
        (SELECT COUNT(*) FROM captured)
        / (SELECT COUNT(*) FROM gt_pairs),
        2
    ) AS recall_percent
""").fetchdf()

print(result)
```

```python
# ==========================================
# NAME + TARGETED ADDRESS BLOCKING
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE combined_name_address AS

SELECT
    source1_entity_id,
    candidate_id
FROM pair_candidates

UNION

SELECT
    v.source1_entity_id,
    t.candidate_id
FROM val_addr_targeted v

JOIN train_addr_targeted t
    ON v.country = t.country
   AND v.address_number <> ''
   AND t.address_number <> ''
   AND ltrim(v.address_number, '0')
       = ltrim(t.address_number, '0')
   AND v.street_word <> ''
   AND t.street_word <> ''
   AND v.street_word = t.street_word
""")

result = con.execute("""
WITH gt_pairs AS (
    SELECT
        source1_entity_id,
        trim(x) AS candidate_id
    FROM val_gt_100,
    UNNEST(
        string_split(
            COALESCE(matched_entity_ids, ''),
            ','
        )
    ) AS u(x)
    WHERE trim(x) <> ''
),

captured AS (
    SELECT
        g.source1_entity_id,
        g.candidate_id
    FROM gt_pairs g
    JOIN combined_name_address c
        ON g.source1_entity_id = c.source1_entity_id
       AND g.candidate_id = c.candidate_id
)

SELECT
    (SELECT COUNT(*) FROM gt_pairs) AS total_true_matches,

    (SELECT COUNT(*) FROM captured) AS captured_true_matches,

    ROUND(
        100.0 *
        (SELECT COUNT(*) FROM captured)
        / (SELECT COUNT(*) FROM gt_pairs),
        2
    ) AS combined_recall_percent,

    (SELECT COUNT(*) FROM combined_name_address)
        AS total_candidate_pairs
""").fetchdf()

print(result)
```

```python
# ==========================================
# CHECK ANY NAME-TOKEN OVERLAP
# ON THE 81 REMAINING MISSES
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE remaining_misses AS

WITH gt_pairs AS (
    SELECT
        source1_entity_id,
        trim(x) AS candidate_id

    FROM val_gt_100,
    UNNEST(
        string_split(
            COALESCE(matched_entity_ids, ''),
            ','
        )
    ) AS u(x)

    WHERE trim(x) <> ''
)

SELECT
    g.source1_entity_id,
    g.candidate_id

FROM gt_pairs g

LEFT JOIN combined_name_address c
    ON g.source1_entity_id = c.source1_entity_id
   AND g.candidate_id = c.candidate_id

WHERE c.candidate_id IS NULL
""")

result = con.execute("""
SELECT
    COUNT(*) AS remaining_misses,

    SUM(
        CASE
            WHEN
                (
                    s.token1 <> '' AND
                    (
                        s.token1 = t.token1 OR
                        s.token1 = t.token2 OR
                        s.token1 = t.token3
                    )
                )
                OR
                (
                    s.token2 <> '' AND
                    (
                        s.token2 = t.token1 OR
                        s.token2 = t.token2 OR
                        s.token2 = t.token3
                    )
                )
                OR
                (
                    s.token3 <> '' AND
                    (
                        s.token3 = t.token1 OR
                        s.token3 = t.token2 OR
                        s.token3 = t.token3
                    )
                )
            THEN 1
            ELSE 0
        END
    ) AS with_any_common_token

FROM remaining_misses m

JOIN val_s1_tokens s
    ON m.source1_entity_id = s.source1_entity_id

JOIN train_tokens t
    ON m.candidate_id = t.candidate_id
""").fetchdf()

print(result)
```

```python
# ==========================================
# TEST SINGLE NAME TOKEN FREQUENCY
# ON 81 REMAINING MISSES
# ==========================================

# Recreate token frequency table from the existing train_tokens table
con.execute("""
CREATE OR REPLACE TEMP TABLE train_token_freq AS

SELECT
    country,
    token,
    COUNT(*) AS freq
FROM (
    SELECT country, token1 AS token
    FROM train_tokens
    WHERE token1 <> ''

    UNION ALL

    SELECT country, token2 AS token
    FROM train_tokens
    WHERE token2 <> ''

    UNION ALL

    SELECT country, token3 AS token
    FROM train_tokens
    WHERE token3 <> ''
)
GROUP BY country, token
""")

for limit in [10, 25, 50, 100, 250, 500, 1000]:

    result = con.execute(f"""
    SELECT
        {limit} AS max_frequency,

        COUNT(*) FILTER (
            WHERE
                (
                    s.token1 <> '' AND
                    (
                        (s.token1 = t.token1 AND f1.freq <= {limit}) OR
                        (s.token1 = t.token2 AND f2.freq <= {limit}) OR
                        (s.token1 = t.token3 AND f3.freq <= {limit})
                    )
                )
                OR
                (
                    s.token2 <> '' AND
                    (
                        (s.token2 = t.token1 AND f1.freq <= {limit}) OR
                        (s.token2 = t.token2 AND f2.freq <= {limit}) OR
                        (s.token2 = t.token3 AND f3.freq <= {limit})
                    )
                )
                OR
                (
                    s.token3 <> '' AND
                    (
                        (s.token3 = t.token1 AND f1.freq <= {limit}) OR
                        (s.token3 = t.token2 AND f2.freq <= {limit}) OR
                        (s.token3 = t.token3 AND f3.freq <= {limit})
                    )
                )
        ) AS recoverable

    FROM remaining_misses m

    JOIN val_s1_tokens s
        ON m.source1_entity_id = s.source1_entity_id

    JOIN train_tokens t
        ON m.candidate_id = t.candidate_id

    LEFT JOIN train_token_freq f1
        ON t.country = f1.country
       AND t.token1 = f1.token

    LEFT JOIN train_token_freq f2
        ON t.country = f2.country
       AND t.token2 = f2.token

    LEFT JOIN train_token_freq f3
        ON t.country = f3.country
       AND t.token3 = f3.token
    """).fetchdf()

    print(result.to_string(index=False))
```

```python
# ==========================================
# CHARACTER N-GRAM TEST ON 81 MISSES
# ==========================================

import re
import pandas as pd

# Get the actual names for the 81 missed true matches
miss_names = con.execute("""
WITH candidates AS (
    SELECT
        entity_id,
        business_name
    FROM read_csv(
        'train_source2.tsv',
        delim='\t',
        header=true
    )

    UNION ALL

    SELECT
        entity_id,
        business_name
    FROM read_csv(
        'train_source3.tsv',
        delim='\t',
        header=true
    )
)

SELECT
    m.source1_entity_id,
    m.candidate_id,
    s.business_name AS s1_name,
    c.business_name AS candidate_name

FROM remaining_misses m

JOIN read_csv(
    'train_source1.tsv',
    delim='\t',
    header=true
) s
    ON m.source1_entity_id = s.entity_id

JOIN candidates c
    ON m.candidate_id = c.entity_id
""").fetchdf()


def compact(text):
    if pd.isna(text):
        return ""
    return re.sub(r'\s+', '', str(text).lower())


def ngrams(text, n):
    text = compact(text)
    if len(text) < n:
        return set()
    return {
        text[i:i+n]
        for i in range(len(text) - n + 1)
    }


for n in [3, 4, 5]:

    miss_names[f"s1_{n}gram"] = miss_names["s1_name"].apply(
        lambda x: ngrams(x, n)
    )

    miss_names[f"cand_{n}gram"] = miss_names["candidate_name"].apply(
        lambda x: ngrams(x, n)
    )

    miss_names[f"shared_{n}gram"] = miss_names.apply(
        lambda r: len(
            r[f"s1_{n}gram"] &
            r[f"cand_{n}gram"]
        ),
        axis=1
    )

    print(
        f"{n}-gram shared >= 1:",
        (
            miss_names[f"shared_{n}gram"] >= 1
        ).sum()
    )

    print(
        f"{n}-gram shared >= 2:",
        (
            miss_names[f"shared_{n}gram"] >= 2
        ).sum()
    )

print("\nSample:")
print(
    miss_names[
        [
            "s1_name",
            "candidate_name",
            "shared_3gram",
            "shared_4gram",
            "shared_5gram"
        ]
    ].head(20).to_string(index=False)
)
```

```python
# ==========================================
# 4-GRAM OVERLAP DISTRIBUTION
# ==========================================

for threshold in [1, 2, 3, 5, 8, 10, 15, 20]:

    count = (
        miss_names["shared_4gram"] >= threshold
    ).sum()

    print(
        f"4-gram overlap >= {threshold}: {count} / {len(miss_names)}"
    )

print("\nDistribution:")
print(
    miss_names["shared_4gram"]
    .describe()
)
```

```python
# ==========================================
# FROZEN VALIDATION BASELINE
# ==========================================

result = con.execute("""
WITH gt_pairs AS (
    SELECT
        source1_entity_id,
        trim(x) AS candidate_id

    FROM val_gt_100,

    UNNEST(
        string_split(
            COALESCE(matched_entity_ids, ''),
            ','
        )
    ) AS u(x)

    WHERE trim(x) <> ''
),

name_captured AS (
    SELECT
        g.source1_entity_id,
        g.candidate_id

    FROM gt_pairs g

    JOIN pair_candidates p
        ON g.source1_entity_id = p.source1_entity_id
       AND g.candidate_id = p.candidate_id
),

combined_candidates AS (
    SELECT
        source1_entity_id,
        candidate_id

    FROM pair_candidates

    UNION

    SELECT
        source1_entity_id,
        candidate_id

    FROM address_pair_candidates
),

combined_captured AS (
    SELECT
        g.source1_entity_id,
        g.candidate_id

    FROM gt_pairs g

    JOIN combined_candidates c
        ON g.source1_entity_id = c.source1_entity_id
       AND g.candidate_id = c.candidate_id
)

SELECT

    -- Ground truth size
    (SELECT COUNT(*) FROM gt_pairs)
        AS total_true_matches,

    -- Name only
    (SELECT COUNT(*) FROM name_captured)
        AS name_captured,

    ROUND(
        100.0 *
        (SELECT COUNT(*) FROM name_captured)
        /
        (SELECT COUNT(*) FROM gt_pairs),
        2
    ) AS name_recall,

    -- Name + address
    (SELECT COUNT(*) FROM combined_captured)
        AS combined_captured,

    ROUND(
        100.0 *
        (SELECT COUNT(*) FROM combined_captured)
        /
        (SELECT COUNT(*) FROM gt_pairs),
        2
    ) AS combined_recall,

    -- Candidate sizes
    (SELECT COUNT(*) FROM pair_candidates)
        AS name_candidates,

    (SELECT COUNT(*) FROM combined_candidates)
        AS combined_candidates

""").fetchdf()

print(result.to_string(index=False))
```

```python
# ==========================================
# CORE NAME RECOVERY TEST
# ==========================================

import re
import pandas as pd

LEGAL_WORDS = {
    "inc", "incorporated",
    "llc", "ltd", "limited",
    "private", "pvt",
    "plc", "corp", "corporation",
    "company", "co",
    "pc"
}

def core_name(text):
    if pd.isna(text):
        return set()

    words = re.findall(
        r'[^\W\d_]+',
        str(text).lower(),
        flags=re.UNICODE
    )

    return {
        w for w in words
        if w not in LEGAL_WORDS and len(w) >= 2
    }

# Current true misses relative to the frozen name-pair blocker
con.execute("""
CREATE OR REPLACE TEMP TABLE current_name_misses AS

WITH gt_pairs AS (
    SELECT
        source1_entity_id,
        trim(x) AS candidate_id

    FROM val_gt_100,
    UNNEST(
        string_split(
            COALESCE(matched_entity_ids, ''),
            ','
        )
    ) AS u(x)

    WHERE trim(x) <> ''
)

SELECT
    g.source1_entity_id,
    g.candidate_id

FROM gt_pairs g

LEFT JOIN pair_candidates p
    ON g.source1_entity_id = p.source1_entity_id
   AND g.candidate_id = p.candidate_id

WHERE p.candidate_id IS NULL
""")

print(
    con.execute("""
        SELECT COUNT(*) AS missed
        FROM current_name_misses
    """).fetchdf()
)

# Read names only for the current missed true pairs
miss_df = con.execute("""
WITH records AS (
    SELECT entity_id, business_name
    FROM read_csv(
        'train_source2.tsv',
        delim='\t',
        header=true
    )

    UNION ALL

    SELECT entity_id, business_name
    FROM read_csv(
        'train_source3.tsv',
        delim='\t',
        header=true
    )
)

SELECT
    m.source1_entity_id,
    m.candidate_id,
    s.business_name AS s1_name,
    r.business_name AS candidate_name

FROM current_name_misses m

JOIN read_csv(
    'train_source1.tsv',
    delim='\t',
    header=true
) s
    ON m.source1_entity_id = s.entity_id

JOIN records r
    ON m.candidate_id = r.entity_id
""").fetchdf()

miss_df["s1_core"] = miss_df["s1_name"].apply(core_name)
miss_df["candidate_core"] = miss_df["candidate_name"].apply(core_name)

miss_df["common_core_words"] = miss_df.apply(
    lambda r: r["s1_core"] & r["candidate_core"],
    axis=1
)

print(
    "At least 2 common core words:",
    (miss_df["common_core_words"].str.len() >= 2).sum()
)

print(
    "At least 1 common core word:",
    (miss_df["common_core_words"].str.len() >= 1).sum()
)
```

```python
# ==========================================
# CORE NAME EXACT / SUBSET TEST
# ==========================================

miss_df["core_equal"] = miss_df.apply(
    lambda r:
        r["s1_core"] == r["candidate_core"],
    axis=1
)

miss_df["core_subset"] = miss_df.apply(
    lambda r:
        r["s1_core"].issubset(r["candidate_core"])
        or r["candidate_core"].issubset(r["s1_core"]),
    axis=1
)

print("Current name-block misses:", len(miss_df))

print(
    "Exact same core words:",
    miss_df["core_equal"].sum()
)

print(
    "One core word-set contains the other:",
    miss_df["core_subset"].sum()
)

print("\nExamples of exact core matches:")
print(
    miss_df[
        miss_df["core_equal"]
    ][
        ["s1_name", "candidate_name"]
    ].head(20).to_string(index=False)
)
```

```python
# ==========================================
# CANONICAL CORE NAME KEY
# ==========================================

LEGAL_WORDS = {
    "inc", "incorporated",
    "llc", "ltd", "limited",
    "private", "pvt",
    "plc", "corp", "corporation",
    "company", "co", "pc"
}

def canonical_key(row):
    words = [
        row["token1"],
        row["token2"],
        row["token3"]
    ]

    words = [
        w for w in words
        if w
        and w not in LEGAL_WORDS
        and len(w) >= 3
    ]

    return "|".join(sorted(set(words)))


# Create canonical key for training candidates
train_core = train_tokens.to_pandas() if False else None
```

```python
# ==========================================
# CANONICAL CORE NAME KEY
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE train_core AS

SELECT
    candidate_id,
    country,

    array_to_string(
        list_sort(
            list_distinct(
                list_filter(
                    [token1, token2, token3],
                    x ->
                        x <> ''
                        AND length(x) >= 3
                        AND x NOT IN (
                            'inc','incorporated',
                            'llc','ltd','limited',
                            'private','pvt',
                            'plc','corp','corporation',
                            'company','co','pc'
                        )
                )
            )
        ),
        '|'
    ) AS core_key

FROM train_tokens
""")

con.execute("""
CREATE OR REPLACE TEMP TABLE val_core AS

SELECT
    source1_entity_id,
    country,

    array_to_string(
        list_sort(
            list_distinct(
                list_filter(
                    [token1, token2, token3],
                    x ->
                        x <> ''
                        AND length(x) >= 3
                        AND x NOT IN (
                            'inc','incorporated',
                            'llc','ltd','limited',
                            'private','pvt',
                            'plc','corp','corporation',
                            'company','co','pc'
                        )
                )
            )
        ),
        '|'
    ) AS core_key

FROM val_s1_tokens
""")

print("✅ Canonical core-name keys created")
```

```python
# ==========================================
# CANONICAL CORE-NAME CANDIDATES
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE core_candidates AS

SELECT DISTINCT
    v.source1_entity_id,
    t.candidate_id

FROM val_core v

JOIN train_core t
    ON v.country = t.country
   AND v.core_key <> ''
   AND t.core_key <> ''
   AND v.core_key = t.core_key
""")

print("✅ Core-name candidates created")

print(
    con.execute("""
        SELECT
            COUNT(*) AS candidate_pairs,
            COUNT(DISTINCT source1_entity_id) AS s1_count
        FROM core_candidates
    """).fetchdf()
)
```

```python
# ==========================================
# CORE-NAME BLOCKING RECALL
# ==========================================

result = con.execute("""
WITH gt_pairs AS (
    SELECT
        source1_entity_id,
        trim(x) AS candidate_id
    FROM val_gt_100,
    UNNEST(
        string_split(
            COALESCE(matched_entity_ids, ''),
            ','
        )
    ) AS u(x)
    WHERE trim(x) <> ''
),

captured AS (
    SELECT
        g.source1_entity_id,
        g.candidate_id
    FROM gt_pairs g
    JOIN core_candidates c
        ON g.source1_entity_id = c.source1_entity_id
       AND g.candidate_id = c.candidate_id
)

SELECT
    (SELECT COUNT(*) FROM gt_pairs) AS total_true_matches,

    (SELECT COUNT(*) FROM captured) AS captured_true_matches,

    ROUND(
        100.0 *
        (SELECT COUNT(*) FROM captured)
        / (SELECT COUNT(*) FROM gt_pairs),
        2
    ) AS recall_percent
""").fetchdf()

print(result)
```

```python
# ==========================================
# ADD CORE-NAME TO CURRENT BEST BLOCKER
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE combined_candidates_v4 AS

SELECT
    source1_entity_id,
    candidate_id
FROM combined_name_address

UNION

SELECT
    source1_entity_id,
    candidate_id
FROM core_candidates
""")

result = con.execute("""
WITH gt_pairs AS (
    SELECT
        source1_entity_id,
        trim(x) AS candidate_id
    FROM val_gt_100,
    UNNEST(
        string_split(
            COALESCE(matched_entity_ids, ''),
            ','
        )
    ) AS u(x)
    WHERE trim(x) <> ''
),

captured AS (
    SELECT
        g.source1_entity_id,
        g.candidate_id
    FROM gt_pairs g
    JOIN combined_candidates_v4 c
        ON g.source1_entity_id = c.source1_entity_id
       AND g.candidate_id = c.candidate_id
)

SELECT
    (SELECT COUNT(*) FROM gt_pairs) AS total_true_matches,

    (SELECT COUNT(*) FROM captured)
        AS captured_true_matches,

    ROUND(
        100.0 *
        (SELECT COUNT(*) FROM captured)
        / (SELECT COUNT(*) FROM gt_pairs),
        2
    ) AS recall_percent,

    (SELECT COUNT(*) FROM combined_candidates_v4)
        AS total_candidate_pairs
""").fetchdf()

print(result)
```

```python
# ==========================================
# SINGLE NAME TOKEN BLOCKER - FREQ <= 1000
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE single_token_1000_candidates AS

SELECT DISTINCT
    s.source1_entity_id,
    t.candidate_id

FROM val_s1_tokens s

JOIN train_tokens t
    ON s.country = t.country
   AND s.token1 <> ''
   AND s.token1 = t.token1

JOIN train_token_freq f
    ON t.country = f.country
   AND t.token1 = f.token

WHERE f.freq <= 1000

UNION

SELECT DISTINCT
    s.source1_entity_id,
    t.candidate_id

FROM val_s1_tokens s

JOIN train_tokens t
    ON s.country = t.country
   AND s.token2 <> ''
   AND s.token2 = t.token2

JOIN train_token_freq f
    ON t.country = f.country
   AND t.token2 = f.token

WHERE f.freq <= 1000

UNION

SELECT DISTINCT
    s.source1_entity_id,
    t.candidate_id

FROM val_s1_tokens s

JOIN train_tokens t
    ON s.country = t.country
   AND s.token3 <> ''
   AND s.token3 = t.token3

JOIN train_token_freq f
    ON t.country = f.country
   AND t.token3 = f.token

WHERE f.freq <= 1000
""")

print("✅ Single-token candidates created")

print(
    con.execute("""
        SELECT
            COUNT(*) AS candidate_pairs,
            COUNT(DISTINCT source1_entity_id) AS s1_count
        FROM single_token_1000_candidates
    """).fetchdf()
)
```

```python
# ==========================================
# RARE ADDRESS NUMBER BLOCKER
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE number_freq AS

SELECT
    country,
    ltrim(address_number, '0') AS number_clean,
    COUNT(*) AS freq

FROM train_addr_targeted

WHERE address_number <> ''

GROUP BY
    country,
    ltrim(address_number, '0')
""")

con.execute("""
CREATE OR REPLACE TEMP TABLE number_candidates AS

SELECT DISTINCT
    v.source1_entity_id,
    t.candidate_id

FROM val_addr_targeted v

JOIN number_freq f
    ON v.country = f.country
   AND v.address_number <> ''
   AND ltrim(v.address_number, '0') = f.number_clean
   AND f.freq <= 500

JOIN train_addr_targeted t
    ON t.country = v.country
   AND t.address_number <> ''
   AND ltrim(t.address_number, '0') = f.number_clean
""")

print("✅ Number candidates created")

print(
    con.execute("""
        SELECT
            COUNT(*) AS candidate_pairs,
            COUNT(DISTINCT source1_entity_id) AS s1_count
        FROM number_candidates
    """).fetchdf()
)
```

```python
# ==========================================
# CURRENT 56 MISSES + ADDRESS WORD OVERLAP
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE remaining_misses_v7 AS

WITH gt_pairs AS (
    SELECT
        source1_entity_id,
        trim(x) AS candidate_id
    FROM val_gt_100,
    UNNEST(
        string_split(
            COALESCE(matched_entity_ids, ''),
            ','
        )
    ) AS u(x)
    WHERE trim(x) <> ''
)

SELECT
    g.source1_entity_id,
    g.candidate_id
FROM gt_pairs g
LEFT JOIN combined_candidates_v7 c
    ON g.source1_entity_id = c.source1_entity_id
   AND g.candidate_id = c.candidate_id
WHERE c.candidate_id IS NULL
""")

# Get addresses for only these missed true pairs
miss_addr = con.execute("""
WITH records AS (
    SELECT entity_id, business_address
    FROM read_csv('train_source2.tsv', delim='\t', header=true)

    UNION ALL

    SELECT entity_id, business_address
    FROM read_csv('train_source3.tsv', delim='\t', header=true)
)

SELECT
    m.source1_entity_id,
    m.candidate_id,
    s.business_address AS s1_address,
    r.business_address AS candidate_address
FROM remaining_misses_v7 m

JOIN read_csv(
    'train_source1.tsv',
    delim='\t',
    header=true
) s
    ON m.source1_entity_id = s.entity_id

JOIN records r
    ON m.candidate_id = r.entity_id
""").fetchdf()


def meaningful_address_words(text):
    if pd.isna(text):
        return set()

    words = re.findall(
        r'[^\W\d_]+',
        str(text).lower(),
        flags=re.UNICODE
    )

    # Ignore very short/common address words
    ignore = {
        "the", "and", "of", "road", "rd",
        "street", "st", "lane", "ln",
        "avenue", "ave", "floor"
    }

    return {
        w for w in words
        if len(w) >= 4 and w not in ignore
    }


miss_addr["s1_words"] = miss_addr["s1_address"].apply(
    meaningful_address_words
)

miss_addr["candidate_words"] = miss_addr[
    "candidate_address"
].apply(meaningful_address_words)

miss_addr["common_words"] = miss_addr.apply(
    lambda r:
        r["s1_words"] & r["candidate_words"],
    axis=1
)

print("Current remaining misses:", len(miss_addr))

print(
    "At least 1 common address word:",
    (miss_addr["common_words"].str.len() >= 1).sum()
)

print(
    "At least 2 common address words:",
    (miss_addr["common_words"].str.len() >= 2).sum()
)
```

```python
# ==========================================
# CHECK ADDRESS-PAIR RECOVERY ON 56 MISSES
# ==========================================

result = con.execute("""
SELECT
    COUNT(*) AS remaining_misses,

    COUNT(a.candidate_id) AS recovered_by_address_pair

FROM remaining_misses_v7 m

LEFT JOIN address_pair_candidates a
    ON m.source1_entity_id = a.source1_entity_id
   AND m.candidate_id = a.candidate_id
""").fetchdf()

print(result)
```

```python
# ==========================================
# STREET-WORD BLOCKING
# TEST DIFFERENT FREQUENCY LIMITS
# ==========================================

# Frequency of street words
con.execute("""
CREATE OR REPLACE TEMP TABLE street_word_freq AS

SELECT
    country,
    street_word,
    COUNT(*) AS freq

FROM train_addr_targeted

WHERE street_word <> ''

GROUP BY
    country,
    street_word
""")

# Test several frequency limits
for limit in [25, 50, 100, 250, 500]:

    result = con.execute(f"""
    SELECT
        {limit} AS max_frequency,

        COUNT(*) AS candidate_pairs,

        COUNT(DISTINCT v.source1_entity_id)
            AS s1_count

    FROM val_addr_targeted v

    JOIN street_word_freq f
        ON v.country = f.country
       AND v.street_word = f.street_word
       AND f.freq <= {limit}

    JOIN train_addr_targeted t
        ON v.country = t.country
       AND v.street_word = t.street_word

    WHERE v.street_word <> ''
    """).fetchdf()

    print(result.to_string(index=False))
```

```python
# ==========================================
# STREET-WORD RECOVERY ON 56 MISSES
# ==========================================

for limit in [25, 50, 100, 250, 500]:

    result = con.execute(f"""
    SELECT
        {limit} AS max_frequency,

        COUNT(*) AS recoverable_true_matches

    FROM remaining_misses_v7 m

    JOIN val_addr_targeted s
        ON m.source1_entity_id = s.source1_entity_id

    JOIN train_addr_targeted t
        ON m.candidate_id = t.candidate_id
       AND s.country = t.country
       AND s.street_word <> ''
       AND t.street_word <> ''
       AND s.street_word = t.street_word

    JOIN street_word_freq f
        ON t.country = f.country
       AND t.street_word = f.street_word
       AND f.freq <= {limit}
    """)

    print(result.to_string(index=False))
```

```python
# ==========================================
# STREET-WORD RECOVERY ON 56 MISSES
# ==========================================

for limit in [25, 50, 100, 250, 500]:

    result = con.execute(f"""
    SELECT
        {limit} AS max_frequency,
        COUNT(*) AS recoverable_true_matches

    FROM remaining_misses_v7 m

    JOIN val_addr_targeted s
        ON m.source1_entity_id = s.source1_entity_id

    JOIN train_addr_targeted t
        ON m.candidate_id = t.candidate_id
       AND s.country = t.country
       AND s.street_word <> ''
       AND t.street_word <> ''
       AND s.street_word = t.street_word

    JOIN street_word_freq f
        ON t.country = f.country
       AND t.street_word = f.street_word
       AND f.freq <= {limit}
    """).fetchdf()

    print(result.to_string(index=False))
```

```python
# ==========================================
# FINAL CONSOLIDATED BLOCKER - VALIDATION
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE final_validation_candidates AS

SELECT source1_entity_id, candidate_id
FROM combined_name_address

UNION

SELECT source1_entity_id, candidate_id
FROM core_candidates

UNION

SELECT source1_entity_id, candidate_id
FROM single_token_1000_candidates

UNION

SELECT source1_entity_id, candidate_id
FROM number_candidates
""")

result = con.execute("""
WITH gt_pairs AS (
    SELECT
        source1_entity_id,
        trim(x) AS candidate_id
    FROM val_gt_100,
    UNNEST(
        string_split(
            COALESCE(matched_entity_ids, ''),
            ','
        )
    ) AS u(x)
    WHERE trim(x) <> ''
),

captured AS (
    SELECT
        g.source1_entity_id,
        g.candidate_id
    FROM gt_pairs g
    JOIN final_validation_candidates c
        ON g.source1_entity_id = c.source1_entity_id
       AND g.candidate_id = c.candidate_id
)

SELECT
    (SELECT COUNT(*) FROM gt_pairs)
        AS total_true_matches,

    (SELECT COUNT(*) FROM captured)
        AS captured_true_matches,

    (SELECT COUNT(*) FROM gt_pairs)
      -
    (SELECT COUNT(*) FROM captured)
        AS missed_true_matches,

    ROUND(
        100.0 *
        (SELECT COUNT(*) FROM captured)
        /
        (SELECT COUNT(*) FROM gt_pairs),
        2
    ) AS blocking_recall_percent,

    (SELECT COUNT(*)
     FROM final_validation_candidates)
        AS total_candidate_pairs,

    (SELECT COUNT(DISTINCT source1_entity_id)
     FROM final_validation_candidates)
        AS s1_with_candidates
""").fetchdf()

print(result)
```

```python
# ==========================================
# EXACT NORMALIZED ADDRESS BLOCKER
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE train_full_address AS

SELECT
    entity_id AS candidate_id,
    country,

    lower(
        regexp_replace(
            coalesce(business_address, ''),
            '[^\\p{L}\\p{N}]+',
            '',
            'g'
        )
    ) AS address_key

FROM (
    SELECT entity_id, country, business_address
    FROM read_csv(
        'train_source2.tsv',
        delim='\t',
        header=true
    )

    UNION ALL

    SELECT entity_id, country, business_address
    FROM read_csv(
        'train_source3.tsv',
        delim='\t',
        header=true
    )
)
""")

con.execute("""
CREATE OR REPLACE TEMP TABLE val_full_address AS

SELECT
    entity_id AS source1_entity_id,
    country,

    lower(
        regexp_replace(
            coalesce(business_address, ''),
            '[^\\p{L}\\p{N}]+',
            '',
            'g'
        )
    ) AS address_key

FROM read_csv(
    'train_source1.tsv',
    delim='\t',
    header=true
)

WHERE entity_id IN (
    SELECT source1_entity_id
    FROM val_gt_100
)
""")

con.execute("""
CREATE OR REPLACE TEMP TABLE exact_address_candidates AS

SELECT DISTINCT
    v.source1_entity_id,
    t.candidate_id

FROM val_full_address v

JOIN train_full_address t
    ON v.country = t.country
   AND v.address_key <> ''
   AND t.address_key <> ''
   AND v.address_key = t.address_key
""")

print("✅ Exact-address candidates created")

print(
    con.execute("""
        SELECT
            COUNT(*) AS candidate_pairs,
            COUNT(DISTINCT source1_entity_id) AS s1_count
        FROM exact_address_candidates
    """).fetchdf()
)
```

```python
# ==========================================
# ADD EXACT ADDRESS TO CURRENT BLOCKER
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE final_validation_candidates_v2 AS

SELECT
    source1_entity_id,
    candidate_id
FROM final_validation_candidates

UNION

SELECT
    source1_entity_id,
    candidate_id
FROM exact_address_candidates
""")

result = con.execute("""
WITH gt_pairs AS (
    SELECT
        source1_entity_id,
        trim(x) AS candidate_id
    FROM val_gt_100,
    UNNEST(
        string_split(
            COALESCE(matched_entity_ids, ''),
            ','
        )
    ) AS u(x)
    WHERE trim(x) <> ''
),

captured AS (
    SELECT
        g.source1_entity_id,
        g.candidate_id
    FROM gt_pairs g
    JOIN final_validation_candidates_v2 c
        ON g.source1_entity_id = c.source1_entity_id
       AND g.candidate_id = c.candidate_id
)

SELECT
    (SELECT COUNT(*) FROM gt_pairs)
        AS total_true_matches,

    (SELECT COUNT(*) FROM captured)
        AS captured_true_matches,

    (SELECT COUNT(*) FROM gt_pairs)
      -
    (SELECT COUNT(*) FROM captured)
        AS missed_true_matches,

    ROUND(
        100.0 *
        (SELECT COUNT(*) FROM captured)
        /
        (SELECT COUNT(*) FROM gt_pairs),
        2
    ) AS blocking_recall_percent,

    (SELECT COUNT(*) FROM final_validation_candidates_v2)
        AS total_candidate_pairs

""").fetchdf()

print(result)
```

```python
# ==========================================
# CHEAP SCORE FOR CURRENT CANDIDATES
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE scored_candidates AS

SELECT
    c.source1_entity_id,
    c.candidate_id,

    -- Count how many of the 3 name tokens are shared
    (
        CASE
            WHEN s.token1 <> ''
             AND (
                    s.token1 = t.token1
                 OR s.token1 = t.token2
                 OR s.token1 = t.token3
             )
            THEN 1 ELSE 0
        END
        +
        CASE
            WHEN s.token2 <> ''
             AND (
                    s.token2 = t.token1
                 OR s.token2 = t.token2
                 OR s.token2 = t.token3
             )
            THEN 1 ELSE 0
        END
        +
        CASE
            WHEN s.token3 <> ''
             AND (
                    s.token3 = t.token1
                 OR s.token3 = t.token2
                 OR s.token3 = t.token3
             )
            THEN 1 ELSE 0
        END
    ) AS name_token_score,

    -- Address number agreement
    CASE
        WHEN a.address_number <> ''
         AND b.address_number <> ''
         AND try_cast(a.address_number AS BIGINT)
             = try_cast(b.address_number AS BIGINT)
        THEN 1 ELSE 0
    END AS same_number,

    -- Street-word agreement
    CASE
        WHEN a.street_token <> ''
         AND b.street_token <> ''
         AND a.street_token = b.street_token
        THEN 1 ELSE 0
    END AS same_street,

    -- Exact canonical core name
    CASE
        WHEN vc.core_key <> ''
         AND tc.core_key <> ''
         AND vc.core_key = tc.core_key
        THEN 1 ELSE 0
    END AS same_core_name,

    -- Exact normalized address
    CASE
        WHEN va.address_key <> ''
         AND ba.address_key <> ''
         AND va.address_key = ba.address_key
        THEN 1 ELSE 0
    END AS exact_address

FROM final_validation_candidates_v2 c

JOIN val_s1_tokens s
    ON c.source1_entity_id = s.source1_entity_id

JOIN train_tokens t
    ON c.candidate_id = t.candidate_id

LEFT JOIN val_address_better a
    ON c.source1_entity_id = a.source1_entity_id

LEFT JOIN train_address_better b
    ON c.candidate_id = b.candidate_id
   AND a.country = b.country

LEFT JOIN val_core vc
    ON c.source1_entity_id = vc.source1_entity_id

LEFT JOIN train_core tc
    ON c.candidate_id = tc.candidate_id
   AND vc.country = tc.country

LEFT JOIN val_full_address va
    ON c.source1_entity_id = va.source1_entity_id

LEFT JOIN train_full_address ba
    ON c.candidate_id = ba.candidate_id
   AND va.country = ba.country
""")

print("✅ Candidate scores created")

print(
    con.execute("""
        SELECT
            COUNT(*) AS total_candidates,
            AVG(name_token_score) AS avg_name_score,
            SUM(same_number) AS number_matches,
            SUM(same_street) AS street_matches,
            SUM(same_core_name) AS core_name_matches,
            SUM(exact_address) AS exact_address_matches
        FROM scored_candidates
    """).fetchdf()
)
```

```python
# ==========================================
# CHEAP SCORE FOR CURRENT CANDIDATES
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE scored_candidates AS

SELECT
    c.source1_entity_id,
    c.candidate_id,

    -- Number of matching name tokens
    (
        CASE
            WHEN s.token1 <> ''
             AND (
                    s.token1 = t.token1
                 OR s.token1 = t.token2
                 OR s.token1 = t.token3
             )
            THEN 1 ELSE 0
        END
        +
        CASE
            WHEN s.token2 <> ''
             AND (
                    s.token2 = t.token1
                 OR s.token2 = t.token2
                 OR s.token2 = t.token3
             )
            THEN 1 ELSE 0
        END
        +
        CASE
            WHEN s.token3 <> ''
             AND (
                    s.token3 = t.token1
                 OR s.token3 = t.token2
                 OR s.token3 = t.token3
             )
            THEN 1 ELSE 0
        END
    ) AS name_token_score,

    -- Same address number
    CASE
        WHEN a.address_number <> ''
         AND b.address_number <> ''
         AND try_cast(a.address_number AS BIGINT)
             = try_cast(b.address_number AS BIGINT)
        THEN 1 ELSE 0
    END AS same_number,

    -- Same street word
    CASE
        WHEN a.street_word <> ''
         AND b.street_word <> ''
         AND a.street_word = b.street_word
        THEN 1 ELSE 0
    END AS same_street,

    -- Same canonical core name
    CASE
        WHEN vc.core_key <> ''
         AND tc.core_key <> ''
         AND vc.core_key = tc.core_key
        THEN 1 ELSE 0
    END AS same_core_name

FROM final_validation_candidates_v2 c

JOIN val_s1_tokens s
    ON c.source1_entity_id = s.source1_entity_id

JOIN train_tokens t
    ON c.candidate_id = t.candidate_id

LEFT JOIN val_addr_targeted a
    ON c.source1_entity_id = a.source1_entity_id

LEFT JOIN train_addr_targeted b
    ON c.candidate_id = b.candidate_id
   AND a.country = b.country

LEFT JOIN val_core vc
    ON c.source1_entity_id = vc.source1_entity_id

LEFT JOIN train_core tc
    ON c.candidate_id = tc.candidate_id
   AND vc.country = tc.country
""")

print("✅ Candidate scores created")

print(
    con.execute("""
        SELECT
            COUNT(*) AS total_candidates,
            AVG(name_token_score) AS avg_name_score,
            SUM(same_number) AS number_matches,
            SUM(same_street) AS street_matches,
            SUM(same_core_name) AS core_name_matches
        FROM scored_candidates
    """).fetchdf()
)
```

```python
# ==========================================
# RANK CANDIDATES + TEST TOP K
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE ranked_candidates AS

SELECT
    source1_entity_id,
    candidate_id,

    (
        3 * name_token_score
        + 3 * same_core_name
        + 2 * same_number
        + 2 * same_street
    ) AS cheap_score,

    ROW_NUMBER() OVER (
        PARTITION BY source1_entity_id
        ORDER BY
            (
                3 * name_token_score
                + 3 * same_core_name
                + 2 * same_number
                + 2 * same_street
            ) DESC,
            candidate_id
    ) AS rank_num

FROM scored_candidates
""")

# Ground truth pairs
con.execute("""
CREATE OR REPLACE TEMP TABLE gt_pairs_frozen AS

SELECT
    source1_entity_id,
    trim(x) AS candidate_id

FROM val_gt_100,

UNNEST(
    string_split(
        COALESCE(matched_entity_ids, ''),
        ','
    )
) AS u(x)

WHERE trim(x) <> ''
""")

# Test Top-K
for k in [50, 100, 200]:

    result = con.execute(f"""
    WITH topk AS (
        SELECT
            source1_entity_id,
            candidate_id
        FROM ranked_candidates
        WHERE rank_num <= {k}
    ),

    captured AS (
        SELECT
            g.source1_entity_id,
            g.candidate_id
        FROM gt_pairs_frozen g
        JOIN topk t
            ON g.source1_entity_id = t.source1_entity_id
           AND g.candidate_id = t.candidate_id
    )

    SELECT
        {k} AS top_k,

        (SELECT COUNT(*) FROM captured)
            AS captured_true_matches,

        ROUND(
            100.0 *
            (SELECT COUNT(*) FROM captured)
            /
            (SELECT COUNT(*) FROM gt_pairs_frozen),
            2
        ) AS recall_percent,

        COUNT(*) AS total_topk_candidates

    FROM topk
    """).fetchdf()

    print(result.to_string(index=False))
```

```python
# ==========================================
# IMPROVED RANKING WITH TOKEN RARITY
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE candidate_ranking_v2 AS

SELECT
    c.source1_entity_id,
    c.candidate_id,

    c.name_token_score,
    c.same_number,
    c.same_street,
    c.same_core_name,

    -- Rarity score for matching name tokens
    (
        CASE
            WHEN s.token1 <> ''
             AND (
                    s.token1 = t.token1
                 OR s.token1 = t.token2
                 OR s.token1 = t.token3
             )
            THEN 1.0 / LOG(2 + COALESCE(f1.freq, 1000000))
            ELSE 0
        END
        +
        CASE
            WHEN s.token2 <> ''
             AND (
                    s.token2 = t.token1
                 OR s.token2 = t.token2
                 OR s.token2 = t.token3
             )
            THEN 1.0 / LOG(2 + COALESCE(f2.freq, 1000000))
            ELSE 0
        END
        +
        CASE
            WHEN s.token3 <> ''
             AND (
                    s.token3 = t.token1
                 OR s.token3 = t.token2
                 OR s.token3 = t.token3
             )
            THEN 1.0 / LOG(2 + COALESCE(f3.freq, 1000000))
            ELSE 0
        END
    ) AS rarity_score

FROM scored_candidates c

JOIN val_s1_tokens s
    ON c.source1_entity_id = s.source1_entity_id

JOIN train_tokens t
    ON c.candidate_id = t.candidate_id

LEFT JOIN train_token_freq f1
    ON t.country = f1.country
   AND t.token1 = f1.token

LEFT JOIN train_token_freq f2
    ON t.country = f2.country
   AND t.token2 = f2.token

LEFT JOIN train_token_freq f3
    ON t.country = f3.country
   AND t.token3 = f3.token
""")

# Final ranking score
con.execute("""
CREATE OR REPLACE TEMP TABLE ranked_candidates_v2 AS

SELECT
    source1_entity_id,
    candidate_id,

    (
        3 * name_token_score
        + 3 * same_core_name
        + 2 * same_number
        + 2 * same_street
        + 10 * rarity_score
    ) AS ranking_score,

    ROW_NUMBER() OVER (
        PARTITION BY source1_entity_id
        ORDER BY
            (
                3 * name_token_score
                + 3 * same_core_name
                + 2 * same_number
                + 2 * same_street
                + 10 * rarity_score
            ) DESC,
            candidate_id
    ) AS rank_num

FROM candidate_ranking_v2
""")

print("✅ Improved ranking created")
```

```python
# ==========================================
# TEST IMPROVED RANKING
# ==========================================

for k in [50, 100, 200, 300]:

    result = con.execute(f"""
    WITH topk AS (
        SELECT
            source1_entity_id,
            candidate_id
        FROM ranked_candidates_v2
        WHERE rank_num <= {k}
    ),

    captured AS (
        SELECT
            g.source1_entity_id,
            g.candidate_id
        FROM gt_pairs_frozen g
        JOIN topk t
            ON g.source1_entity_id = t.source1_entity_id
           AND g.candidate_id = t.candidate_id
    )

    SELECT
        {k} AS top_k,
        (SELECT COUNT(*) FROM captured) AS captured_true_matches,
        ROUND(
            100.0 *
            (SELECT COUNT(*) FROM captured)
            /
            (SELECT COUNT(*) FROM gt_pairs_frozen),
            2
        ) AS recall_percent,
        COUNT(*) AS total_candidates
    FROM topk
    """)

    print(result.fetchdf().to_string(index=False))
```

```python
# ==========================================
# DIVERSE TOP-K RANKING
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE diverse_candidates AS

WITH ranked AS (
    SELECT
        *,
        
        -- Name-focused score
        (
            3 * name_token_score
            + 3 * same_core_name
        ) AS name_score,

        -- Address-focused score
        (
            2 * same_number
            + 2 * same_street
        ) AS address_score,

        -- Overall score
        (
            3 * name_token_score
            + 3 * same_core_name
            + 2 * same_number
            + 2 * same_street
        ) AS overall_score

    FROM scored_candidates
),

name_ranked AS (
    SELECT
        source1_entity_id,
        candidate_id
    FROM ranked

    QUALIFY ROW_NUMBER() OVER (
        PARTITION BY source1_entity_id
        ORDER BY
            name_score DESC,
            address_score DESC,
            candidate_id
    ) <= 100
),

address_ranked AS (
    SELECT
        source1_entity_id,
        candidate_id
    FROM ranked

    QUALIFY ROW_NUMBER() OVER (
        PARTITION BY source1_entity_id
        ORDER BY
            address_score DESC,
            name_score DESC,
            candidate_id
    ) <= 100
),

overall_ranked AS (
    SELECT
        source1_entity_id,
        candidate_id
    FROM ranked

    QUALIFY ROW_NUMBER() OVER (
        PARTITION BY source1_entity_id
        ORDER BY
            overall_score DESC,
            candidate_id
    ) <= 100
)

SELECT source1_entity_id, candidate_id
FROM name_ranked

UNION

SELECT source1_entity_id, candidate_id
FROM address_ranked

UNION

SELECT source1_entity_id, candidate_id
FROM overall_ranked
""")

print("✅ Diverse candidate set created")

print(
    con.execute("""
        SELECT
            COUNT(*) AS candidate_pairs,
            COUNT(DISTINCT source1_entity_id) AS s1_count
        FROM diverse_candidates
    """).fetchdf()
)
```

```python
# ==========================================
# RECALL OF DIVERSE TOP-K
# ==========================================

result = con.execute("""
WITH captured AS (
    SELECT
        g.source1_entity_id,
        g.candidate_id
    FROM gt_pairs_frozen g

    JOIN diverse_candidates d
        ON g.source1_entity_id = d.source1_entity_id
       AND g.candidate_id = d.candidate_id
)

SELECT
    (SELECT COUNT(*) FROM gt_pairs_frozen)
        AS total_true_matches,

    COUNT(*) AS captured_true_matches,

    ROUND(
        100.0 * COUNT(*)
        / (SELECT COUNT(*) FROM gt_pairs_frozen),
        2
    ) AS recall_percent,

    (SELECT COUNT(*) FROM diverse_candidates)
        AS total_candidate_pairs
FROM captured
""").fetchdf()

print(result)
```

```python
# ==========================================
# FULL CORE NAME TEST ON CURRENT 56 MISSES
# ==========================================

current_misses = con.execute("""
    SELECT source1_entity_id, candidate_id
    FROM remaining_misses_v7
""").fetchdf()

# miss_names already contains the names of the earlier missed pairs
current = miss_names.merge(
    current_misses,
    on=["source1_entity_id", "candidate_id"],
    how="inner"
).copy()

LEGAL_WORDS = {
    "inc", "incorporated",
    "llc", "ltd", "limited",
    "private", "pvt",
    "plc", "corp", "corporation",
    "company", "co", "pc"
}

def full_core(text):
    if pd.isna(text):
        return set()

    words = re.findall(
        r'[^\W\d_]+',
        str(text).lower(),
        flags=re.UNICODE
    )

    return {
        w for w in words
        if len(w) >= 3
        and w not in LEGAL_WORDS
    }

current["s1_full_core"] = current["s1_name"].apply(full_core)
current["candidate_full_core"] = current["candidate_name"].apply(full_core)

current["core_equal"] = current.apply(
    lambda r: r["s1_full_core"] == r["candidate_full_core"],
    axis=1
)

current["core_subset"] = current.apply(
    lambda r:
        r["s1_full_core"].issubset(r["candidate_full_core"])
        or
        r["candidate_full_core"].issubset(r["s1_full_core"]),
    axis=1
)

print("Current remaining misses:", len(current))

print(
    "Exact full-core match:",
    current["core_equal"].sum()
)

print(
    "Core subset match:",
    current["core_subset"].sum()
)

print("\nExamples:")
print(
    current[
        ["s1_name", "candidate_name", "s1_full_core", "candidate_full_core"]
    ].head(15).to_string(index=False)
)
```

```python
# ==========================================
# COMPACT / WEBSITE NAME TEST
# ==========================================

import re

def compact_business_name(text):
    if pd.isna(text):
        return ""

    text = str(text).lower().strip()

    # Remove common website markers
    text = re.sub(r'^https?://', '', text)
    text = re.sub(r'^www\.', '', text)
    text = text.replace('@', '')

    # Remove common domain suffixes
    text = re.sub(
        r'\.(com|org|net|co|in|us|fr|biz)$',
        '',
        text
    )

    # Keep only letters and digits
    text = re.sub(
        r'[^\w]',
        '',
        text,
        flags=re.UNICODE
    )

    return text


current["s1_compact"] = current["s1_name"].apply(
    compact_business_name
)

current["candidate_compact"] = current[
    "candidate_name"
].apply(
    compact_business_name
)

current["compact_equal"] = current.apply(
    lambda r:
        r["s1_compact"] != ""
        and r["candidate_compact"] != ""
        and r["s1_compact"] == r["candidate_compact"],
    axis=1
)

current["compact_contained"] = current.apply(
    lambda r:
        r["s1_compact"] != ""
        and r["candidate_compact"] != ""
        and (
            r["s1_compact"] in r["candidate_compact"]
            or
            r["candidate_compact"] in r["s1_compact"]
        ),
    axis=1
)

print("Current remaining misses:", len(current))

print(
    "Exact compact-name match:",
    current["compact_equal"].sum()
)

print(
    "One compact name contains the other:",
    current["compact_contained"].sum()
)

print("\nExamples:")
print(
    current[
        [
            "s1_name",
            "candidate_name",
            "s1_compact",
            "candidate_compact"
        ]
    ].head(20).to_string(index=False)
)
```

```python
# ==========================================
# TRANSLITERATED COMPACT NAME TEST
# ==========================================

from unidecode import unidecode

def translit_compact(text):
    if pd.isna(text):
        return ""

    x = unidecode(str(text).lower())

    # keep only letters/digits
    x = re.sub(r'[^a-z0-9]+', '', x)

    return x


current["s1_translit_compact"] = current["s1_name"].apply(
    translit_compact
)

current["candidate_translit_compact"] = current[
    "candidate_name"
].apply(
    translit_compact
)

current["translit_equal"] = current.apply(
    lambda r:
        r["s1_translit_compact"] != ""
        and r["candidate_translit_compact"] != ""
        and r["s1_translit_compact"]
            == r["candidate_translit_compact"],
    axis=1
)

current["translit_contained"] = current.apply(
    lambda r:
        r["s1_translit_compact"] != ""
        and r["candidate_translit_compact"] != ""
        and (
            r["s1_translit_compact"]
                in r["candidate_translit_compact"]
            or
            r["candidate_translit_compact"]
                in r["s1_translit_compact"]
        ),
    axis=1
)

print("Current remaining misses:", len(current))

print(
    "Exact transliterated match:",
    current["translit_equal"].sum()
)

print(
    "Transliterated containment:",
    current["translit_contained"].sum()
)

print("\nExamples:")
print(
    current[
        [
            "s1_name",
            "candidate_name",
            "s1_translit_compact",
            "candidate_translit_compact"
        ]
    ].head(20).to_string(index=False)
)
```

```python
# ==========================================
# 4-GRAM FREQUENCY ANALYSIS
# ==========================================

def compact_ascii(text):
    if pd.isna(text):
        return ""

    x = unidecode(str(text).lower())
    x = re.sub(r'[^a-z0-9]+', '', x)
    return x


def get_4grams(text):
    text = compact_ascii(text)

    if len(text) < 4:
        return set()

    return {
        text[i:i+4]
        for i in range(len(text) - 3)
    }


# 4-grams for current 56 true misses
current["s1_4grams"] = current["s1_name"].apply(get_4grams)
current["candidate_4grams"] = current["candidate_name"].apply(get_4grams)

current["shared_4grams"] = current.apply(
    lambda r:
        r["s1_4grams"] & r["candidate_4grams"],
    axis=1
)

print("Current misses:", len(current))

for threshold in [1, 2, 3, 5, 8, 10]:

    print(
        f"Shared 4-grams >= {threshold}:",
        (
            current["shared_4grams"].str.len()
            >= threshold
        ).sum()
    )
```

```python
# ==========================================
# POSITIONAL 4-CHAR SIGNATURE TEST
# ON CURRENT 56 MISSES
# ==========================================

def positional_4grams(text):
    if pd.isna(text):
        return set()

    x = unidecode(str(text).lower())
    x = re.sub(r'[^a-z0-9]+', '', x)

    if len(x) < 4:
        return set()

    result = {
        x[:4],
        x[-4:]
    }

    mid = len(x) // 2
    result.add(x[max(0, mid-2):mid+2])

    return {g for g in result if len(g) == 4}


current["s1_pos4"] = current["s1_name"].apply(positional_4grams)
current["candidate_pos4"] = current[
    "candidate_name"
].apply(positional_4grams)

current["common_pos4"] = current.apply(
    lambda r:
        r["s1_pos4"] & r["candidate_pos4"],
    axis=1
)

print("Current misses:", len(current))

print(
    "At least 1 positional 4-char match:",
    (current["common_pos4"].str.len() >= 1).sum()
)

print(
    "At least 2 positional 4-char matches:",
    (current["common_pos4"].str.len() >= 2).sum()
)

print("\nExamples:")
print(
    current[
        ["s1_name", "candidate_name", "common_pos4"]
    ].head(20).to_string(index=False)
)
```

```python
# ==========================================
# TEST EVIDENCE SCORE THRESHOLDS
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE evidence_scored AS

SELECT
    source1_entity_id,
    candidate_id,

    (
        3 * name_token_score
        + 3 * same_core_name
        + 2 * same_number
        + 2 * same_street
    ) AS evidence_score

FROM scored_candidates
""")

for threshold in [1, 2, 3, 4, 5, 6, 7, 8]:

    result = con.execute(f"""
    WITH filtered AS (
        SELECT
            source1_entity_id,
            candidate_id
        FROM evidence_scored
        WHERE evidence_score >= {threshold}
    ),

    captured AS (
        SELECT
            g.source1_entity_id,
            g.candidate_id
        FROM gt_pairs_frozen g
        JOIN filtered f
            ON g.source1_entity_id = f.source1_entity_id
           AND g.candidate_id = f.candidate_id
    )

    SELECT
        {threshold} AS min_score,

        (SELECT COUNT(*) FROM filtered)
            AS candidate_pairs,

        (SELECT COUNT(*) FROM captured)
            AS captured_true_matches,

        ROUND(
            100.0 *
            (SELECT COUNT(*) FROM captured)
            /
            (SELECT COUNT(*) FROM gt_pairs_frozen),
            2
        ) AS recall_percent
    """).fetchdf()

    print(result.to_string(index=False))
```

```python
# ==========================================
# ENHANCED CORE NAME TEST
# ==========================================

ENHANCED_STOPWORDS = {
    # Legal/business suffixes
    "inc", "incorporated", "llc", "ltd", "limited",
    "private", "pvt", "plc", "corp", "corporation",
    "company", "co", "pc", "pllc", "llp", "lp",
    
    # Professional/company designation noise
    "md", "do", "dds", "dmd", "dvm", "phd",
    "rn", "pa", "apc", "esq",
    
    # Common connector words
    "and", "of", "the"
}

def enhanced_core(text):
    if pd.isna(text):
        return set()

    words = re.findall(
        r'[^\W\d_]+',
        str(text).lower(),
        flags=re.UNICODE
    )

    return {
        w for w in words
        if len(w) >= 3
        and w not in ENHANCED_STOPWORDS
    }

current["s1_enhanced_core"] = current["s1_name"].apply(
    enhanced_core
)

current["candidate_enhanced_core"] = current[
    "candidate_name"
].apply(enhanced_core)

current["enhanced_exact"] = current.apply(
    lambda r:
        r["s1_enhanced_core"]
        == r["candidate_enhanced_core"],
    axis=1
)

current["enhanced_subset"] = current.apply(
    lambda r:
        r["s1_enhanced_core"].issubset(
            r["candidate_enhanced_core"]
        )
        or
        r["candidate_enhanced_core"].issubset(
            r["s1_enhanced_core"]
        ),
    axis=1
)

print("Current difficult misses:", len(current))

print(
    "Enhanced exact core matches:",
    current["enhanced_exact"].sum()
)

print(
    "Enhanced core subset matches:",
    current["enhanced_subset"].sum()
)

print("\nExamples:")
print(
    current[
        [
            "s1_name",
            "candidate_name",
            "s1_enhanced_core",
            "candidate_enhanced_core"
        ]
    ].head(20).to_string(index=False)
)
```

```python
# ==========================================
# COMPACT 8-CHARACTER PREFIX TEST
# ==========================================

def compact8(text):
    if pd.isna(text):
        return ""

    x = unidecode(str(text).lower())
    x = re.sub(r'[^a-z0-9]+', '', x)

    return x[:8]


current["s1_compact8"] = current["s1_name"].apply(compact8)
current["candidate_compact8"] = current["candidate_name"].apply(compact8)

current["same_compact8"] = current.apply(
    lambda r:
        r["s1_compact8"] != ""
        and r["candidate_compact8"] != ""
        and r["s1_compact8"] == r["candidate_compact8"],
    axis=1
)

print("Current difficult misses:", len(current))

print(
    "Same compact 8-character prefix:",
    current["same_compact8"].sum()
)

print("\nExamples:")
print(
    current[
        [
            "s1_name",
            "candidate_name",
            "s1_compact8",
            "candidate_compact8"
        ]
    ].head(20).to_string(index=False)
)
```

```python
# ==========================================
# COMPACT-8 PREFIX BLOCKER
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE train_compact8 AS

SELECT
    entity_id AS candidate_id,
    country,

    substr(
        regexp_replace(
            lower(coalesce(business_name, '')),
            '[^a-z0-9]+',
            '',
            'g'
        ),
        1,
        8
    ) AS compact8

FROM (
    SELECT entity_id, country, business_name
    FROM read_csv(
        'train_source2.tsv',
        delim='\t',
        header=true
    )

    UNION ALL

    SELECT entity_id, country, business_name
    FROM read_csv(
        'train_source3.tsv',
        delim='\t',
        header=true
    )
)
""")

con.execute("""
CREATE OR REPLACE TEMP TABLE val_compact8 AS

SELECT
    entity_id AS source1_entity_id,
    country,

    substr(
        regexp_replace(
            lower(coalesce(business_name, '')),
            '[^a-z0-9]+',
            '',
            'g'
        ),
        1,
        8
    ) AS compact8

FROM read_csv(
    'train_source1.tsv',
    delim='\t',
    header=true
)

WHERE entity_id IN (
    SELECT source1_entity_id
    FROM val_gt_100
)
""")

result = con.execute("""
SELECT
    COUNT(*) AS candidate_pairs,
    COUNT(DISTINCT v.source1_entity_id) AS s1_count

FROM val_compact8 v

JOIN train_compact8 t
    ON v.country = t.country
   AND v.compact8 <> ''
   AND t.compact8 <> ''
   AND v.compact8 = t.compact8
""").fetchdf()

print("✅ Compact-8 candidate size checked")
print(result)
```

```python
# ==========================================
# ADAPTIVE CANDIDATE FILTER
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE adaptive_candidates AS

WITH scored AS (
    SELECT
        *,
        (
            3 * name_token_score
            + 3 * same_core_name
            + 2 * same_number
            + 2 * same_street
        ) AS evidence_score
    FROM scored_candidates
),

ranked AS (
    SELECT
        *,
        ROW_NUMBER() OVER (
            PARTITION BY source1_entity_id
            ORDER BY
                evidence_score DESC,
                candidate_id
        ) AS rn
    FROM scored
)

SELECT
    source1_entity_id,
    candidate_id

FROM ranked

WHERE
    -- Strong evidence: always keep
       name_token_score >= 2

    OR same_core_name = 1

    OR (same_number = 1 AND same_street = 1)

    OR rn <= 100
""")

print("✅ Adaptive candidates created")

print(
    con.execute("""
        SELECT
            COUNT(*) AS candidate_pairs,
            COUNT(DISTINCT source1_entity_id) AS s1_count,
            AVG(cnt) AS avg_candidates_per_s1,
            MAX(cnt) AS max_candidates_per_s1
        FROM (
            SELECT
                source1_entity_id,
                COUNT(*) AS cnt
            FROM adaptive_candidates
            GROUP BY source1_entity_id
        )
    """).fetchdf()
)
```

```python
# ==========================================
# CORRECT ADAPTIVE BLOCKER BENCHMARK
# ==========================================

result = con.execute("""
WITH captured AS (
    SELECT
        g.source1_entity_id,
        g.candidate_id

    FROM gt_pairs_frozen g

    JOIN adaptive_candidates a
        ON g.source1_entity_id = a.source1_entity_id
       AND g.candidate_id = a.candidate_id
)

SELECT
    (SELECT COUNT(*) FROM adaptive_candidates)
        AS actual_candidate_pairs,

    (SELECT COUNT(DISTINCT source1_entity_id)
     FROM adaptive_candidates)
        AS s1_count,

    ROUND(
        (SELECT COUNT(*) FROM adaptive_candidates) * 1.0
        / (SELECT COUNT(DISTINCT source1_entity_id)
           FROM adaptive_candidates),
        2
    ) AS avg_candidates_per_s1,

    (SELECT MAX(cnt)
     FROM (
        SELECT source1_entity_id, COUNT(*) AS cnt
        FROM adaptive_candidates
        GROUP BY source1_entity_id
     ))
        AS max_candidates_per_s1,

    (SELECT COUNT(*) FROM captured)
        AS captured_true_matches,

    ROUND(
        100.0 *
        (SELECT COUNT(*) FROM captured)
        / (SELECT COUNT(*) FROM gt_pairs_frozen),
        2
    ) AS recall_percent
""").fetchdf()

print(result.to_string(index=False))
```

```python
# ==========================================
# RARE COMPACT-8 BLOCKING TEST
# ==========================================

# Frequency of compact8 in S2 + S3
con.execute("""
CREATE OR REPLACE TEMP TABLE compact8_freq AS

SELECT
    country,
    compact8,
    COUNT(*) AS freq
FROM train_compact8
WHERE compact8 <> ''
GROUP BY country, compact8
""")

# Current true matches NOT captured by our final blocker
con.execute("""
CREATE OR REPLACE TEMP TABLE final_misses AS

WITH gt_pairs AS (
    SELECT
        source1_entity_id,
        trim(x) AS candidate_id
    FROM val_gt_100,
    UNNEST(
        string_split(
            COALESCE(matched_entity_ids, ''),
            ','
        )
    ) AS u(x)
    WHERE trim(x) <> ''
)

SELECT
    g.source1_entity_id,
    g.candidate_id
FROM gt_pairs g

LEFT JOIN final_validation_candidates_v2 c
    ON g.source1_entity_id = c.source1_entity_id
   AND g.candidate_id = c.candidate_id

WHERE c.candidate_id IS NULL
""")

for limit in [5, 10, 25, 50]:

    result = con.execute(f"""
    WITH candidates AS (

        SELECT DISTINCT
            v.source1_entity_id,
            t.candidate_id

        FROM val_compact8 v

        JOIN compact8_freq f
            ON v.country = f.country
           AND v.compact8 = f.compact8
           AND f.freq <= {limit}

        JOIN train_compact8 t
            ON v.country = t.country
           AND v.compact8 = t.compact8

        WHERE v.compact8 <> ''
    ),

    new_true_matches AS (

        SELECT
            m.source1_entity_id,
            m.candidate_id

        FROM final_misses m

        JOIN candidates c
            ON m.source1_entity_id = c.source1_entity_id
           AND m.candidate_id = c.candidate_id
    )

    SELECT
        {limit} AS max_frequency,

        (SELECT COUNT(*) FROM candidates)
            AS candidate_pairs,

        (SELECT COUNT(*)
         FROM new_true_matches)
            AS new_true_matches
    """)

    print(result.fetchdf().to_string(index=False))
```

```python
# ==========================================
# ANY-POSITION INFORMATIVE NAME TOKEN
# ON CURRENT 56 MISSES
# ==========================================

import re
import pandas as pd

STOPWORDS = {
    "inc", "incorporated", "llc", "ltd", "limited",
    "private", "pvt", "plc", "corp", "corporation",
    "company", "co", "pc", "pllc", "llp", "lp",
    "and", "of", "the", "for", "group",
    "services", "service"
}

def all_informative_words(text):
    if pd.isna(text):
        return set()

    words = re.findall(
        r'[^\W\d_]+',
        str(text).lower(),
        flags=re.UNICODE
    )

    return {
        w for w in words
        if len(w) >= 4
        and w not in STOPWORDS
    }

current["all_s1_words"] = current["s1_name"].apply(
    all_informative_words
)

current["all_candidate_words"] = current["candidate_name"].apply(
    all_informative_words
)

current["all_common_words"] = current.apply(
    lambda r:
        r["all_s1_words"] & r["all_candidate_words"],
    axis=1
)

print("Current misses:", len(current))

print(
    "At least 1 common informative word:",
    (current["all_common_words"].str.len() >= 1).sum()
)

print(
    "At least 2 common informative words:",
    (current["all_common_words"].str.len() >= 2).sum()
)

print("\nExamples:")
print(
    current[
        ["s1_name", "candidate_name", "all_common_words"]
    ].head(20).to_string(index=False)
)
```

```python
# ==========================================
# FETCH NAMES + ADDRESSES FOR CURRENT
# 181K VALIDATION CANDIDATES
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE fuzzy_candidate_data AS

SELECT
    c.source1_entity_id,
    c.candidate_id,

    s.business_name AS s1_name,
    s.business_address AS s1_address,
    s.country AS s1_country,

    t.business_name AS candidate_name,
    t.business_address AS candidate_address,
    t.country AS candidate_country

FROM final_validation_candidates_v2 c

JOIN read_csv(
    'train_source1.tsv',
    delim='\t',
    header=true
) s
    ON c.source1_entity_id = s.entity_id

JOIN (
    SELECT
        entity_id,
        business_name,
        business_address,
        country
    FROM read_csv(
        'train_source2.tsv',
        delim='\t',
        header=true
    )

    UNION ALL

    SELECT
        entity_id,
        business_name,
        business_address,
        country
    FROM read_csv(
        'train_source3.tsv',
        delim='\t',
        header=true
    )
) t
    ON c.candidate_id = t.entity_id
""")

print("✅ Fuzzy candidate data created")

print(
    con.execute("""
        SELECT
            COUNT(*) AS rows,
            COUNT(DISTINCT source1_entity_id) AS s1_count
        FROM fuzzy_candidate_data
    """).fetchdf()
)
```

```python
# ==========================================
# LOAD DUCKDB TABLE INTO PANDAS
# ==========================================

fuzzy_candidate_data = con.execute("""
    SELECT *
    FROM fuzzy_candidate_data
""").fetchdf()

print("✅ Fuzzy candidate DataFrame loaded")
print(fuzzy_candidate_data.shape)
```

```python
# ==========================================
# FUZZY NAME SCORES - FAST VERSION
# ==========================================

from rapidfuzz.fuzz import WRatio, token_set_ratio

# Clean names
s1_names = (
    fuzzy_candidate_data["s1_name"]
    .fillna("")
    .astype(str)
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

candidate_names = (
    fuzzy_candidate_data["candidate_name"]
    .fillna("")
    .astype(str)
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

# Calculate fuzzy similarities
fuzzy_candidate_data["name_wratio"] = [
    WRatio(a, b)
    for a, b in zip(s1_names, candidate_names)
]

fuzzy_candidate_data["name_token_set"] = [
    token_set_ratio(a, b)
    for a, b in zip(s1_names, candidate_names)
]

print("✅ Fuzzy name scores calculated")

print(
    fuzzy_candidate_data[
        ["name_wratio", "name_token_set"]
    ].describe()
)
```

```python
# ==========================================
# FUZZY NAME THRESHOLD TEST
# ==========================================

# Best of the two fuzzy name scores
fuzzy_candidate_data["fuzzy_name_score"] = (
    fuzzy_candidate_data[
        ["name_wratio", "name_token_set"]
    ].max(axis=1)
)

# Ground-truth pairs
gt_pairs = con.execute("""
    SELECT
        source1_entity_id,
        trim(x) AS candidate_id

    FROM val_gt_100,

    UNNEST(
        string_split(
            COALESCE(matched_entity_ids, ''),
            ','
        )
    ) AS u(x)

    WHERE trim(x) <> ''
""").fetchdf()

for threshold in [50, 60, 70, 75, 80, 85, 90]:

    filtered = fuzzy_candidate_data[
        fuzzy_candidate_data["fuzzy_name_score"] >= threshold
    ]

    # Convert to set for fast lookup
    candidate_set = set(
        zip(
            filtered["source1_entity_id"],
            filtered["candidate_id"]
        )
    )

    captured = sum(
        (row.source1_entity_id, row.candidate_id)
        in candidate_set
        for row in gt_pairs.itertuples()
    )

    total_candidates = len(filtered)

    recall = (
        100.0 * captured / len(gt_pairs)
    )

    print(
        f"Threshold >= {threshold:2d} | "
        f"Candidates = {total_candidates:6d} | "
        f"Captured = {captured:3d} | "
        f"Recall = {recall:6.2f}%"
    )
```

```python
# ==========================================
# FUZZY ADDRESS SCORES
# ==========================================

from rapidfuzz.fuzz import WRatio, token_set_ratio

s1_addresses = (
    fuzzy_candidate_data["s1_address"]
    .fillna("")
    .astype(str)
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

candidate_addresses = (
    fuzzy_candidate_data["candidate_address"]
    .fillna("")
    .astype(str)
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

# Address WRatio
fuzzy_candidate_data["address_wratio"] = [
    WRatio(a, b)
    for a, b in zip(s1_addresses, candidate_addresses)
]

# Address token-set similarity
fuzzy_candidate_data["address_token_set"] = [
    token_set_ratio(a, b)
    for a, b in zip(s1_addresses, candidate_addresses)
]

print("✅ Address fuzzy scores calculated")

print(
    fuzzy_candidate_data[
        ["address_wratio", "address_token_set"]
    ].describe()
)
```

```python
# ==========================================
# ADD EXISTING EVIDENCE FEATURES
# ==========================================

evidence_features = con.execute("""
    SELECT
        source1_entity_id,
        candidate_id,
        name_token_score,
        same_number,
        same_street,
        same_core_name
    FROM scored_candidates
""").fetchdf()

fuzzy_candidate_data = fuzzy_candidate_data.merge(
    evidence_features,
    on=["source1_entity_id", "candidate_id"],
    how="left"
)

print("✅ Evidence features merged")

print(
    fuzzy_candidate_data[
        [
            "name_token_score",
            "same_number",
            "same_street",
            "same_core_name"
        ]
    ].head()
)
```

```python
# ==========================================
# COMBINED FUZZY + EVIDENCE SCORE
# ==========================================

# Best fuzzy score for name
fuzzy_candidate_data["fuzzy_name_score"] = (
    fuzzy_candidate_data[
        ["name_wratio", "name_token_set"]
    ].max(axis=1)
)

# Best fuzzy score for address
fuzzy_candidate_data["fuzzy_address_score"] = (
    fuzzy_candidate_data[
        ["address_wratio", "address_token_set"]
    ].max(axis=1)
)

# Combined ranking score
fuzzy_candidate_data["combined_score"] = (
    0.35 * fuzzy_candidate_data["fuzzy_name_score"]
    + 0.25 * fuzzy_candidate_data["fuzzy_address_score"]
    + 8 * fuzzy_candidate_data["same_number"]
    + 5 * fuzzy_candidate_data["same_street"]
    + 8 * fuzzy_candidate_data["same_core_name"]
)

print("✅ Combined score created")

print(
    fuzzy_candidate_data[
        [
            "fuzzy_name_score",
            "fuzzy_address_score",
            "same_number",
            "same_street",
            "same_core_name",
            "combined_score"
        ]
    ].describe()
)
```

```python
# ==========================================
# TEST COMBINED FUZZY RANKING
# ==========================================

# Rank candidates for each S1
ranked_fuzzy = fuzzy_candidate_data.sort_values(
    by=[
        "source1_entity_id",
        "combined_score",
        "fuzzy_name_score",
        "fuzzy_address_score"
    ],
    ascending=[True, False, False, False]
).copy()

ranked_fuzzy["rank"] = (
    ranked_fuzzy
    .groupby("source1_entity_id")
    .cumcount() + 1
)

# Ground truth
gt_pairs = con.execute("""
    SELECT
        source1_entity_id,
        trim(x) AS candidate_id

    FROM val_gt_100,

    UNNEST(
        string_split(
            COALESCE(matched_entity_ids, ''),
            ','
        )
    ) AS u(x)

    WHERE trim(x) <> ''
""").fetchdf()

for k in [50, 100, 200, 300, 500]:

    topk = ranked_fuzzy[
        ranked_fuzzy["rank"] <= k
    ]

    candidate_set = set(
        zip(
            topk["source1_entity_id"],
            topk["candidate_id"]
        )
    )

    captured = sum(
        (row.source1_entity_id, row.candidate_id)
        in candidate_set
        for row in gt_pairs.itertuples()
    )

    print(
        f"Top {k:3d} | "
        f"Candidates = {len(topk):6d} | "
        f"Captured = {captured:3d} | "
        f"Recall = {100 * captured / len(gt_pairs):6.2f}%"
    )
```

```python
# ==========================================
# RANK FUSION
# NAME + ADDRESS + EVIDENCE
# ==========================================

for k in [50, 100, 200]:

    name_top = (
        fuzzy_candidate_data
        .sort_values(
            ["source1_entity_id", "fuzzy_name_score"],
            ascending=[True, False]
        )
        .groupby("source1_entity_id")
        .head(k)
    )

    address_top = (
        fuzzy_candidate_data
        .sort_values(
            ["source1_entity_id", "fuzzy_address_score"],
            ascending=[True, False]
        )
        .groupby("source1_entity_id")
        .head(k)
    )

    evidence_top = (
        fuzzy_candidate_data
        .sort_values(
            [
                "source1_entity_id",
                "same_core_name",
                "same_number",
                "same_street",
                "name_token_score"
            ],
            ascending=[True, False, False, False, False]
        )
        .groupby("source1_entity_id")
        .head(k)
    )

    fused = pd.concat(
        [
            name_top[
                ["source1_entity_id", "candidate_id"]
            ],
            address_top[
                ["source1_entity_id", "candidate_id"]
            ],
            evidence_top[
                ["source1_entity_id", "candidate_id"]
            ]
        ]
    ).drop_duplicates()

    candidate_set = set(
        zip(
            fused["source1_entity_id"],
            fused["candidate_id"]
        )
    )

    captured = sum(
        (row.source1_entity_id, row.candidate_id)
        in candidate_set
        for row in gt_pairs.itertuples()
    )

    print(
        f"Rank fusion K={k:3d} | "
        f"Candidates={len(fused):6d} | "
        f"Captured={captured:3d} | "
        f"Recall={100 * captured / len(gt_pairs):6.2f}%"
    )
```

```python
# ==========================================
# LAST MEANINGFUL ADDRESS WORD TEST
# ==========================================

def last_meaningful_words(text):
    if pd.isna(text):
        return set()

    words = re.findall(
        r'[^\W\d_]+',
        str(text).lower(),
        flags=re.UNICODE
    )

    ignore = {
        "the", "and", "of", "road", "rd",
        "street", "st", "lane", "ln",
        "avenue", "ave", "floor",
        "private", "limited", "inc",
        "llc", "corp", "company"
    }

    words = [
        w for w in words
        if len(w) >= 4 and w not in ignore
    ]

    # Last 3 meaningful words
    return set(words[-3:])


# Use the current 55 missed-pair addresses
current_miss_addr = con.execute("""
WITH records AS (
    SELECT entity_id, business_address
    FROM read_csv(
        'train_source2.tsv',
        delim='\t',
        header=true
    )

    UNION ALL

    SELECT entity_id, business_address
    FROM read_csv(
        'train_source3.tsv',
        delim='\t',
        header=true
    )
)

SELECT
    m.source1_entity_id,
    m.candidate_id,
    s.business_address AS s1_address,
    r.business_address AS candidate_address

FROM remaining_misses_v7 m

JOIN read_csv(
    'train_source1.tsv',
    delim='\t',
    header=true
) s
    ON m.source1_entity_id = s.entity_id

JOIN records r
    ON m.candidate_id = r.entity_id
""").fetchdf()


current_miss_addr["s1_last_words"] = (
    current_miss_addr["s1_address"]
    .apply(last_meaningful_words)
)

current_miss_addr["candidate_last_words"] = (
    current_miss_addr["candidate_address"]
    .apply(last_meaningful_words)
)

current_miss_addr["common_last_words"] = (
    current_miss_addr.apply(
        lambda r:
            r["s1_last_words"] &
            r["candidate_last_words"],
        axis=1
    )
)

print(
    "Current misses:",
    len(current_miss_addr)
)

print(
    "At least 1 common last address word:",
    (
        current_miss_addr["common_last_words"].str.len()
        >= 1
    ).sum()
)

print(
    "At least 2 common last address words:",
    (
        current_miss_addr["common_last_words"].str.len()
        >= 2
    ).sum()
)
```

```python
# ==========================================
# EXACT LAST-2 ADDRESS WORD TEST
# ON CURRENT 56 MISSES
# ==========================================

def last_two_words(text):
    if pd.isna(text):
        return ()

    words = re.findall(
        r'[^\W\d_]+',
        str(text).lower(),
        flags=re.UNICODE
    )

    ignore = {
        "the", "and", "of", "road", "rd",
        "street", "st", "lane", "ln",
        "avenue", "ave", "floor",
        "private", "limited", "inc",
        "llc", "corp", "company"
    }

    words = [
        w for w in words
        if len(w) >= 4 and w not in ignore
    ]

    return tuple(words[-2:])


current_miss_addr["s1_last2"] = (
    current_miss_addr["s1_address"]
    .apply(last_two_words)
)

current_miss_addr["candidate_last2"] = (
    current_miss_addr["candidate_address"]
    .apply(last_two_words)
)

current_miss_addr["exact_last2"] = (
    current_miss_addr["s1_last2"]
    == current_miss_addr["candidate_last2"]
)

current_miss_addr["unordered_last2"] = (
    current_miss_addr["s1_last2"].apply(set)
    ==
    current_miss_addr["candidate_last2"].apply(set)
)

print("Current misses:", len(current_miss_addr))

print(
    "Exact same last 2 words:",
    current_miss_addr["exact_last2"].sum()
)

print(
    "Same last 2 words ignoring order:",
    current_miss_addr["unordered_last2"].sum()
)

print("\nExamples:")
print(
    current_miss_addr[
        [
            "s1_address",
            "candidate_address",
            "s1_last2",
            "candidate_last2"
        ]
    ].head(20).to_string(index=False)
)
```

```python
# ==========================================
# RARE MEANINGFUL ADDRESS WORDS
# ON CURRENT 56 MISSES
# ==========================================

import re
import pandas as pd

ADDRESS_STOPWORDS = {
    "the", "and", "of", "road", "street", "st",
    "rd", "lane", "ln", "avenue", "ave",
    "floor", "unit", "no", "near", "city",
    "district", "county", "state"
}

def all_address_words(text):
    if pd.isna(text):
        return set()

    words = re.findall(
        r'[^\W\d_]+',
        str(text).lower(),
        flags=re.UNICODE
    )

    return {
        w for w in words
        if len(w) >= 4
        and w not in ADDRESS_STOPWORDS
    }


current_miss_addr["s1_addr_words"] = (
    current_miss_addr["s1_address"]
    .apply(all_address_words)
)

current_miss_addr["candidate_addr_words"] = (
    current_miss_addr["candidate_address"]
    .apply(all_address_words)
)

current_miss_addr["common_addr_words"] = (
    current_miss_addr.apply(
        lambda r:
            r["s1_addr_words"] &
            r["candidate_addr_words"],
        axis=1
    )
)

print("Current misses:", len(current_miss_addr))

for n in [1, 2, 3, 4]:

    print(
        f"At least {n} common meaningful address words:",
        (
            current_miss_addr["common_addr_words"].str.len()
            >= n
        ).sum()
    )
```

```python
# ==========================================
# IMPORTANT 4-GRAMS IN CURRENT 55 MISSES
# ==========================================

from collections import Counter

# Get current 55 true-miss names
current55_names = miss_names.merge(
    con.execute("""
        SELECT
            source1_entity_id,
            candidate_id
        FROM current_final_misses
    """).fetchdf(),
    on=["source1_entity_id", "candidate_id"],
    how="inner"
).copy()

def get_4gram_list(text):
    if pd.isna(text):
        return []

    x = unidecode(str(text).lower())
    x = re.sub(r'[^a-z0-9]+', '', x)

    if len(x) < 4:
        return []

    return list({
        x[i:i+4]
        for i in range(len(x) - 3)
    })


# Count in how many S1/candidate pairs
# each 4-gram appears in BOTH names
pair_gram_counts = Counter()

for _, row in current55_names.iterrows():

    s1_grams = set(get_4gram_list(row["s1_name"]))
    cand_grams = set(get_4gram_list(row["candidate_name"]))

    shared = s1_grams & cand_grams

    for gram in shared:
        pair_gram_counts[gram] += 1


print("Current 55 difficult pairs:", len(current55_names))

print("\nMost useful shared 4-grams:")

for gram, count in pair_gram_counts.most_common(30):
    print(f"{gram:8s} → {count} pairs")
```

```python
# ==========================================
# INFORMATIVE 4-GRAMS
# REMOVE BUSINESS/LEGAL WORDS FIRST
# ==========================================

LEGAL_WORDS = {
    "inc", "incorporated",
    "llc", "ltd", "limited",
    "private", "pvt",
    "plc", "corp", "corporation",
    "company", "co", "pc",
    "pllc", "llp", "lp",
    "group", "services", "service",
    "and", "of", "the"
}

def informative_4grams(text):
    if pd.isna(text):
        return set()

    # Transliterate first
    x = unidecode(str(text).lower())

    # Keep words
    words = re.findall(r'[a-z0-9]+', x)

    # Remove generic business words
    words = [
        w for w in words
        if w not in LEGAL_WORDS
        and len(w) >= 3
    ]

    # Join informative words
    x = "".join(words)

    if len(x) < 4:
        return set()

    return {
        x[i:i+4]
        for i in range(len(x) - 3)
    }


current["s1_info4"] = current["s1_name"].apply(
    informative_4grams
)

current["candidate_info4"] = current[
    "candidate_name"
].apply(
    informative_4grams
)

current["shared_info4"] = current.apply(
    lambda r:
        r["s1_info4"] & r["candidate_info4"],
    axis=1
)

print("Current difficult misses:", len(current))

for threshold in [1, 2, 3, 5, 8, 10]:

    print(
        f"Informative 4-grams >= {threshold}:",
        (
            current["shared_info4"].str.len()
            >= threshold
        ).sum()
    )

print("\nMost useful 4-grams:")

from collections import Counter

counter = Counter()

for grams in current["shared_info4"]:
    counter.update(grams)

for gram, count in counter.most_common(20):
    print(f"{gram:8s} -> {count} pairs")
```

```python
# ==========================================
# LONGEST-WORD 4-CHAR SIGNATURE TEST
# ==========================================

NAME_STOPWORDS = {
    "inc", "incorporated",
    "llc", "ltd", "limited",
    "private", "pvt",
    "plc", "corp", "corporation",
    "company", "co", "pc",
    "pllc", "llp", "lp",
    "group", "services", "service",
    "and", "of", "the"
}

def longest_word_signatures(text):
    if pd.isna(text):
        return set()

    words = re.findall(
        r'[a-zA-Z0-9]+',
        str(text).lower()
    )

    words = [
        w for w in words
        if len(w) >= 4
        and w not in NAME_STOPWORDS
    ]

    if not words:
        return set()

    # Longest meaningful word
    word = max(words, key=len)

    if len(word) < 4:
        return set()

    signatures = {
        word[:4],
        word[-4:]
    }

    return signatures


current["sigs_s1"] = current["s1_name"].apply(
    longest_word_signatures
)

current["sigs_candidate"] = current[
    "candidate_name"
].apply(
    longest_word_signatures
)

current["common_longest_sigs"] = current.apply(
    lambda r:
        r["sigs_s1"] & r["sigs_candidate"],
    axis=1
)

print("Current difficult misses:", len(current))

print(
    "At least 1 longest-word signature:",
    (
        current["common_longest_sigs"].str.len()
        >= 1
    ).sum()
)

print(
    "Both signatures match:",
    (
        current["common_longest_sigs"].str.len()
        >= 2
    ).sum()
)

print("\nExamples:")
print(
    current[
        [
            "s1_name",
            "candidate_name",
            "sigs_s1",
            "sigs_candidate",
            "common_longest_sigs"
        ]
    ].head(20).to_string(index=False)
)
```

```python
# ==========================================
# ADDRESS WORD-PAIR FREQUENCY
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE train_address_word_pairs AS

SELECT
    candidate_id,
    country,

    CASE
        WHEN w1 < w2
        THEN w1 || '|' || w2
        ELSE w2 || '|' || w1
    END AS word_pair

FROM (
    SELECT
        entity_id AS candidate_id,
        country,

        lower(w1) AS w1,
        lower(w2) AS w2

    FROM (
        SELECT
            entity_id,
            country,
            unnest(
                regexp_extract_all(
                    lower(coalesce(business_address, '')),
                    '[^\\W\\d_]{4,}'
                )
            ) AS w1,
            unnest(
                regexp_extract_all(
                    lower(coalesce(business_address, '')),
                    '[^\\W\\d_]{4,}'
                )
            ) AS w2

        FROM read_csv(
            'train_source2.tsv',
            delim='\t',
            header=true
        )

        UNION ALL

        SELECT
            entity_id,
            country,
            unnest(
                regexp_extract_all(
                    lower(coalesce(business_address, '')),
                    '[^\\W\\d_]{4,}'
                )
            ) AS w1,
            unnest(
                regexp_extract_all(
                    lower(coalesce(business_address, '')),
                    '[^\\W\\d_]{4,}'
                )
            ) AS w2

        FROM read_csv(
            'train_source3.tsv',
            delim='\t',
            header=true
        )
    )
)

WHERE w1 <> ''
  AND w2 <> ''
  AND w1 <> w2
""")

print("✅ Address word-pair table created")
```

```python
# ==========================================
# ADDRESS WORD-PAIR FREQUENCY
# ==========================================

con.execute("""
CREATE OR REPLACE TEMP TABLE train_address_pair_freq AS

SELECT
    country,
    word_pair,
    COUNT(DISTINCT candidate_id) AS freq
FROM train_address_word_pairs
GROUP BY
    country,
    word_pair
""")

print("✅ Address word-pair frequency table created")

# Build word-pairs for the current 55 missed true matches
from itertools import combinations

def make_word_pairs(words):
    words = sorted(set(words))
    return {
        a + "|" + b
        for a, b in combinations(words, 2)
        if a != b
    }

miss_pair_rows = []

for _, row in current55_addr.iterrows():

    s1_words = row["s1_addr_words"]
    cand_words = row["candidate_addr_words"]

    common = set(s1_words) & set(cand_words)

    for pair in make_word_pairs(common):

        miss_pair_rows.append({
            "source1_entity_id": row["source1_entity_id"],
            "candidate_id": row["candidate_id"],
            "word_pair": pair
        })

miss_pairs_df = pd.DataFrame(miss_pair_rows)

print(
    "Current 55 misses:",
    len(current55_addr)
)

print(
    "Common address word-pairs found:",
    len(miss_pairs_df)
)

# Look up frequency of these pairs
pair_freq_result = con.execute("""
SELECT
    m.source1_entity_id,
    m.candidate_id,
    m.word_pair,
    f.freq
FROM miss_pairs_df m
LEFT JOIN train_address_pair_freq f
    ON m.word_pair = f.word_pair
ORDER BY
    COALESCE(f.freq, 999999999)
""").fetchdf()

print("\nRare pair coverage:")

for limit in [10, 25, 50, 100, 250, 500]:

    covered = (
        pair_freq_result
        .groupby(
            ["source1_entity_id", "candidate_id"]
        )["freq"]
        .apply(
            lambda x:
                (x.fillna(999999999) <= limit).any()
        )
        .sum()
    )

    print(
        f"Pair frequency <= {limit}: "
        f"{covered} / {len(current55_addr)} misses"
    )
```

```python
# ==========================================
# LONGEST-WORD 4-GRAM TEST
# CURRENT 55 MISSES
# ==========================================

from unidecode import unidecode
import re
import pandas as pd

NAME_STOPWORDS = {
    "inc", "incorporated",
    "llc", "ltd", "limited",
    "private", "pvt",
    "plc", "corp", "corporation",
    "company", "co", "pc",
    "pllc", "llp", "lp",
    "group", "services", "service",
    "and", "of", "the"
}

# Get exactly the current 55 missed true pairs
current55 = miss_names.merge(
    con.execute("""
        SELECT
            source1_entity_id,
            candidate_id
        FROM current_final_misses
    """).fetchdf(),
    on=["source1_entity_id", "candidate_id"],
    how="inner"
).copy()


def longest_word_4grams(text):
    if pd.isna(text):
        return set()

    x = unidecode(str(text).lower())

    words = re.findall(r'[a-z0-9]+', x)

    words = [
        w for w in words
        if len(w) >= 4
        and w not in NAME_STOPWORDS
    ]

    if not words:
        return set()

    longest = max(words, key=len)

    return {
        longest[i:i+4]
        for i in range(len(longest) - 3)
    }


current55["s1_long4"] = current55["s1_name"].apply(
    longest_word_4grams
)

current55["candidate_long4"] = current55[
    "candidate_name"
].apply(
    longest_word_4grams
)

current55["shared_long4"] = current55.apply(
    lambda r:
        r["s1_long4"] & r["candidate_long4"],
    axis=1
)

print("Current misses:", len(current55))

for threshold in [1, 2, 3, 4, 5]:

    print(
        f"Shared 4-grams >= {threshold}:",
        (
            current55["shared_long4"].str.len()
            >= threshold
        ).sum()
    )
```

```python
# ==========================================
# SAME-WORD 4-GRAM OVERLAP
# CURRENT 55 MISSES
# ==========================================

def longest_meaningful_word(text):
    if pd.isna(text):
        return ""

    x = unidecode(str(text).lower())

    words = re.findall(r'[a-z0-9]+', x)

    words = [
        w for w in words
        if len(w) >= 4
        and w not in NAME_STOPWORDS
    ]

    return max(words, key=len) if words else ""


def word_4grams(word):
    if len(word) < 4:
        return set()

    return {
        word[i:i+4]
        for i in range(len(word) - 3)
    }


current55["s1_longest_word"] = current55["s1_name"].apply(
    longest_meaningful_word
)

current55["candidate_longest_word"] = current55[
    "candidate_name"
].apply(
    longest_meaningful_word
)

current55["s1_longest_4g"] = current55[
    "s1_longest_word"
].apply(word_4grams)

current55["candidate_longest_4g"] = current55[
    "candidate_longest_word"
].apply(word_4grams)

current55["same_word_4g"] = current55.apply(
    lambda r:
        r["s1_longest_4g"] &
        r["candidate_longest_4g"],
    axis=1
)

print("Current misses:", len(current55))

for n in [1, 2, 3, 4]:

    print(
        f"Same longest word, shared 4-grams >= {n}:",
        (
            current55["same_word_4g"].str.len()
            >= n
        ).sum()
    )
```

```python
# ==========================================
# DUCKDB FULL-TEXT SEARCH INDEX
# ==========================================

con.execute("INSTALL fts;")
con.execute("LOAD fts;")

con.execute("""
CREATE OR REPLACE TABLE train_fts AS

SELECT
    entity_id AS candidate_id,
    country,
    business_name,
    business_address,

    -- compact form helps website-like names
    lower(
        regexp_replace(
            coalesce(business_name, ''),
            '[^\\p{L}\\p{N}]+',
            '',
            'g'
        )
    ) AS compact_name

FROM read_csv(
    'train_source2.tsv',
    delim='\t',
    header=true
)

UNION ALL

SELECT
    entity_id AS candidate_id,
    country,
    business_name,
    business_address,

    lower(
        regexp_replace(
            coalesce(business_name, ''),
            '[^\\p{L}\\p{N}]+',
            '',
            'g'
        )
    ) AS compact_name

FROM read_csv(
    'train_source3.tsv',
    delim='\t',
    header=true
)
""")

print("✅ train_fts table created")

con.execute("""
PRAGMA create_fts_index(
    'train_fts',
    'candidate_id',
    'business_name',
    'business_address',
    'compact_name',
    overwrite = 1
)
""")

print("✅ FTS index created")
```

```python
# ============================================================
# FAST LIGHTWEIGHT CANDIDATE GENERATION
# NO FTS / NO 10M-ROW INDEX
# ============================================================

import pandas as pd
import re
import time

start = time.time()

# ------------------------------------------------------------
# 1. Load only the columns we need from S2 and S3
# ------------------------------------------------------------

s2 = pd.read_csv(
    "train_source2.tsv",
    sep="\t",
    usecols=["entity_id", "business_name", "business_address", "country"]
)

print("S2 loaded:", len(s2))

s3 = pd.read_csv(
    "train_source3.tsv",
    sep="\t",
    usecols=["entity_id", "business_name", "business_address", "country"]
)

print("S3 loaded:", len(s3))

# Combine
candidates = pd.concat([s2, s3], ignore_index=True)

print("Total S2 + S3:", len(candidates))


# ------------------------------------------------------------
# 2. Normalize business names
# ------------------------------------------------------------

def normalize_name(x):
    if pd.isna(x):
        return ""
    
    x = str(x).lower()
    x = re.sub(r"[^a-z0-9\s]", " ", x)
    x = re.sub(r"\s+", " ", x).strip()
    
    return x


candidates["norm_name"] = candidates["business_name"].map(normalize_name)


# ------------------------------------------------------------
# 3. Create token index
# ------------------------------------------------------------

token_index = {}

for idx, name in enumerate(candidates["norm_name"]):

    tokens = set(name.split())

    for token in tokens:

        # Ignore extremely short/common tokens
        if len(token) < 4:
            continue

        if token not in token_index:
            token_index[token] = []

        token_index[token].append(idx)


print("Token index created:", len(token_index))


# ------------------------------------------------------------
# 4. Generate candidates for 100 validation S1
# ------------------------------------------------------------

all_candidates = []

for _, row in test_s1.iterrows():

    s1_id = row["source1_entity_id"]
    name = normalize_name(row["business_name"])

    tokens = set(name.split())

    possible = set()

    # Use up to 3 useful tokens
    useful_tokens = [
        t for t in tokens
        if len(t) >= 4 and t in token_index
    ]

    # Prefer rarer tokens
    useful_tokens.sort(
        key=lambda x: len(token_index[x])
    )

    for token in useful_tokens[:3]:

        # Limit huge blocks
        if len(token_index[token]) <= 50000:
            possible.update(token_index[token])

    for idx in possible:
        all_candidates.append(
            (s1_id, candidates.iloc[idx]["entity_id"])
        )


candidate_pairs_fast = pd.DataFrame(
    all_candidates,
    columns=["source1_entity_id", "candidate_entity_id"]
).drop_duplicates()


# ------------------------------------------------------------
# 5. Calculate validation recall
# ------------------------------------------------------------

gt_lookup = dict(
    zip(
        test_s1["source1_entity_id"],
        test_s1["matched_entity_ids"]
    )
)

total_true = 0
captured_true = 0

for s1_id, gt_string in gt_lookup.items():

    if pd.isna(gt_string) or str(gt_string).strip() == "":
        continue

    true_ids = set(str(gt_string).split(","))

    retrieved = set(
        candidate_pairs_fast.loc[
            candidate_pairs_fast["source1_entity_id"] == s1_id,
            "candidate_entity_id"
        ]
    )

    total_true += len(true_ids)
    captured_true += len(true_ids & retrieved)


recall = captured_true / total_true

print("\n================ RESULT ================")
print("True matches      :", total_true)
print("Captured          :", captured_true)
print("Recall            :", round(recall * 100, 2), "%")
print("Candidates        :", len(candidate_pairs_fast))
print("Avg candidates/S1 :", round(len(candidate_pairs_fast) / 100, 2))
print("Time              :", round(time.time() - start, 1), "sec")
```
